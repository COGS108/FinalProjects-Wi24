```python
# imports for notebook
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
from scipy.optimize import curve_fit
import patsy
import statsmodels.api as sm
import warnings
import jinja2

# silence future warnings (primarily in EDA section)
warnings.filterwarnings("ignore", category=FutureWarning)

# imports needed for data scraping
import re
import requests
from bs4 import BeautifulSoup
```

```python
# Use requests to retrieve the data from spotrac and use beautifulsoup to parse the outcome
url = 'https://www.spotrac.com/nfl/contracts/sort-value/limit-2000/'
response = requests.get(url)
soup = BeautifulSoup(response.text, 'html.parser')
rows = soup.find_all('tr')
ans = []
for row in rows:
    cells = row.find_all('td')
    data = [cell.text.strip() for cell in cells]
    ans.append(data)
```

We can check the result from scraping the data from the Spotrac website, and see that we need to clean the data quite a bit.

```python
ans[:4]
```

We see that `ans` is a list, where each element is another list with the information we collected from each row on Spotrac. We can clean the data by removing all the newline characters and extra unnecessary symbols.

```python
def fix_formatting(text):
    text = re.sub(r'\([^)]*\)', '', text)
    return text.replace('|', '')
```

```python
# We get raw data that requires further cleaning.
# Printing the data shows that we have some missing values and bunches of \ns and spaces, we work to get rid of all of them.
ans = ans[1:]
new = [[s.replace('\n', '') for s in i] for i in ans]
temp1 = [[fix_formatting(string) for string in sublist] for sublist in new]
temp2 = [[w for s in l for w in s.split('  ')] for l in temp1]
temp3 = [[e for e in s if e != ''] for s in temp2]
temp4 = [[e for e in s if 'Original' not in e] for s in temp3]
temp4[0]
```

Within each element of `ans`, we can see that the second member of that element is the player name, and the fourth member is the contract year period. We can fix the player name formatting by using a regular expression to catch the name and overwriting `element[1]`, and saving the contract start year to `element[3]`.

```python
# temp4 is list of rows from table
# element is list of elements from each row
for element in temp4:
    #splitted will find first and last names, and each suffix individually (e.g. 'John Smith-Doe II' -> 'John', 'Smith-Doe', 'I', 'I')
    splitted = re.findall('[A-Z]-?[a-z]*', element[1])

    # Dividing by 2 removes extra last name parts from beginning of splitted
    splitted = splitted[len(splitted)//2:]
    # If splitted is more than 2 parts, there is compound last name or suffix
    if len(splitted) > 2:
        #If the first part of the last name is less than 3 characters long, more than likely it is a prefix (e.g. "Mc")
        if len(splitted[1]) < 3:
            splitted[1] = ''.join(splitted[1:])
        # Otherwise, it's a suffix; keep it spaced to remove it later
        else:
            splitted[1] = ' '.join(splitted[1:])

    element[1] = ' '.join(splitted[:2]).strip()
    if "I I" in element[1] or "I I I" in element[1]:
        element[1] = element[1].replace("I I I",'');
        element[1] = element[1].replace("I I",'');
    

    element[3] = int(element[3].split('-')[0])

temp4[0]
```

Now we can build our Spotrac DataFrame using our cleaned data:

```python
spotrac_df = pd.DataFrame(temp4, columns = ['rank','player','position','start_year','sign_age','contract_yrs','contract_value','average_annual_value','sign_bonus',"guaranteed_at_sign","actual_guaranteed"])
spotrac_df.head()
```

Next, we want to convert the player's position into the appropriate abbreviation:

```python
def standardize_position(pos):
    pos = pos.lower()
    
    # Dictionary mapping positions to abbreviations
    position_abbreviations = {
        'quarterback': 'QB',
        'cornerback': 'CB',
        'running back': 'RB',
        'halfback': 'HB',
        'tailback': 'TB',
        'fullback': 'FB',
        'inside linebacker': 'ILB',
        'outside linebacker': 'OLB',
        'linebacker': 'LB',
        'defensive end': 'DE',
        'defensive tackle': 'DT',
        'center': 'C',
        'strong safety': 'SS',
        'free safety': 'FS',
        'safety': 'S',
        'wide receiver': 'WR',
        'tight end': 'TE',
        'left tackle': 'LT',
        'right tackle': 'RT',
        'tackle': 'T',
        'guard': 'G',
        'edge': 'E'
    }
    
    # Replace positions with abbreviations
    pos = position_abbreviations.get(pos, '?')

    return pos
```

```python
spotrac_df['position'] = spotrac_df['position'].apply(standardize_position)
spotrac_df = spotrac_df[spotrac_df['position'] != '?']
```

Next, we want to limit our dataset to entries between 2020 and 2023, to align with the scope we defined for our project earlier. We then convert the variables to the right of `start_year` to integers and floats where appropriate by removing dollar signs and commas. Given that our question caters to analysis of contract values and sign-on bonuses, we chose to drop the variables that were irrelevant to our study, and add column `base_contract`, which are the values of `contract_value` subtracted by `sign_bonus`.

```python
# Spotrac Cleaning
spotrac_df = spotrac_df[(spotrac_df['start_year'] >= 2019) & (spotrac_df['start_year'] <= 2023)]
strs = ['sign_age','contract_yrs']
strs2 = ['contract_value','average_annual_value','sign_bonus',"guaranteed_at_sign","actual_guaranteed"]
for col in strs:
    spotrac_df[col] = pd.to_numeric(spotrac_df[col], errors='coerce').fillna(0).astype(int) 
for col in strs2:
    spotrac_df[col] = spotrac_df[col].str.replace(r'\$', '', regex=True).str.replace(r',', '', regex=True) # Get rid of extra signs
    spotrac_df[col] = pd.to_numeric(spotrac_df[col], errors='coerce').fillna(0).astype(float) #typecast

spotrac_df = spotrac_df.drop(columns=['rank','contract_yrs', 'sign_age', 'average_annual_value','guaranteed_at_sign', 'actual_guaranteed'], axis=1)
spotrac_df['base_contract'] = spotrac_df['contract_value'] - spotrac_df['sign_bonus']
```

Finally, we want to add the $log_{10}$ value of `contract_value`, `sign_bonus`, and `base_contract`, which will help address range concerns later when analyzing our data.

```python
spotrac_df['contract_value_log10'] = np.log10(spotrac_df['contract_value'] + 1)
spotrac_df['sign_bonus_log10'] = np.log10(spotrac_df['sign_bonus'] + 1)
spotrac_df['base_contract_log10'] = np.log10(spotrac_df['base_contract'])
```

## Linear Regression using OLS Model

Here we will perform Linear Regression in order to measure the relationship between the sign-on bonus as a predictor and the different peformance statistics as outcomes. Each outcome will be compared in separate linear regression models. 

### Defensive Statistic Analysis

For our analysis, we will perform linear regressions for `SACK`,`TOT`,`INT`, and `FF` to determine if there is a relationship between them and `sign_bonus_log10`. We will interpret these results later.

```python
# linear regressions for performance metrics
outcome_sack, predictor_sack = patsy.dmatrices('SACK ~ sign_bonus_log10', df_diff)
res_sack = sm.OLS(outcome_sack, predictor_sack).fit()

outcome_tot, predictor_tot = patsy.dmatrices('TOT ~ sign_bonus_log10', df_diff)
res_tot = sm.OLS(outcome_tot, predictor_tot).fit()

outcome_int, predictor_int = patsy.dmatrices('INT ~ sign_bonus_log10', df_diff)
res_int = sm.OLS(outcome_int, predictor_int).fit()

outcome_FF, predictor_FF = patsy.dmatrices('FF ~ sign_bonus_log10', df_diff)
res_FF = sm.OLS(outcome_FF, predictor_FF).fit()
```

We developed a function to extract the intercepts, coefficients, standard error, and p-values from the linear regressions and later on we will combine them into a table to make it easy to visualize the results. 

```python
def linear_table(linear_model, col_name):
    out_table = linear_model.summary().tables[1]
    out_df = pd.DataFrame(out_table.data, columns=out_table[0])
    out_df = out_df.iloc[1:, 0:5]
    out_df = out_df.drop(out_df.columns[3], axis=1)
    out_df.iloc[0,0] = col_name
    out_df.columns = ['intercept', 'coef', 'std err', 'P>|t|']
    return out_df
```

```python
Defense_TOT_intercepts = linear_table(res_tot, "TOT Intercept")
Defense_SACK_intercepts = linear_table(res_sack, "SACK Intercept")
Defense_FF_intercepts = linear_table(res_FF, "FF Intercept")
Defense_INT_intercepts = linear_table(res_int, "INT Intercept")
```

Now that we have our linear regressions as a dataframe that can be more easily manipulated and interpreted. 

```python
defense_intercepts = pd.concat([Defense_TOT_intercepts, Defense_SACK_intercepts, Defense_FF_intercepts, Defense_INT_intercepts], ignore_index=True)
idx = pd.IndexSlice
slice_ = idx[::2, :]  # Slice to select every other row

# Apply styling to the subset defined by the slice
styled_defense_intercepts = defense_intercepts.style.set_properties(**{'background-color': '#6b8eae'}, subset=slice_)

# Display the styled DataFrame
styled_defense_intercepts 
```

In our analysis of defensive statistics including Sacks (SACK), and Interceptions (INT), we observed that the p-values associated with these metrics were greater than 0.05. This suggests that there is no statistically significant influence of sign-on bonuses on player performance in these categories.

Looking at all linear regressions together, the change in player performance for some statistics were indeed negatively correlated with sign-on bonus when the confounding variable base contract is held constant. However, the effect size was much smaller than anticipated, so the changes in performance would not be noticeable in the real world application of the NFL drafting and contract process.  

When comparing the relationship between sign-on bonus and offensive performance statistics, our results show that sign-on bonus cannot strongly predict change in offensive performance for players between their first contract year and the previous year. Although the sign-on bonus was supposedly a significant predictor for receiving touchdowns, and even more significant when contract value is kept constant, the effect size is very small. There is less than one unit of change in performance for a 10 times increase in sign-on bonus. Therefore, these changes in performance are miniscule and not noticeable in the big picture for football player performance.

For defensive statistics, the null hypothesis is not rejected, indicating that the sign-on bonus does not directly impact players' performance This suggests that there is no statistically significant relationship between the sign-on bonus and the performance of football defensive players. However, when the confounding variable base contract value is included, both the sign-on bonus and base contract value become significant predictors of performance in FF and INT. Similar to offensive statistics, increasing sign-on bonus is correlated with a slightly negative decrease in performance, that is not very meaningful compared to the large increase in sign-on bonus.

Our project faced limitations due to a small sample size of NFL player statistics from 2020-2023, which may not fully represent broader NFL trends since 1920. Additionally, about a third of the raw cleaned data was discarded due to incompleteness (previous contract year missing, players not having all/some of the chosen performance metrics).

Touching back on previous work, our findings differed from our cited reports, since we sought to find a relationship on the individual player level rather than their team as a whole. Although we did find statistical significance between sign-on bonus and some player performance statistics however it wasn't of significance. Based on our findings, we conclude that our initial hypothesis of predicting players may get lazy with their performance was false; our results show that players are staying steady with their performance even after the luxury of a sign-on bonus. Further research could be conducted to observe whether differing results can be found if the time period was further extended beyond the first season after signing, and more performance metrics were considered at once.
