# Abstract

Our group was curious to explore the relationship between temperature and a variety of social factors. Given the rising occurrence of heat waves around the world, we decided it would be important to ask the question of how these heat waves affect human behavior. In particular, we believed that heat waves might prompt people to be less rational and more aggressive, so we decided to explore how heat waves affect violent crime rates in cities. In addition, we were curious if any change in behavior due to heat was stronger in cities where people are not as used to high heat.

In order to accomplish these tasks, we used crime and temperature datasets from three cities in the US: Los Angeles, Phoenix, and Chicago. We then took note of spikes in temperature and observed how crime rates changed over the following days.

Through our analysis, we found that heat waves did not significantly or consistently affect violent crime rates. However, we did notice that changes in crime during heat waves were more common in the colder climate of Chicago, although these changes did not happen consistently. We also observed that crime rates fluctuate over the seasons, with generally more crime occuring in the summer months, but had little evidence to support the claim that temporary heat waves have an immediate effect on violent crime rates. 

# Research Question


**How do heat waves affect violent crime rates within a given area and how is this variance influenced by regions accustomed to heat versus those that are unaccustomed?**



## Background and Prior Work

A publication from Boston University mentions that mental health is significantly impacted during heat waves, as the number of hospitalizations for mental health conditions increases during these times. Additionally, it was found that standardized testing scores dropped during heat waves (Thornton 2021). This research points to a negative effect on people’s mental and emotional states, which we believe can lead people to more spontaneous and aggressive behavior.

An article in the Journal of Urban Health addressed a related subject matter. Through a time series analysis conducted in Philadelphia, Pennsylvania, researchers discovered a positive, linear correlation between variations in the daily mean heat index from the seasonal average and the incidence rates of violent crime and disorderly conduct (Schinasi and Hamra, 2017).

Our project aims to analyze the relationship between unexpected heatwaves and violent crimes, as well as how this relationship differs in regions accustomed to heat versus those that are unaccustomed. Already, there is lots of prior research and existing literature surrounding how crime rates tend to be higher in places that are hotter. Specifically, our research differs in that we want to see if unexpected heat causes people to act more aggressively and if people living in hot regions react differently to heat waves from people living in a cold region. More broadly, our question focuses on trying to see if uncomfortable and unaccustomed changes in weather will have an effect on people's moods. The question we are asking primarily looks at people's mood, we are simply choosing to measure violent crime as a means of assessment. This is important, because that means when collecting data, we have to make sure that the crimes observed only include those that accurately reflect mood as the motivating factor. Therefore, crimes motivated by need such as theft or robbery are not included, whereas crimes such as domestic abuse or public disturbance are.

What we would hope to see is that unexpected weather, particularly heat, has some relation to peoples' moods (irritability, aggressiveness, etc.) in such a way that people behave differently, reflected in the particular crimes we are measuring. From previous works, we can hypothesize that this relation is likely a positive one.

# Hypothesis


We predict that unexpected heat waves do increase violent crime levels. # Results

### Los Angeles Crime/Weather EDA

```python
plt.rcParams["figure.figsize"] = [12, 3.50]

# Loading DataFrames in
la_crime = pd.read_csv('data/LA_CRIME_2013_2023.csv')
la_crime['DATE OCC'] = pd.to_datetime(la_crime['DATE OCC'], format='%Y-%m-%d', errors='coerce')

la_weather = pd.read_csv('data/LA_WEATHER_2013_2023.csv')
la_weather['DATE'] = pd.to_datetime(la_weather['DATE'], format='%Y-%m-%d', errors='coerce')

# ------------- 2013-2023 -------------
grouped_crime_full = la_crime.groupby('DATE OCC').size().to_frame("count").reset_index()
crime_full = grouped_crime_full.plot(x='DATE OCC', y='count', label = 'Crime Count (left)', title = 'Los Angeles: Crime and Weather (2013-2023)')
weather_full = la_weather.plot(color = 'red', x = 'DATE', y = 'TAVG', secondary_y=True, ax = crime_full, label = 'Temp (°F)')

# Place legend outside of axes
crime_full.get_legend().set_bbox_to_anchor((1, 1.2))

# ------------- 2023 -------------
la_crime_2023 = la_crime[la_crime['DATE OCC'].dt.year == 2023]
grouped_crime_2023 = la_crime_2023.groupby('DATE OCC').size().to_frame("count").reset_index()
crime_2023 = grouped_crime_2023.plot(x='DATE OCC', y='count', label = 'Crime Count (left)', title = 'Los Angeles: Crime and Weather (2023)')

la_weather_2023 = la_weather[la_weather['DATE'].dt.year == 2023]
weather_2023 = la_weather_2023.plot(color = 'red', x = 'DATE', y = 'TAVG', secondary_y=True, ax = crime_2023, label = 'Temp (°F)')

# Place legend outside of axes
crime_2023.get_legend().set_bbox_to_anchor((1, 1.2))
```

Both of the above graphs show line plots of the average temperature (red) and the number of violent crime occurrences (blue) in the city of Los Angeles. The top graph displays the trends over the course of 10 years, whereas the bottom graph looks at the trend over one year (2023). While the 10 year graph displays that the temperature is distributed in an oscillating pattern (which is to be expected), we cannot say much of the same for the crime counts. The bottom graph gives us more insight into answering our question, as we can see heat waves more clearly due to the shorter timespan. While there are observable temperature spikes throughout the year, there does not seem to be a consistent increase in crime occurrences during or after these heat waves. There is possibly a mild correlation between the change in seasons and the violent crime rate, as seen by a slightly higher average crime rate in the summer months, but there is not much evidence here pointing us towards the belief that heat waves cause a temporary increase in violent crime rates.

As we did with the LA data, we have constructed line plots on a ten-year and one-year interval for Phoenix. The ten-year plot reveals a more striking oscillation over the seasons for temperature, yet the crime data does not vary tremendously. This may be explained by the fact that Phoenix stays at a warm or hot temperature throughout the year, and so people may be more used to heat and thus it may not play a large role in increased crime rates. At the one-year level, we can see some heat waves over the year, but just like LA, there is little to no indication that crime increases during or right after these heat waves.

# Discussion and Conclusion

Our exploration into the relationship between heat waves and violent crime rates in cities yielded unexpected results. Despite our initial hypothesis that there would be a potential link between violent crime and heat waves, our analyses did not reveal a clear and consistent pattern that would suggest so in Los Angeles, Chicago, or Phoenix.

Across all three cities, we cleaned our data so that we could visualize temperature and crime as a line graph overlaid on top of one another. For each city we had graphs spanning the entire decade of available weather data, but chose to break it down into graphs spanning smaller time periods in order to more easily try and locate heatwaves and the plot of crime shortly thereafter. For all three cities, depending on their crime cataloging standards, we would drop labels for crimes that most likely had confounds beyond just behavior.

In Los Angeles and Phoenix, where residents are more accustomed to warmer temperatures, we observed minimal fluctuations in violent crime rates during heat waves. This finding suggests that factors beyond temperature alone may influence crime rates, including socioeconomic conditions, law enforcement strategies, and community resilience.

Chicago, however, presented a slightly different picture. While our analysis showed a general increase in violent crime during the warmer summer months, the impact of heat waves on crime rates was less pronounced. This could be attributed to the city's colder climate, where warmer temperatures contribute to factors, such as being able to go outside for longer, increasing the crime rate. But, there was no notable increase in crime outside of the general trend that warmer climate is correlated to higher crime rates, even here there was no data that suggested a heatwave caused significant increases to violent crime during or in the days immediately following.

Our study is not without limitations. The datasets used were limited to publicly available records, which may not fully capture all instances of violent crime. Additionally, our analysis focused solely on temperature as a proxy for heat waves, overlooking other potential environmental factors such as humidity and air quality, and violent crime was a proxy for aggressive or violent behavior. Furthermore, our study was constrained to three cities in the United States, limiting the generalizability of our findings to other regions and contexts.

Overall, our findings suggest that the relationship between heat waves and violent crime rates is complex and multifaceted. While temperature may play a role in shaping behavioral responses, other contextual factors are likely at play. For now, we cannot say that there are immediate and consistent crimes resulting for heat waves due to an effect on behavior.
