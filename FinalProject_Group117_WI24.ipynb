# Abstract


Our project aims to predict arrests and police stops of certain demographics in LA County. We decided to look into this topic because of the history of racial profiling and police abuse by LAPD throughout history. LAPD claims they are fixing their problem of racial profiling, so we wanted to analyze recent arrest data and racial population data in LA County in order to determine if racial profiling in LAPD policing has actually been decreasing or if it has stayed constant. We hypothesized that LAPD has continued or increased racial profiling in their policing.

We gathered data for our investigation from arrest data recorded by LAPD between 2010 and 2019 as well as US Census data in LA county from 2010 and 2020. Using the data collcted, we were able to visualize number of arrests made based on areas of LA County, number of arrests from 2010-2019, LA racial population, and number of arrests compared to ethnicity population.

From our analysis of the data collected, we found that the total number of arrests in LA county have decreased between 2010 and 2019. However, in looking at the populations of different races in LA County in comparison to the number of arrests of different racial groups, we found that Black people are arrested significantly more in comparison to their population. We also found that there has been no significand changes in the populations of racial identities in LA County. 

Looking at the data collected and from our analysis, we can conclude that although total number of arrests are decreasing in LA County, there is still a significant disparity in the number of Black LA County residents arrested compared to their population. This suggests that there is still significant racial profiling happening within LAPD policing.

## Background and Prior Work

 There is a long history of police abusing people of color in Los Angeles <a name="FELKER-KANTOR"></a>[<sup>1</sup>](#cite_note-1). The LAPD did not respond to calls for reform in the wake of the Watts uprising, and police brutality continued up to and beyond the LA rebellion of 1992, which was a response to the police beating Rodney King <a name="FELKER-KANTOR"></a>[<sup>1</sup>](#cite_note-1). While there were some reforms after the 92 rebellion, the LAPD continued their project of racist violence and community control <a name="FELKER-KANTOR"></a>[<sup>1</sup>](#cite_note-1). Specifically 1996, the DOJ filed a lawsuit against the LAPD <a name="FELKER-KANTOR"></a>[<sup>1</sup>](#cite_note-1).  To avoid going to trial the city council agreed to a consent decree that enabled federal oversight of LAPD <a name="FELKER-KANTOR"></a>[<sup>1</sup>](#cite_note-1). The consent decree also mandated strict regulations, oversight by federal circuit courts,  and annual audits of how the reforms had been progressing <a name="FELKER-KANTOR"></a>[<sup>1</sup>](#cite_note-1). The goal of our project is to identify whether the LAPD continues to engage in racist practices after the reforms mandated by the DOJ. In particular, LAPD began to utilize big data in policing in 2001 <a name="FELKER-KANTOR"></a>[<sup>1</sup>](#cite_note-1). Felker-Kantor argues that the statistical focus on the number of arrests to determine who had potential for promotion encouraged police officers to wrongfully arrest or stop and search people without suspicion wrongfully. We seek to understand whether there has been a change in the problem of racial violence in policing.

```python
def convert_float(str_in):
    str_in = str_in.replace('%', '')
    str_in = str_in.replace(',', '')
    output = int(str_in)         
    return output
```

## Arrest Data from 2010 to 2019

```python
arrest_df = pd.read_csv('https://drive.usercontent.google.com/download?id=1GQN967mTfWD0K7UhhjBYJlyUBURHj4e3&confirm=xxx')
```

```python
#remove unnecesary columns
arrest_df = arrest_df.drop(['Cross Street', 'Booking Date','Booking Time', 'Booking Location', 'Booking Location Code', 'Disposition Description'], axis=1)
arrest_df = arrest_df.dropna(subset=['Area ID', 'Area Name', 'Reporting District', 
                                     'Age', 'Sex Code', 'Descent Code', 'LAT', 'LON', 'Location'])

arrest_df.drop(arrest_df[arrest_df.LON > -100].index, inplace=True)


#extract year information from date data
arrest_df.insert(3,"Year", pd.to_datetime(arrest_df['Arrest Date']).dt.year)

#fix descent codes
descent_codes_fixed = {
    'A': 'Asian', 
    'B': 'Black', 
    'C': 'Asian', 
    'D': 'Asian',
    'F': 'Asian', 
    'G': 'Pacific Islander',
    'H': 'Hispanic/Latin/Mexican',
    'I': 'American Indian/Alaskan Native',
    'J': 'Asian',
    'K': 'Asian',
    'L': 'Asian',
    'O': 'Other',
    'P': 'Pacific Islander',
    'S': 'Pacific Islander',
    'U': 'Pacific Islander',
    'V': 'Asian',
    'W': 'White',
    'X': 'Unknown',
    'Z': 'Asian'
}

arrest_df['Descent Code'] = arrest_df['Descent Code'].replace(descent_codes_fixed)
```

## US Census LA Race 2010

```python
race_2010_df = pd.read_csv('Data_sets/US_Census_LA_Race_2010.csv')
```

```python
race_2010_df.set_index('Label', inplace=True)

#Simplified Rows and Columns Name
race_2010_df.columns = ['Agoura Hills-Malibu', 'Compton', 'Downey-Norwalk', 
                        'East San Gabriel Valley', 'Inglewood', 'Long Beach-Lakewood', 'Los Angeles', 
                        'Newhall', 'North Antelope Valley', 'Palos Verdes', 'Pasadena', 
                        'San Fernando Valley', 'Santa Monica', 'South Antelope Valley', 'South Bay Cities',
                        'South Gate-East Los Angeles', 'Southwest San Gabriel Valley', 'Torrance',
                        'Upper San Gabriel Valley', 'Whittier']

race_2010_df.rename({ "Hispanic or Latino" : "Hispanic/Latin/Mexican", 
                   "White alone": "White",
                   "Black or African American alone" : "Black", 
                   "American Indian and Alaska Native alone" : "American Indian/Alaskan Native",
                   "Asian alone" : "Asian",
                   "Native Hawaiian and Other Pacific Islander alone" : "Pacific Islander",
                   "Some Other Race alone" : "Other"
                  }, inplace=True)

#Compute sum per racial identities/Ethnicities
race_2010_df = race_2010_df.applymap(convert_float)
race_2010_df['Total'] = race_2010_df.sum(axis=1)
```

### Arrest Over Area in LA County
This barplot helps visualize the number of arrests across different neighborhoods in Los Angeles County and notice any geographical trends. It represents the distribution of arrests over a specified timeline, which allows us to identify any significant patterns or outliers. We wanted to emphasize the frequency of law enforcement and help see which areas may have higher criminal activity rates. 

```python
#generating a pivot table with totals for neighborhood and race

arrest_race_pivot = pd.pivot_table(arrest_df, values = 'Report ID', index = 'Descent Code', aggfunc = 'count')

location_pivot = pd.pivot_table(arrest_df, values = 'Report ID', index = 'Area Name', aggfunc = 'count')

year_pivot = pd.pivot_table(arrest_df, values = 'Report ID', index = 'Year', aggfunc = 'count')
```

```python
#Generate a plot of arrests across locations 
plot = location_pivot.plot(kind = 'bar', rot = 45, figsize = (10,6), align = 'center')
plot.set_title('Central L.A Has the Highest Total Number of Arrests')
# Label the Axes 
plot.set_ylabel('Number of Arrests')
plot.set_xlabel('Area Name')

plot.legend(["Number of Arrests"])
```

This bar plot shows the number of arrests throughout the different neighborhoods in Los Angeles County from 2010 - 2019. After cleaning and extracting the data from the dataset, we can see that the most arrests happened in Central LA, with over 120,000 arrests. On the other hand, we noticed that most neighborhood's arrests was a little under 80,000 while some outliers passed 100,000 arrests such as Hollywood and Pacific. Since we were comparing the relationship between the categoric variable and the numeric amount, we decided to utilize a bar plot to visualize our data. 

```python
#Obtain geospatial data for mapping
arrest_gdf = gpd.read_file("Data_sets/LAPD Divisions.zip")
arrest_gdf = arrest_gdf.set_index("APREC")

tmp = location_pivot
tmp.index = location_pivot.index.str.upper()
arrest_gdf  = arrest_gdf.merge(tmp, on=arrest_gdf.index, how='left')
```

### Arrest Percentage of Each Racial Group

Plotting the arrest percentage of each racial group creates a comprehensive overview of the arrests that happen among different racial groups. This helps us observe any potential disparities or biases from law enforcement practices and how these activities may have disproportionately impacted racial groups. We can address which groups that have experienced systematic inequalities and see if there is a correlation between the demographics of people with police arrests.  

```python
perc_df = race_2020_df['Total']
perc_df = perc_df.to_frame()
s = arrest_race_pivot
perc_df = pd.concat([perc_df, s], axis=1).dropna()

perc_df.columns = ['Population %', 'Arrest %']

perc_df['Population %'] = (perc_df['Population %'] / perc_df['Population %'].sum()) * 100
perc_df['Arrest %'] = (perc_df['Arrest %'] / perc_df['Arrest %'].sum()) * 100
```

```python
perc_df
```

```python
perc_df.plot(kind='barh')
plt.title("Percentage of Arrest of different Racial Identities/Ethnicities vs. Population in LA County")
plt.xlabel("Percentages")
plt.grid()
```

This graph shows the relationship of the number of arrests amongst the racial identities/ethnicities within Los Angeles County in percentages. We noticed that within the Black community, there are more arrests, with about 28.75% that happened than the overall population of about 7.8%. Another observation was that there were no significant changes in the population and arrest percentages for the American Indians/Alaskan Native community. As for the Hispanic/Latin/Mexican community, both variables are almost the same. They make up about 49.53% of the population and have about 46.62% of the amount of arrests within the county.
