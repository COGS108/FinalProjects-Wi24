### Dataset #1: LinkedIn Job Postings - 2023
LinkedIn job Postings - 2023 is a sample of every job posted to LinkedIn during two random days in 2023, and includes information such as the title, description, location and pay for each job.  We believe this is a representative sample of all jobs across the United States.  We intend to get Job ID, Job title, Salaries, Programming Language, and Location columns from LinkedIn Job Posting -2023:

  **Job ID**: We keep job id just to make every row unique and make it clear that they are different job postings.\
  **Job title**: Used to determine if a job is data science or software engineering by performing text analysis on the description looking for the terms “data science”, "machine learning", "developer" or “software engineering”, or similar.\
  **Minimum salary**:  We chose minimum salary because it was more abundant than median salary, and more meaningful than maximum salary. We also need to clean the salary column to make them consistent in the same pay period (monthly/yearly) by using the 'pay_period' column\
  **Programming language**:  From the "job description" column, perform text analysis and identify if (Python but not Java), (Java but not Python), or (both Python and Java) are mentioned.  Discard all jobs that don’t mention either\
  **Level**:  Discard all jobs that are not mid-senior level using "Formatted_Experience_Level" Column\
  **Posting location**:  Used to normalize salary by cost-of-living index, which will vary by location.

### Dataset #2: 2024 GS Locality Pay Table
We recognize it is important to compare salaries after normalizing for their buying power.  We used the US federal government's locality pay adjustments as a proxy for cost of living.  In efforts to maintain fair and transparent salaries, the Office of Personnel Management (OPM) has broken the country into 58 distinct zones (locality pay areas), each with an associated cost of living that gets added to a federal employee's base pay.  From the dataset, we extracted:

  **LOCNAME**: the 2-3 letter code of the locality pay area.  There are 58 locality pay areas.\
  **GRADE**: government pay works on a grade and step system.  We chose to use the locality pay from Grade 1.\
  **ANNUAL1**: the locality pay in US dollars added to a federal employee's annual salary based on their locality pay area, assuming they were Step 1.

### Dataset #3: Locality Pay Area definitions
Dataset #2 gave us some measure of cost of living in each of the locality pay areas are, but we have no idea where those locality pay areas are in relation to the jobs in Dataset #1.  Therefore, it was necessary to look at the locality pay areas as they are definited by OPM, and manually determine their centroids.  These data were compiled in locality_latlon.xlsx.  We will class jobs from Dataset #1 based on the Posting Location's distance from various locality pay areas' centroids.  From this spreadsheet, we are interested in

  **locname**: 2-3 letter codes unique to each locality pay area and consistent with the LOCNAME from Dataset #2.\
  **centroid_lat**: the latitude of the centroid of that locality pay area in decimal degrees.\
  **centroid_lon**: the longitude of the centroid of that locality pay area in decimal degrees, with 0 being the Prime Meridian and negative longitudes denoting degrees West of the prime meridian.

## 2024 GS Locality Pay Table

We start by loading the spreadsheet, removing all data from the pay steps we're not interested in, renaming the columns to better match coding conventions, and filtering the dataset so we have one value for each locality pay area corresponding to the annual locality pay of a grade 1, step 1 federal employee in that area.

## Locality Pay Area definitions

We start by loading the spreadsheet, and omitting the pay area description, which will not be helpful in further analysis.

```python
locality_pay_areas = pd.read_excel('locality_latlon.xlsx',usecols=['locname', 'centroid_lat','centroid_lon']).drop(55)
locality_pay_areas.head()
```

Note that the locality pay area RUS (Rest of US) doesn't have a centroid, this is a special case for if a job doesn't all in any of the other locality pay areas in CONUS that will have to be handled.

Now, we can merge the 2024 GS Locality Pay Table and the Locality Pay Area definitions to get one DataFrame with pay area code, annual locality pay, and the centroid of that area.

```python
locality_pay = pd.merge(locality_pay_areas, locality_pay_schedule, on='locname')
locality_pay['centroid'] = locality_pay.apply(lambda row: (row['centroid_lat'], row['centroid_lon']), axis=1)
locality_pay = locality_pay.drop(['centroid_lat','centroid_lon'], axis=1)
locality_pay.head()
```

Now we have cost-of-living area for various locations in the US.  However, the original dataset does not have locality pay area.  Therefore, we need to determine each job listing's locality pay area based on Posting Location.  We will do this by determining which locality pay area centroid is closest to the job listing.

```python
from geopy.geocoders import Nominatim

def city_to_latlon(city):
    geolocator = Nominatim(user_agent="myApp")
    location = geolocator.geocode(city)
    if location:
        return location.latitude, location.longitude
    else:
        assert 'city'+city+'not found'

nyc_latlon = city_to_latlon('New York')
nyc_latlon
```

We can see we have created a function that can find a city and return its lat/lon to be compared against the centroids of locality pay areas.

```python
from geopy.distance import geodesic

def latlon_to_lpa(job_latlon):
    calculate_distance = lambda centroid:geodesic(job_latlon, centroid).kilometers
    locality_pay_nonans = locality_pay.dropna(axis=1)
    distance = locality_pay_nonans.centroid.apply(calculate_distance)
    closest_distance = distance.idxmin()
    if closest_distance > 200: #if the closest lpa centroid is more than 200km away, just use the Rest of US salary
        return "RUS"
    else:
        return locality_pay_nonans.locname.iloc[closest_distance]

latlon_to_lpa(nyc_latlon)
```

As expected, we see that that city of New York lies in locality pay area code NY, which corresponds to "New York-Newark, NY-NJ-CT-PA", which is correct.  Note we have handled the exception for locality pay area RUS by saying if a job is more than 200mi from the nearest locality pay area centroid, it probably isn't in a metro area locality pay area and can be assigned as RUS.  This 200mi threshold was determined by visually looking at [a map](https://www.federalpay.org/gs/locality).  Additionally, we handle the exception where the location can't be found, these are just given a locality pay from the Rest of US (RUS) locality pay area.  Finally, we must normalize the job listing pay with the GS-1 locality pay.

Finally, we join the functions made earlier together.  The city lookup tool `geolocator` will fail for some obscure or poorly-labeled Posting Locations so it is necessary to use a for loop instead of Pandas' `df.apply()` feature.

```python
def location_to_lp(location):
    try:
        latlon = city_to_latlon(location)
        lpa = latlon_to_lpa(latlon)
        return locality_pay[locality_pay.locname == lpa].iloc[0].annual
    except:
        return 25684 #if city can't be found, it must be small so use the RUS locality pay

from tqdm import tqdm
job['lp'] = 'failed'
for i in tqdm(range(len(job))):
    row = job.iloc[i,:]
    try:
        job.iloc[i,-1] = location_to_lp(row.location)
    except:
        job.iloc[i,-1] = 25684
```

Finally, we filter out jobs with no valid locality pay area code, and normalize the annual salary by the locality pay.  Our data cleaning is complete.

```python
job['sal_norm_lp'] = job.yr_salary / job.lp
job.head()
assert len(job) == 168
```

## More Statistical Tests

One may suspect that programming language and job choice are confounding variables.  Each discipline has its preferred language and maybe language acts as a proxy for job or vice versa.  Let's use the Chi-Squared Test for Independence to see if there is any significant difference between all combinations of jobs and programming languages.

```python
for i in job_no_outliers.index:
    if (job_no_outliers['Python'][i]==1) & (job_no_outliers['Java'][i]==0):
        job_no_outliers.loc[i,'chi-test']='Python'
    elif (job_no_outliers['Python'][i]==0) & (job_no_outliers['Java'][i]==1):
        job_no_outliers.loc[i,'chi-test']='Java'
    else:
        job_no_outliers.loc[i,'chi-test']='Both'
job_no_outliers
```

```python
crosstab = pd.crosstab(job_no_outliers['title'], job_no_outliers['chi-test'])
crosstab
```

```python
stats.chi2_contingency(crosstab)
```

We can reject the null hypothesis that there is no a relationship between programming languages and job titles as the p-value is 2.708e-18, less than 0.01. There is a significant difference in programming languages between Data Scientists and Software Engineers, meaning that programming languages might be confounding variables that can affect the job titles on salary.

As one final test, let's perform a one-way ANOVA test to see if any of the mean salaries differ across groups.

```python
from scipy.stats import f_oneway
f_oneway(job_no_outliers[(job_no_outliers.title == 'Data Scientist') & (job_no_outliers['chi-test'] == 'Both')].sal_norm_lp,
         job_no_outliers[(job_no_outliers.title == 'Data Scientist') & (job_no_outliers['chi-test'] == 'Python')].sal_norm_lp,
         job_no_outliers[(job_no_outliers.title == 'Data Scientist') & (job_no_outliers['chi-test'] == 'Java')].sal_norm_lp,
         job_no_outliers[(job_no_outliers.title == 'Software Engineer') & (job_no_outliers['chi-test'] == 'Both')].sal_norm_lp,
         job_no_outliers[(job_no_outliers.title == 'Software Engineer') & (job_no_outliers['chi-test'] == 'Python')].sal_norm_lp,
         job_no_outliers[(job_no_outliers.title == 'Software Engineer') & (job_no_outliers['chi-test'] == 'Java')].sal_norm_lp)
```

Again, nope.  We fail to reject the null hypothesis that they differ.

# Discussion and Conclusion

Our project sought to investigate whether the choice of programming language and job title significantly influenced the normalized mean salary for mid to senior-level positions of data scientist and software engineer.

In our EDA, there is an apparent relationship between programming skills and job titles. We found that data scientist positions require Python, while software engineer positions look for candidates who are proficient in Java. In our statistical analysis, we initially found a difference in mean salary between the two professions. Yet, our t-test revealed that the difference in mean salary across the two groups is not statistically significant. Our Chi-Squared test suggested with high statistical significance that programming language and job titles are confounding variables that influence average salary. While these are leading in the industry, they don’t encompass the full diversity of technical skills that employers look for. The proficiency in other programming languages, like SQL, R, and C++, is growing in popularity and may affect potential salaries. Further, our methodology included eliminating lots of job postings that lacked transparent compensation details. Although it helped us enhance the validity of our data, it also contributed to limiting the breadth of our analysis.
