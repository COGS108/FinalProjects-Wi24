# Abstract

This paper examines the relationship between movie genres and critic sentiment scores on Rotten Tomatoes over the 2010 to 2020 decade. Using a mix of machine learning, statistics, and visualizations, we investigate how sentiment varies across genres. We focus on drama, comedy, documentary, action & adventure, and mystery & suspense genres for analysis. 

Our findings consistently underscore that documentaries tend to receive higher average sentiment scores, contrasting action & adventure, comedy, and mystery & suspense genres which receive lower but steady sentiment scores.

An interesting find is that the documentary genre is the one genre that contains movies with a higher likelihood of staying true to a single genre without mixing in other styles and genres. The other genres all contain movies that can be found in multiple differing genres. 

To analyze this data, we deploy Support Vector Machines, a sophisticated yet widely used tool to help us uncover patterns and trends. By using this tool, we gain valuable insights into how sentiment changes across genres and across time, providing a deeper understanding of how critics perceive different types of movies.

Overall, our analysis reveals shifts in overall sentiment over the years, possibly influenced by external factors. Surprisingly, genres like comedy and action & adventure often have lower sentiment scores, challenging conventional assumptions about their critical appeal. These findings provide insights into film criticism, shedding light on the changing dynamics of movie genres and how they are perceived by critics.

# Research Question

What is the relationship between a movie’s genre and the sentiment score of its collection of critic reviews on Rotten Tomatoes, defined as the aggregate positivity or negativity of the reviews associated with the movie over the period of 2010 to 2020? Specifically, we will explore whether certain genres (drama, comedy, mystery and suspense, documentary, action and adventure) tend to have higher or lower sentiment scores over time, and whether the sentiment score varies for each genre year by year.

### rotten_tomatoes_critic_reviews

Let's first load the dataset and look at the quality of the data:

```python
import pandas as pd

critic_reviews_df = pd.read_csv('./datasets/rotten_tomatoes_critic_reviews.csv')
critic_reviews_df.head()
```

Note that each observation is a single review mapped to a movie where the movie is described by the rotten_tomatoes_link variable. This means movies have potentially multiple reviews to them. We want to create a dataframe that maps rotten_tomatoes_link to a list of reviews, where each review is just the value from review_content. We don't really care about any other variable as we don't care about the critics, publishers, review scores, and date the review was made.

```python
critic_reviews_df_slim = critic_reviews_df.iloc[:, [0, -1]]
critic_reviews_df_slim.head()
```

Right now, multiple values of review_content will map to a single value of rotten_tomatoes_link. We want it so that a single value of rotten_tomatoes_link maps to a list of review_content values. This will make it so that a single movie maps to multiple reviews.

We also want to be careful of null reviews. Let's get rid of all values of review_content that are invalid and then perform our "transformation".

```python
# Check if 'review_content' has any null values
has_null = critic_reviews_df_slim['review_content'].isnull().any()

# If 'review_content' has null values, filter them out
if has_null:
    print("Found null values in review_content")
    critic_reviews_df_slim = critic_reviews_df_slim[~critic_reviews_df_slim['review_content'].isnull()]

# Create a new DataFrame with a list of review_content values associated with each rotten_tomatoes_link
reviews_df = critic_reviews_df_slim.groupby('rotten_tomatoes_link')['review_content'].apply(list).reset_index()

# Rename the 'review_content' column to 'reviews'
reviews_df.rename(columns={'review_content': 'reviews'}, inplace=True)

reviews_df.head()
```

That's all the cleaning we can do for now with respect to this dataset! Later, we'll join the information we find here with our second dataset. But we've to clean the second dataset first...

### rotten_tomatoes_movies

Like our first dataset, let's load in the second one and look at the quality of data:

```python
movies_df = pd.read_csv('./datasets/rotten_tomatoes_movies.csv')
movies_df.head()
```

Now we need to do the same procedure we did in the previous dataset and map each movie to a list, only the list is the list of genres associated with that movie.

```python
movies_to_genres = filtered_movies_df.groupby('rotten_tomatoes_link').agg({
    'genres': list,
    'movie_title': 'first',
    'original_release_date': 'first'
})
movies_to_genres.head()
```

### Creating the Sentiment Analysis Classifier

Now that we have a list of reviews for each movie entry, we need to create a sentiment analysis classifier that is trained on external sentiment analysis data, which can then classify the positivity and negativity of each review per movie entry. Then, we can quantify the positivity and negativity sentiment of each movie for further analysis.

The training data is taken from a GitHub repository that has a properly classified movie review dataset, a perfect analogue for the Rotten Tomato reviews we already have.

```python
# train Data
trainData = pd.read_csv("https://raw.githubusercontent.com/Vasistareddy/sentiment_analysis/master/data/train.csv")
# test Data
testData = pd.read_csv("https://raw.githubusercontent.com/Vasistareddy/sentiment_analysis/master/data/test.csv")
```

```python
trainData.head()
```

The following code is using the TfidfVectorizer from Scikit-learn to transform our text data into a format that can be used in machine learning models. Now let's convert our raw data into a matrix of TF-IDF features.

```python
from sklearn.feature_extraction.text import TfidfVectorizer

# Create feature vectors
vectorizer = TfidfVectorizer(min_df = 5,
                             max_df = 0.8,
                             sublinear_tf = True,
                             use_idf = True)
train_vectors = vectorizer.fit_transform(trainData['Content'])
test_vectors = vectorizer.transform(testData['Content'])
```

The classifier model used will be the Support Vector Machine (SVM) since it provides a nice margin surrounding the linear decision boundary the model creates, increasing the accuracy of our model on future data. Previous research referred to in the background section notes that SVM performs better than other models that create linear decision boundaries for movie review sentiment analysis.

```python
import time
from sklearn import svm
from sklearn.metrics import classification_report

# Perform classification with SVM
classifier_linear = svm.SVC(kernel='linear')

t0 = time.time()
classifier_linear.fit(train_vectors, trainData['Label'])
t1 = time.time()

prediction_linear = classifier_linear.predict(test_vectors)
# print(test_vectors.shape)
# print(prediction_linear.shape)

t2 = time.time()
time_linear_train = t1-t0
time_linear_predict = t2-t1

# results
print("Training time: %fs; Prediction time: %fs" % (time_linear_train, time_linear_predict))
report = classification_report(testData['Label'], prediction_linear, output_dict=True)
print('positive: ', report['pos'])
print('negative: ', report['neg'])
```

Now that the classifier is trained, we can classify every review for every movie entry and create a sentiment value from 0.0 to 1.0 that determines the mean sentiment for a single movie. A score of 0.0 means the sentiment is extremely negative while a score of 1.0 means the sentiment is extremely positive, while a score of 0.5 means a neutral sentiment.

Below is a function that will be applied to every row of our merged dataframe that will give us the sentiment scores based on the review columns in our dataset so that each movie received an overall sentiment score.

```python
def generateSentiment(reviews, classifier, vectorizer):
    reviews_vector = vectorizer.transform(reviews)
    predictions = classifier.predict(reviews_vector)
    
    # Convert 'pos' to 1 and 'neg' to 0
    converted_values = [1 if value == 'pos' else 0 for value in predictions]
    
    # Calculate the mean
    mean_value = sum(converted_values) / len(converted_values)
    return mean_value
```

```python
df_merged['sentiment'] = df_merged['reviews'].apply(lambda x: generateSentiment(x, classifier_linear, vectorizer))
```

### Statistical Analysis

Apart from visual analysis, statistical tests can be a great avenue in determining the relationships between variables. Let's first use the ANOVA test to test differences in mean sentiment scores for each genre and see if there is a statistically significant relationships between genre and sentiment.

```python
from scipy.stats import f_oneway

# Perform ANOVA
genre_groups = [df_exploded[df_exploded['genres'] == genre]['sentiment'] for genre in df_exploded['genres'].unique()]
f_statistic, p_value = f_oneway(*genre_groups)

print("F-statistic:", f_statistic)
print("p-value:", p_value)
```

The p-value is so small that it may as well be zero! This means that we can reject the null hypothesis that there is no difference in mean sentiment scores across genres. So, it is true that there is a statistically significant difference in mean sentiment across our selected genres.

Let's do further post-hoc analysis like Tukey's HSD (Honestly Significant Difference) to identify which specific genres differ significantly from each other in terms of sentiment scores.

```python
import statsmodels.api as sm
from statsmodels.stats.multicomp import pairwise_tukeyhsd

# Perform Tukey's HSD test
tukey_results = pairwise_tukeyhsd(df_exploded['sentiment'], df_exploded['genres'])

# Print the summary of Tukey's HSD test
print(tukey_results.summary())
```

Most pairs have significant differences between their mean sentiment scores, except for the following three pairs:
- Action & Adventure and Comedy
- Action & Adventure and Mystery & Suspense
- Comedy and Mystery & Suspense

This correlates nicely with the information given to us by the heatmap and the stacked line plots as the three genres of Action & Adventure, Comedy, and Mystery & Suspense are the only genres with non-significant pairwise differences, as well as having relatively similar sentiment score over the years.

Next, let's compare genre with their sentiment scores using OLS regression. While the visual analysis have already showed clear differences in sentiment score for specific genres, more information can be attained for the genres of Action & Adventure, Comedy, and Mystery & Suspense. We first start by creating dummy variables for each genre. For each movie in our original DataFrame, we classify a '1' in a genre slot if the movie is of that genre and a '0' otherwise.

```python
# Create dummy variables for the 'genres' column
dummy_genre = pd.get_dummies(df_exploded['genres'], prefix='genres', drop_first=False)

# Concatenate the dummy variables
df_with_dummies = pd.concat([df_exploded, dummy_genre], axis=1)

# Group by movie title and aggregate the dummy variables
aggregated_df = df_with_dummies.groupby('movie_title').sum()

# Reset the index to make 'movie_title' a column again
aggregated_df = aggregated_df.reset_index()

aggregated_df
```

Next, we run OLS:

```python
X = aggregated_df[['genres_Action & Adventure', 'genres_Comedy', 'genres_Documentary', 'genres_Drama', 'genres_Mystery & Suspense']]  # Independent variables (dummy variables)
y = aggregated_df['sentiment']  # Dependent variable

X = sm.add_constant(X)
model = sm.OLS(y, X).fit()
print(model.summary())
```

Lots of information can be gleamed from this:

- The R-squared result of 0.532 indicates that 53.2% of the variation in the sentiment scores among movies can be attributed to which different movie genres each movie is classified in.
- The F-statistic and its corresponding p-value that's very close to zero indicates that the regression model is statistically significant. This means our dummy variables representing movie genres, together, have a significant impact on predicting sentiment scores.
- The genres in order of strongest to weakest influence on sentiment score (determined by the magnitude of the respective coefficient) are: Documentary, Drama, Comedy, Action & Adventure, and Mystery & Suspense

Finally, let's analyze the fact that movies can be part of multiple genres. If a movie is classified as one genre, how likely will the same movie be classified to a different genre? A correlation matrix can give some interesting answers.

```python
# Extract only the columns corresponding to the dummy variables
dummy_variables = aggregated_df[['genres_Action & Adventure', 'genres_Comedy', 'genres_Documentary', 'genres_Drama', 'genres_Mystery & Suspense']]

# Calculate the correlation matrix
correlation_matrix = dummy_variables.corr()
print("Correlation Matrix:")
print(correlation_matrix)

plt.figure(figsize=(10, 8))
sns.heatmap(correlation_matrix, annot=True, cmap="coolwarm", fmt=".2f", linewidths=0.5)
plt.title("Correlation Matrix")
plt.show()
```

Each cell in the correlation matrix contains a correlation coefficient, which indicates the strength and direction of the linear relationship between two genres. The correlation coefficient ranges from -1 to 1 and to determine strong correlation between two genres, we care if the coefficient is extreme (whether it's close to either -1 or 1). The darker colors on the heatmap match to more "extreme" coefficient values. 

The only strong relationships are between Documentary and Drama: if a movie is in the Documentary genre, it is less likely to be in the Drama genre, and vice versa. Further, the Documentary genre has the most negative correlation coefficient for every other genre it's compared against. This makes sense in theory as documentaries focus on real-life events and have vastly different storytelling and production techniques compared to other movies.

# Discussion and Conclusion

Our study aimed to unravel the relationship between a movie's genre and the sentiment score of its critic reviews on Rotten Tomatoes from 2010 to 2020. We focused on key genres: drama, comedy, mystery and suspense, documentary, action, and adventure. Our analysis unveiled a notable shift in the movie landscape. This change aligns with the evolving preferences of digital audiences, who show a marked preference for episodic and cliffhanger-driven narratives.

With our SVM model that has around 90% accuracy, ANOVA Test, Tukey's HSD (Honestly Significant Difference), and OLS regression, we can proudly say that our model is working well and the result is statistically significant. Firstly, our analysis showed that certain genres consistently evoked stronger sentiments, both positive and negative, among critics. For instance, dramas and documentaries tended to receive higher sentiment scores, reflecting a possible appreciation for their depth and real-world relevance. Conversely, genres like action and adventure often had lower sentiment scores, potentially indicating a critical view of their emphasis on entertainment over substance.

Year-by-year analysis further highlighted that sentiment scores for each genre fluctuated over time, suggesting that external factors such as social trends or global events might influence critical reception. Interestingly, the sentiment scores for comedies showed a gradual decline over the decade, prompting speculation about changing humor sensibilities or the evolving landscape of comedy in cinema.

Another key finding was the impact of technological advancements and streaming services on movie genres. The data suggested a shift in the types of movies being produced and reviewed, with genres like mystery and suspense gaining traction, possibly due to the influence of digital platforms favoring episodic and cliffhanger-driven storytelling.

Our hypothesis proved to be false, as we believed comedy and action & adventure would outperform all other genres in at least one year. It was quite the contrary, as they appeared to be at the bottom nearly every year. This might be due to personal preferences of our taste in movies versus trained professionals who analyze movies on a daily basis. 

Limitations of our study include the reliance on Rotten Tomatoes as the sole source of critic reviews, which might not fully represent the diversity of critical opinion. Additionally, our analysis was confined to a specific timeframe and selected genres, which could limit the generalizability of our findings.

Future research could expand this work by incorporating a wider range of review sources, exploring additional genres, and examining the impact of viewer sentiment alongside critical reviews. Investigating the reasons behind the trends observed in our study could provide deeper insights into the evolving relationship between movie genres and their reception by critics and audiences alike.

In conclusion, our research offers valuable insights into the sentimental dynamics of movie genres, highlighting the ever-evolving landscape of film criticism and production. These findings contribute to our understanding of how genres resonate with critics and can inform future studies in the realms of film studies and sentiment analysis.
