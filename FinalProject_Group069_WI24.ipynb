# Abstract

In our project, we aim to figure out how the accessibility of healthcare, in relation to geographical distribution of hospitals, varies across different demographic groups over the past five years in different US cities. We also want to know if there are disparities in healthcare quality among these groups, including variations in hospital resources, funding, mortality rates, readmission rates, and patient satisfaction scores. We use various figures and OLS regression to analyze the dataset we acquired in order to determine the above relationships. The results suggest that there is a negative relationship between healthcare access and median household incomes. We also found that there is a negative relationship between percentage of Hispanic or Latino citizens and access to healthcare. There is a negative correlation between hospitals that received FHA-insured loans and quality of care. There is no notable relationship between healthcare quality with race or income. In short, there is a relationship between access to healthcare and different demographic groups as well as between access to healthcare and income, but there there is no relationship between healthcare quality and demographic group or income.

**242_All_Historical_Deals:** This dataset is a comprehensive list of all acute care hospital facilities with current or previous mortgage insurance provided by the FHA, adminitered by the Office of Hospital Facilities (OHF). FHA-insured loans can be used for remodeling, expansion, odernization, equiptment, refinancing, and acquisitions, which increases a hospitals ability to give higher quality care to more patients. The important variables in this dataset are "Project Name" (the name of the hospital facility), "Initial Endorsment" (the date of mortage insurance approval), and "Status" (if the loan is still active). Since most of the datatypes are strings, we will need to convert the "initial Endorsement" to a datetime object, filter for more recent dates, and also remove the last two variables from consideration ("Unnamed: 8"	and "Unnamed: 9") which both only contain null values. We will then map this list to the list of all hospital facilities and create a column of boolean values that indicate whether a hospital has received FHA-insured loans.

**Hospitals:** This dataset is a comprehensive list of all healthcare facilities in the United States, collected for the Homeland Infrastructure Foundation-Level Data (HIFLD) database. The important varibales in this data set are "Name", "Address", "City", "State", & "Zip" (to determine geographic hospital density), "Status" (to determine if the facility is still in use), "Beds" (to determine the hospital capcity and hospital-bed density), and "Trauma" (to determine the level of trauma care accessible). This dataset contains floats, ints, and strings. We will need to remove the unnecessary columns such as the latitude and longitude, and then this dataset will be the basis for all future mappings when combining. 

**Hospital_General_Information:** This dataset contains data provided by the Centers for Medicare & Medicaid Services that is used to compare quality of care at over 4000 medicare-certified hospitals. The important variables are "City", "State", "Address", "Hospital Name", & "ZIP Code", (to determine geographic hospital density), and 'Emergency Services', 'Meets criteria for meaningful use of EHRs', 'Hospital overall rating', 'Mortality national comparison', 'Safety of care national comparison', 'Readmission national comparison', 'Patient experience national comparison', 'Effectiveness of care national comparison', 'Timeliness of care national comparison', and 'Efficient use of medical imaging national comparison' (to measure quality of healthcare). The remaining columns, especially all the columns that are 'footnotes' of existing columns will be dropped as they only contain null values; the datatypes are strings and ints, but "ZIP Code" may be converted to a string to comply with formatting in other datasets. We will map these columns onto the 'Hospitals' dataframe and drop any hospitals that do not have matching data. 

**HCAHPS-Hospital.csv:**  Hospital Ratings for the Hospital Consumer Assessment of Healthcare Providers and Systems (HCAHPS), a national survey for hospital patients. Contains data concerning location(address, city, zip code, state), type of hospital, services provided, and overall ratings along with a description for the rating. This dataset includes ints and strings. We will remove individual response data and keep quantitative data from the overall linear scores and patient star ratings, along with pivoting the columns to have each row as a hospital with associated values rather than having each hospital have multiple rows. 

**median_household_income.csv:** This dataset is a five-year estimate of median household income in the US across different cities, which is a good indicator of people’s financial status from different cities. We plan to use the information about different races’ income since they could describe different demographic group’s financial status within that city. Since it’s a large dataset, we plan to drop columns that have mostly null values, as well as the margin of error columns. It has datatypes of only strings, so we may need to convert those to ints or floats to get descriptive data for the income. 

**us-cities-demographics.csv:** US Census Bureau's 2015 American Community Survey.  This dataset contains information on counts of races in cities, average age, gender populations, total population, veterans, and number of those who are foreign-born. Includes floats, ints and strings. We might use this to compare the qualities of hospitals in that area to the demographics present and how that might be correlated. We plan to use all of the columns in order to see which demographic factors have the strongest relationship with healthcare access and quality.  

To combine the datasets, we will merge the dataframes and then drop any rows that have majority null values. The hospitals will be mapped to the Hospitals.csv dataset, as it is a comprehensive list of all hospitals in the US.

### Is there a difference in quality of healthcare between hospitals that have received funding vs not? ###

We want to determine if receiving funding (FHA-insured loans) impacts the quality of care of hospitals. To visualize the difference in means we use a boxplot, and then perform a t-test to see if the difference in means is significant. 

```python
bp = sns.boxplot(x='received_loan', y='overall_rating', data=df_hospital, palette=sns.color_palette())
# change color and size of mean lines to make more noticable
mean_lines = [bp.lines[4], bp.lines[10]]
for line in mean_lines:
    line.set_linewidth(2.25)
    line.set_color('maroon')
bp.set_title('Hospitals that received loans related to lower quality of care ratings')
bp.set_ylabel('Overall Quality of Care Rating')
bp.set_xlabel('Recevied Loan')
```

We find that the average quality of care ratings of both groups are very similar, but the inner quartile range for hospitals that received loans is lower than the IQR for hospitals that did not receive loans, contradicting our hypothesis. This remains true even when we remove null values. 

$H_0: \bar x_{OverallRatingWithLoan} = \bar y_{OverallRatingWithoutLoan}$

$H_a: \bar x_{OverallRatingWithLoan} \ne \bar y_{OverallRatingWithoutLoan}$

```python
group1 = df_hospital.where(df_hospital.received_loan==0).dropna()['overall_rating']
group2 = df_hospital.where(df_hospital.received_loan==1).dropna()['overall_rating']
t_val, p_val = ttest_ind(group1, group2)
t_val, p_val
```

Using an $\alpha = 0.05$ value:

We reject the null in favor of the alternative hypothesis, concluding that the means between the groups are statiscally significantly different and that receiving FHA-insured loans plays a role on quality of care ratings.

We use people per hospital as an inverse indication of healthcare access and median household income to represent the income level of that city and we find that people per hospital is positively correlated with median household income, which suggests that healthcare accessibility is negatively influenced by income. This could potentially be explained by higher income cities having high population densities and therefore a higher people per hospital value. Additionally, when we try to investigate the relationship between access to healthcare and demographic variables such as age, gender, ethnicity using scatter matrix, we find that ethnicity is a possible factor. Further linear regression results show that percentage of Hispanic or Latino citizens has statistically significant negative relationship with healthcare access. 

Regarding our healthcare quality hypothesis, we use the ratings of hospitals since it combined and calculated factors we cared about such as mortality rate, readmission rate, patient satisfaction. We first try to focus on how funding (FHA-insured loans) may influence hospital quality of care and the results show that there is a negative relationship between receiving the loan and quality of care, which rejects our hypothesis. In additon, we found that there is no significant relationship between quality of care and income. As for the race/ethnicity factor, we only find a significant relationship between quality of care and hispanic populations, and the coefficent is negligible. Thus, the entire hypothesis is rejected.  

To further visualize the healthcare quality, we map quality ratings all over the US and the map presents that healthcare quality is lower in the Southwest, Florida, and East Coast and higher from the Northern Midwest down to Texas, indicationg that geographical area is a potential facotr that influences quality of care as well.

The findings of this study illustrate the relationship between healthcare accessibility, healthcare quality and demographic factors in various cities across the United states. Though the initial hypothesis suggests that higher income levels and increased hospital fundings would correspond to better access and quality of healthcare, the results challenge these assumptions. 

The negative relationship between median household income and healthcare accessibility can show the complexity of healthcare access. We found that income plays a role in it, but socioeconomic and educational variables also make a difference. This suggests that intervention should not only focus on economic measures but also use educational measures and community support initiatives. 

The relationship between hispanic populations and quality of care is particularly concerning since it highlights the inequalities persisting within the healthcare system.
