## Background and Prior Work


With the introduction of hybrid and remote learning into every level of education due to the COVID-19, the way that universities teach their students have changed since the peak of the pandemic in 2020. Many colleges now offer their courses online or record their in-person lectures for students to view on their own time. This has made learning much more accessible and convenient for most students, since having access to a recording of a lecture or being able to go to a discussion over Zoom can give students who can’t make it to class more opportunities to learn when they otherwise wouldn’t. This raises the question: what is the impact on academic performance on students who opt into the remote learning format over an in person one? Specifically, how has the addition of remote learning options at UCSD affected the academic performance of students who opt into this learning format versus those who attend class in person?

Based on research that was spurred by the move to online learning during the COVID-19 pandemic, many have found that remote learning generally is detrimental to learning outcomes in students. According to a simulation run to measure the potential losses in learning outcomes by students due to remote learning by Joao Pedro Azevedo, most students scored significantly lower on average due to online learning, scoring 7 points lower on the Programme for International Student Assessment test (22 - 23). Additionally, students in the U.S. have had worse learning outcomes due to online learning, as most have lost “3 months of learning in mathematics and 1.5 months in reading” that normally would have been learned with in-person teaching (Munoz-Najar et al., 38). Overall, it seems that remote learning has had general negative effects on students who have had to learn by this method.

```python
# Drop all the null values
capes = capes.drop(columns=['Evalulation URL'])
capes = capes.dropna(subset=['Average Grade Received'])
capes = capes.dropna(subset=['Average Grade Expected'])
```

```python
#Convert to float
def convert_float(df):
    percentage_cols = ['Percentage Recommended Class', 'Percentage Recommended Professor']
    for col in percentage_cols:
        df[col] = 100* pd.to_numeric(df[col].str.rstrip('%'), errors='coerce') / 100

    grade_cols = ['Average Grade Expected', 'Average Grade Received']
    for col in grade_cols:
        df[col] = pd.to_numeric(df[col].str.extract(r'\(([\d.]+)\)')[0], errors='coerce')

    exclude_cols = ['Instructor', 'Course', 'Quarter'] + percentage_cols + grade_cols
    object_cols = df.select_dtypes(include=['object']).columns

    for col in object_cols:
        if col not in exclude_cols:
            df[col] = pd.to_numeric(df[col], errors='coerce')

    return df
```

## Comparison between Stem and non Stem

```python
post_covid_stem = post_covid[post_covid["stem"] == True]
post_covid_non_stem = post_covid[post_covid["stem"] == False]
pre_covid_stem = pre_covid[pre_covid["stem"] == True]
pre_covid_non_stem = pre_covid[pre_covid["stem"] == False]
during_covid_stem = during_covid[during_covid["stem"] == True]
during_covid_non_stem = during_covid[during_covid["stem"] == False]

stem_covid_avg = during_covid_stem["r_gpa"].mean()
non_stem_covid_avg = during_covid_non_stem["r_gpa"].mean()

stem_pre_t_stat, stem_pre_pval = stats.ttest_1samp(pre_covid_stem["r_gpa"], stem_covid_avg)
non_stem_pre_t_stat, non_stem_pre_pval = stats.ttest_1samp(pre_covid_non_stem["r_gpa"], non_stem_covid_avg)

stem_post_t_stat, stem_post_pval = stats.ttest_1samp(post_covid_stem["r_gpa"], stem_covid_avg)
non_stem_post_t_stat, non_stem_post_pval = stats.ttest_1samp(post_covid_non_stem["r_gpa"], non_stem_covid_avg)
```

After running t-tests on the pre-COVID distribution vs. the average gpa during COVID for non-stem classes and stem classes separately, we see that is unlikely that we would find the average COVID gpa in either pre-COVID distribution randomly. We can infer that no matter the type of course that it taken, the changes during the COVID time period raised the average gpa by a statistically signifcant amount. This means the actions and measures the university took to accomadate for students during COVID affected stem and non-stem students equally.

```python
print("pre covid stem gpa distribution vs. during covid stem avg gpa: " + str(stem_pre_pval))
print("pre covid non-stem gpa distribution vs. during covid non-stem avg gpa: " + str(non_stem_pre_pval))
print("post covid stem gpa distribution vs. during covid stem avg gpa: " + str(stem_post_pval))
print("post covid non-stem gpa distribution vs. during covid non-stem avg gpa: " + str(non_stem_post_pval))
```

After running t-tests on the post-COVID distribution vs. the average gpa during COVID for non-stem classes and stem classes separately, we see that is unlikely that we would find the average COVID gpa in either post-COVID distribution randomly. We can infer that the changes that happened during COVID which spiked student's gpas were reverted after COVID.

```python
ttest_ind(post_covid_stem['r_gpa'], pre_covid_stem["r_gpa"])
ttest_ind(post_covid_non_stem['r_gpa'], pre_covid_non_stem["r_gpa"])
```

For the following analysis, we chose to make use of T-Tests, as our question is primarily one which we were looking to answer by considering the average gpas of classes, and we also wanted to confirm that any such differences were statistically significant, and not just changes that might have occurred by random chance. With this in mind, we used T-Tests on a few different samples from our data to more closely inspect these changes in distributions.

```python
pre_covid_non_stem = capes_covid[capes_covid['quarter'].str.contains(r'17$|18$|19$')]
pre_covid_non_stem = pre_covid_non_stem[pre_covid_non_stem['stem'] == False]
pre_covid_stem = capes_covid[capes_covid['quarter'].str.contains(r'17$|18$|19$')]
pre_covid_stem = pre_covid_stem[pre_covid_stem['stem'] == True]
stats.ttest_ind(pre_covid_non_stem["r_gpa"], pre_covid_stem["r_gpa"])
```

We can use Independent Samples T-Tests to determine whether the means of the non-STEM courses and the STEM courses bear a statistically significant difference in there means for the pre-COVID, during-COVID, and post-COVID periods as well. The first of these T-Tests, as ran in the two cells above, is for the pre-COVID period. As we can see from the incredibly low p-value, there is indeed statistical significance in the difference of the two populations' means in the 2017-2019 time period, and based on the statistic score, non-STEM courses had the higher mean for this period, which aligns with our visualizations of the distribtuion of the two groups in the EDA section.

```python
during_covid_non_stem = capes_covid[capes_covid['quarter'].str.contains(r'20$|21$')]
during_covid_non_stem = during_covid_non_stem[during_covid_non_stem['stem'] == False]
during_covid_stem = capes_covid[capes_covid['quarter'].str.contains(r'20$|21$')]
during_covid_stem = during_covid_stem[during_covid_stem['stem'] == True]

stats.ttest_ind(during_covid_non_stem["r_gpa"], during_covid_stem["r_gpa"])
```

Here we have run the T-Test for the second period, to compare the means of the STEM courses and non-STEM courses during COVID, and from the results, we can see that non-STEM maintains the higher mean out of the two groups, but the p-value is many magnitudes higher than in the pre-COVID era, though still low enough to indicate some statistical significance in the difference of means. This would seem to suggest that while there is some quality of STEM courses versus non-STEM courses which creates a significant difference in their average grades, independent of COVID, the gap in the means of the two groups is considerably shrunk in the COVID period as opposed to pre-COVID.

```python
post_covid_non_stem = capes_covid[capes_covid['quarter'].str.contains(r'22$|23$')]
post_covid_non_stem = post_covid_non_stem[post_covid_non_stem['stem'] == False]
post_covid_stem = capes_covid[capes_covid['quarter'].str.contains(r'22$|23$')]
post_covid_stem = post_covid_stem[post_covid_stem['stem'] == True]

stats.ttest_ind(post_covid_non_stem["r_gpa"], post_covid_stem["r_gpa"])
```

We run one more T-Test to check the means of the STEM and non-STEM courses against each other in the post-COVID period, to see if the difference between the two groups' means returns to something more like the pre-COVID period's, maintains the results from the during-COVID period, or whether we get a different result altogether. And as can be determined from the statistic and p-value above, non-STEM again has the higher average, and the p-value is much lower than in the during-COVID period, though not quite as low as in the pre-COVID period either. ## Comparing Study Hours Overall

```python
before_covid_study = pre_covid['study'].dropna()
during_covid_study = during_covid['study'].dropna()
post_covid_study = post_covid['study'].dropna()

t_stat, p_val = ttest_ind(during_covid_study, before_covid_study, nan_policy='omit')

t_stat_before_during, p_val_before_during = ttest_ind(before_covid_study.dropna(), during_covid_study.dropna(), equal_var=False)
t_stat_during_after, p_val_during_after = ttest_ind(during_covid_study.dropna(), post_covid_study.dropna(), equal_var=False)

print(f"T-statistic Before vs. During: {t_stat_before_during}, P-value: {p_val_before_during}")
print(f"T-statistic During vs. After: {t_stat_during_after}, P-value: {p_val_during_after}")


plt.figure(figsize=(10, 6))
study_hour = pd.concat([before_covid_study, during_covid_study, post_covid_study], axis=1)

study_hour.columns = ['Before COVID-19', 'During COVID-19', 'After COVID-19']
study_hour = study_hour.melt(var_name='Period', value_name='Study Hours per Week')

sns.boxplot(x='Period', y='Study Hours per Week', data=study_hour)
plt.title('Comparison of Study Hours Before, During, and After COVID-19')
plt.show()
```

- The t-stats and p-value suggest that a statistically significant decrease in study hours from the period pre COVID-19 to during COVID-19.It also implies that the mean study hours during COVID-19 were higher than mean study hours before COVID19
- However, there aren't a significant change after the COVID-19 in terms of mean study hours,but there are a few outliers in the study hours suggesting that some students are not used to traditional class settings.
