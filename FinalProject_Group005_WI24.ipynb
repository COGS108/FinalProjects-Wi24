## Background and Prior Work

The frequency of volcanic eruptions in association with average country surface temperatures over the years can provide insight into how these naturally occurring phenomena have altered past and future temperature predictions per country. A combination of assessing volcanic eruption frequency must be examined in association with surface temperatures over time to determine impacts on volcanic activity in relation to temperature.

```python
import matplotlib.pyplot as plt
import numpy as np
from scipy.optimize import curve_fit
import seaborn as sns

temperatures = new_final['AverageTemperature']
eruptions = new_final['eruptions_count']

# Index of outlier
outlier_index = eruptions.idxmax()
outlier_temp = temperatures[outlier_index]
outlier_eruptions = eruptions[outlier_index]

def custom_function(x, a, b, c):
    return a * (x - outlier_temp) ** 2 + b * (x - outlier_temp) + c
popt, _ = curve_fit(custom_function, temperatures, eruptions)

# Modify the custom function to pass through the outlier point
def custom_function_with_constraint(x, a, b, c):
    return custom_function(x, a, b, c) + outlier_eruptions - custom_function(outlier_temp, a, b, c)

popt_with_constraint, _ = curve_fit(custom_function_with_constraint, temperatures, eruptions)
x_values = np.linspace(temperatures.min(), temperatures.max(), 100)

plt.scatter(temperatures, eruptions, color='blue', label='Data points')

plt.plot(x_values, custom_function_with_constraint(x_values, *popt_with_constraint), color='red', label='Fitted curve (with constraint)')

plt.xlabel('Average Land Temperature')
plt.ylabel('Number of Eruptions')
plt.title('Correlation Between Number of Eruptions and Land Temperature')
plt.legend()
plt.show()
```

Based on this graph, there is a lack of correlation between the two factors again. We plotted a custom curve to highlight an outlier that represents a high average land temperature with a high number of eruptions.

Thus the formation and geographical distribution of volcanos impact the elevation and type of volcano which is why stratovolcanoes are the most frequently erupting type as well as the second highest in elevation.
