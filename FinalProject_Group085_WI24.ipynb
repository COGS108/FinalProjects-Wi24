## Background and Prior Work

MBTI is a four letter indicator of a person’s personality type. We will be exploring if it is possible to guess someone’s personality type based off of the way they post on social media. Furthermore, we will look into which of the four attributes is most predictable. 

The scale between introversion (I) and extroversion (E) can be defined by how an individual gains energy. Extroverts are “energized by the outer world of people and things” while introverts prefer to do this by “spending time alone and in their heads”. The second letter is determined by the way someone takes in information. Sensors (S) “process things in a step-by-step linear sequence”. They are more inclined to remember the details of things as opposed to the intuitives (N) who are more inclined to look for “big picture or themes” when learning new things. The third letter determines how a person makes decisions with thinkers (T) being more inclined to use logic and take objective stances while feelers (F) take the impact of their actions on others into consideration. Finally, the fourth letter describes the way someone organizes their lives. People who are judgers (J) are more inclined to be schedule oriented planners, preferring structure and routine, while perceivers (P) prefer spontaneity and are more flexible<a name="cite_ref-1"></a>[<sup>1</sup>](#cite_note-1) (Melissa, MBTIonline).

This set was generated with the intention of machine learning being used to analyze the key differences between the way the differing personalities converse. ### Balance data

For this project, we will choose undersampling method, using the size of smallest personality type (ESFJ)
by randomly removing the majority class. This will help balance our dataset to counteract the skew, as can be seen in the above figure.

```python
target_sample_size = combined_df['type'].value_counts().min()

# Perform undersampling
undersampled_df = pd.DataFrame()

for mbti_type in combined_df['type'].unique():
    subset = combined_df[combined_df['type'] == mbti_type].sample(target_sample_size, random_state=42)
    undersampled_df = pd.concat([undersampled_df, subset])
```

#### Clean post contents

We used Natural Language Toolkit to remove any unecessary objects, such as URLs or punctuation.

```python
# download necessary librarys and dictionaries
nltk.download('wordnet', quiet=True)
nltk.download('stopwords', quiet=True)
lemmatiser = WordNetLemmatizer()
stop_words = stopwords.words('english')
```

```python
def clean_posts(word):
    # format the string
    word = word.lower()
    word = word.strip()
    # remove urls
    word = re.sub('(http:|ftp:|https:|www.)\S*', '', word)
    # remove email address
    word = re.sub('.*@.*\.com', ' ', word)
    # keep only letters
    word = re.sub('[^a-z]', ' ', word)
    # remove stop words and lemmatize the words
    word = " ".join([lemmatiser.lemmatize(w) for w in word.split(' ') if w not in stop_words])
    # remove MBTI types
    for type in mbti_list:
        word = word.replace(type.lower(), '')
    # remove redundent spaces
    word = re.sub('\s+', ' ', word)

    word = word.strip()
    return word
```

```python
undersampled_df['posts'] = undersampled_df['posts'].apply(clean_posts)
```

#### Encode MBTI types into binaries for modeling by following rules

<pre>Introversion (I): 0 |  Extroversion (E): 1
Intuition    (N): 0 |  Sensing      (S): 1
Thinking     (T): 0 |  Feeling      (F): 1
Judging      (J): 0 |  Perceiving   (P): 1</pre>


```python
def binarize_mbti(type):
    type = list(type)
    output = []
    # first letter
    output.append(0 if type[0] == 'I' else 1)
    # second letter
    output.append(0 if type[1] == 'N' else 1)
    # third letter
    output.append(0 if type[2] == 'T' else 1)
    # fourth letter
    output.append(0 if type[3] == 'J' else 1)
    return output

binarize_mbti('ISTP')
```

#### Apply function to the 'type' column

```python
undersampled_df['type'] = undersampled_df['type'].apply(binarize_mbti)
```

### Sentiment Analysis

We want to perform Sentiment Analysis to see if there are any trends that we can find in the data with regards to the different personality types.

We begin by importing the TextBlock class from the textblob library that calculates the sentiment polarity of ranging from -1 to 1. From this polarity score, we classify the scores into 3 different values into the column "sentiment." The three values are: negative (if the score is less than 0), neutral (if the score is 0), and positive (if the score is more than 0). Textblob determines sentiments based on semantic relations and the frequency of each word in the input sentence.

```python
from textblob import TextBlob

# gets polarity score from text
def get_polarity(text):
    return TextBlob(text).sentiment.polarity

undersampled_df['polarity'] = undersampled_df['posts'].apply(get_polarity)

# classifies polarity score into sentiments
def classify_sentiment(polarity):
    if polarity < 0:
        return 'Negative'
    elif polarity > 0:
        return 'Positive'
    else:
        return 'Neutral'
    
undersampled_df['sentiment'] = undersampled_df['polarity'].apply(classify_sentiment)

undersampled_df
```

### Splitting the Data into T/F

We chose to dive deeper into comparing Thinkers and Feelers for a few reasons. There was the highest accuracy percentage in prediction of T/F during the TF-IDF, so we wondered if there were sentiment differences. Additionally, these categories were ones that we considered more in our hypothesis due to emotional nuances, so diving deeper into the differences would help us see any differences.

```python
# convert type column to string for easier plotting
undersampled_df['type_str'] = undersampled_df['type'].apply(lambda x: ''.join(map(str, x)))

# function to unbinarize mbti
def unbinarize_mbti(type):
    output = []
    # first letter
    output.append('I' if type[0] == 0 else 'E')
    # second letter
    output.append('N' if type[1] == 0 else 'S')
    # third letter
    output.append('T' if type[2] == 0 else 'F')
    # fourth letter
    output.append('J' if type[3] == 0 else 'P')
    return ''.join(output)

undersampled_df['type_str'] = undersampled_df['type'].apply(unbinarize_mbti)

# function to classify each row as either thinking or feeling
def TF(type):
    return 'F' if type[2] == 1 else 'T'

undersampled_df['T/F'] = undersampled_df['type'].apply(TF)

undersampled_df
```

## 2-Gram

We wanted to find the most frequent 2-grams of the posts. This refers to a sequence of 2 words next to each other, which takes into account more context than just looking at word clouds. This could give us insight into common wording usage.

```python
from sklearn.feature_extraction.text import CountVectorizer

# create the CountVectorizer object and transform the text data w/ 2-grams
count_vect = CountVectorizer(ngram_range=(2,2)) 
X_counts = count_vect.fit_transform(undersampled_df['posts'])

# counts to frequencies
X_freq = X_counts.sum(axis=0)
freq = [(word, X_freq[0, idx]) for word, idx in count_vect.vocabulary_.items()]
freq =sorted(freq, key = lambda x: x[1], reverse=True)

# 2-grams from most frequent to least
freq[:10]
```

From the output of our frequent 2-grams, phrases like "feel like", "sound like", and "look like" are commonly used where all of these are common phrases to express an individual's perceptions or feelings, which might be particularly relavant to F types.

The frequent use of "pretty much" and "even though" could indicate a preference for qualifying statements or expressing nuanced views, which might be more common among Intuitive (N) types who enjoy exploring complex ideas.

The phrase ‘make sense’ being common could suggest a logical thought process, potentially more prevalent in Thinking (T) types.

The presence of ‘make sure’ might indicate a focus on planning or ensuring certain outcomes, which could be more characteristic of Judging (J) types.

## Topic Modeling

uses Latent Dirichlet Allocation (LDA) to identify the main topics discussed by each MBTI type

```python
from sklearn.decomposition import LatentDirichletAllocation
from sklearn.feature_extraction.text import TfidfVectorizer

mbti_types = undersampled_df['type_str'].unique()

for mbti in mbti_types:
    print(f"\n\nTopics for MBTI type {mbti}:")

    # filter posts for the current MBTI type
    posts = undersampled_df[undersampled_df['type_str'] == mbti]['posts']

    # use tf-idf features for LD
    vectorizer = TfidfVectorizer(max_df=0.95, min_df=2, stop_words='english')
    tfidf = vectorizer.fit_transform(posts)

    lda = LatentDirichletAllocation(n_components=10, random_state=0)
    lda.fit(tfidf)

    # print  the top words associated with each topic
    feature_names = vectorizer.get_feature_names_out()
    for topic_idx, topic in enumerate(lda.components_):
        print(f"Topic #{topic_idx + 1}:")
        print(" ".join([feature_names[i] for i in topic.argsort()[:-10 - 1:-1]]))
```

uses Latent Dirichlet Allocation (LDA) to identify the main topics discussed by Thinking Types vs. Feeling Types

```python
from sklearn.decomposition import LatentDirichletAllocation
from sklearn.feature_extraction.text import TfidfVectorizer

tf_column = undersampled_df['T/F'].unique()

for types in tf_column:
    if (types is "F"):
        print(f"\n\nTopics for Feeling Types:")

    if (types is "T"):
        print(f"\n\nTopics for Thinking Types:")

    # filter posts for the current MBTI type
    posts = undersampled_df[undersampled_df['T/F'] == types]['posts']

    # use tf-idf features for LD
    vectorizer = TfidfVectorizer(max_df=0.95, min_df=2, stop_words='english')
    tfidf = vectorizer.fit_transform(posts)

    lda = LatentDirichletAllocation(n_components=10, random_state=0)
    lda.fit(tfidf)

    # print  the top words associated with each topic
    feature_names = vectorizer.get_feature_names_out()
    for topic_idx, topic in enumerate(lda.components_):
        print(f"Topic #{topic_idx + 1}:")
        print(" ".join([feature_names[i] for i in topic.argsort()[:-10 - 1:-1]]))
```

From the results, it appears that Feeling types frequently discuss topics related to personal experiences, emotions, and social interactions, as seen by words like ‘love’, ‘good’, ‘friend’, and ‘awesome’. This aligns with the characteristics of Feeling types who tend to be empathetic and focus on social harmony.

On the other hand, Thinking types seem to discuss a wider range of topics including politics (‘trump’, ‘vote’, ‘country’), gaming (‘game’, ‘play’, ‘player’), and practical matters (‘money’, ‘work’, ‘buy’, ‘job’). This could reflect the logical and objective nature of Thinking types, as well as their tendency to be more task-oriented.

However, both Feeling and Thinking types seem to discuss topics related to thinking and knowledge (‘think’, ‘know’, ‘type’, ‘thing’), which might suggest that these aspects are not as distinguishable between the two types as hypothesized.
