```python
df = df.dropna(subset=["County", "City"])
```

```python
def simplify_utility(string):
    original = string
    string = string.lower()
    if 'puget sound energy' in string:
        return 'PUGET SOUND ENERGY INC'
    elif 'bonneville power administration' in string:
        return 'BONNEVILLE POWER ADMINISTRATION'
    elif 'pud' in string:
        return 'PUBLIC UTILITY DISTRICT'
    return original

df['Electric Utility'] = df['Electric Utility'].apply(simplify_utility)
```

```python
np.array(df['Electric Utility'].unique())
```

```python
west_coast_states = ["CA", "OR", "WA"]
west_coast_data = df[df['State'].isin(west_coast_states)]

ev_popularity = west_coast_data.groupby('Model Year').size()

years = ev_popularity.index.to_numpy()
ev_count = ev_popularity.values

plt.figure(figsize=(10, 6))
plt.plot(years, ev_count, marker='o')
plt.title('EV Popularity Increase on the West Coast')
plt.xlabel('Model Year')
plt.ylabel('Number of EVs')
plt.grid(True)
plt.xticks(years)
plt.tight_layout()
plt.show()
```

This graph shows us the trend in EV popularity over time. Clearly, the popularity of EVs has increased as time goes on with the exception of a dip in EV count in the year 2019. The dip at the end of the graph, in year 2024, is not valid because obviously the year 2024 just started.

### Section 3 - Multiple regression analysis on MSRP of EVs

In this section, we'll be using the relationships we explored earlier to create a multiple regression model that predicts MSRP of EVs. This will also allow us to better understand the correlations between variables.

```python
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import OneHotEncoder, StandardScaler
from sklearn.compose import ColumnTransformer
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error
from sklearn.model_selection import train_test_split
```

```python
filter_msrp_df = df[df['Base MSRP'] > 0]
X = filter_msrp_df.drop('Base MSRP', axis=1)
y = filter_msrp_df['Base MSRP']

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2)

col_trans = ColumnTransformer([
    ('ohe', OneHotEncoder(drop='first'), ['Make', 'Electric Vehicle Type', 'Clean Alternative Fuel Vehicle (CAFV) Eligibility']),
    ('std-scale', StandardScaler(), ['Model Year', 'Electric Range'])
], remainder='drop')

model = Pipeline([
    ('col-trans', col_trans),
    ('lin-reg', LinearRegression())
])

model.fit(X_train, y_train)
```

```python
pred_train = model.predict(X_train)
rmse_train = mean_squared_error(y_train, pred_train, squared=False)
print("Root Mean Squared Error on training set: " + str(rmse_train))
```

```python
pred_test = model.predict(X_test)
rmse_test = mean_squared_error(y_test, pred_test, squared=False)
print("Root Mean Squared Error on test set: " + str(rmse_test))
```

```python
coefficients = model.named_steps['lin-reg'].coef_
feature_names = (
    list(col_trans.named_transformers_['ohe'].get_feature_names(['Make', 'Electric Vehicle Type', 'Clean Alternative Fuel Vehicle (CAFV) Eligibility']))
    + ['Model Year', 'Electric Range']
)
coefs_df = pd.DataFrame({'Feature': feature_names, 'Coefficient': coefficients})
coefs_df
```

In the above code, we split the data into a training set and a test set using a ratio of 80%-20%. The features we decided to use were Make, Electric Vehicle Type, Clean Alternative Fuel Vehicle Eligibility, Model Year, and Electric Range. The categorical variables like Make, EV Type, and CAFV Eligibility were one-hot encoded, meaning they were turned into numerical values. The quantitiative variables were standardized.

After training, the model had a root-mean-square error of **7945.2** for predicting Base MSRP on the training set. RMSE on the test was **28182.2**.

We also looked at the coefficients of variables used in the multiple linear regression and found that Make tended to have the largest impact on MSRP compared to other variables like EV type which only had a coefficient of -2597.7. The single most impactful factor according to our regression model was whether or not the Make was Porsche, with a coefficient of 67831.

```python
residuals_train = y_train - pred_train
sns.residplot(pred_train, residuals_train, lowess=True)
plt.xlabel('Predicted Values')
plt.ylabel('Residuals')
plt.title('Residual Plot for Training Set')
plt.show()
```

```python
residuals_test = y_test - pred_test
sns.residplot(pred_test, residuals_test, lowess=True)
plt.xlabel('Predicted Values')
plt.ylabel('Residuals')
plt.title('Residual Plot for Test Set')
plt.show()
```

The above residual plots show how far off our predictions are at each predicted value. The residuals for the training set are mostly contrained to -$20,000 to $40,000 off of the true base MSRP. The residuals for the test set are mostly constrained to around -$50,000 to $70,000 off from the true MSRP, with one very notable outlier being the Porsche 918.
