```
import pandas as pd
import numpy as np
from pathlib import Path
from ast import literal_eval
import json
import matplotlib.pyplot as plt
import seaborn as sns
from collections import Counter

import statsmodels.api as sm
from patsy import dmatrices
```

goodreads_reviews_spoilers was too large to be loaded in directly using pd.read_json(), so we used the json package to write a function that would load the dataset into a DataFrame iteratively. In the parse_large_json method, we loop through the dataset in chunks of 10000 (lines of data) at a time, and append each line in this chunk to a list that is later converted into a data table with pandas. This way, computational memory is not used up, since only some of the data occupies the memory at a given time.

```
# The code of this function is partially generated by ChatGPT from the prompt 
# "write a function that iteratively parses a large json file in chunks"
# chatGPT suggested using json decoder funcitonality from the json package;
# we referred to json documentation at https://docs.python.org/3/library/json.html to figure out
# the "chunking" while loop and how to open the json file without loading it in.
# we also wrote the helper method to store parsed lines from the file in list form.

def parse_large_json(json_file_path, process_data_callback):
    with open(json_file_path, 'r') as file:
        buffer = ""
        objects_processed = 0
        max_objects = sum(1 for line in open(json_file_path))

        while True:
            chunk = file.read(10000)  # Adjust the chunk size as needed

            if not chunk or (objects_processed >= max_objects):
                break  # Reached the end of the file

            buffer += chunk

            # Split buffer into lines (each line should be a JSON object)
            lines = buffer.split('\n')

            # Process complete JSON objects
            for line in lines[:-1]:
                try:
                    obj = json.loads(line)
                    process_data_callback(obj)
                    objects_processed += 1
                except json.JSONDecodeError as e:
                    print(f"JSONDecodeError: {e}")
                
                objects_processed = objects_processed + 1

            # Keep the last incomplete line in the buffer
            buffer = lines[-1]


parsed_data_list = []
#helper method to add data to list
def process_data(data):
    parsed_data_list.append(data)


parse_large_json('goodreads_reviews_spoiler_raw.json', process_data)
# Convert the list of parsed data to a Pandas DataFrame
reviews_df = pd.DataFrame(parsed_data_list)

reviews_df.head(10)
print(reviews_df.shape)
```

Using the function we defined during the last step, we can load in the dataset of book genres as well.

Goodreads allows users to select multiple genre labels for each book read, so we create a function to standardize these subgenre labels, mapping them to a more general genre category. Using this mapping, we create a function that assigns to each review a list of the general genres that its existing 'genre' variable (a dictionary of the user-assigned labels) contains the affiliated tags for. Labeling with a list of overarching genres accounts for books that might fit into multiple genre categories, such as fantasy young adult novels, or classical romance novels.

```
genre_mapping = {
    'nonfiction': ['biography', 'autobiography', 'memoir', 'history', 'self-help', 'scientific','non-fiction'],
    'literary fiction': ['literary','literary fiction','classic','poetry',"modern classic","play"],
    'science fiction': ['science fiction','sci-fi','dystopia'],
    'fantasy': ['fantasy', 'paranormal', 'magic', 'mythology'],
    'romance': ['romance', 'contemporary romance', 'historical romance'],
    'mystery/thriller': ['mystery', 'thriller', 'suspense'],
    'young adult': ['young adult', 'teen', 'YA', 'coming of age', 'young-adult']
}

genre_to_cat = {subgenre: category for category, subgenres in genre_mapping.items() for subgenre in subgenres}

def create_genre_list(row):
    genre_set = set()
    for genre_list in row.keys():
        genres = [genre.strip() for genre in genre_list.split(',')]
        genre_set.update(genres)
    genre_set = list(genre_set)
    genre_list = set()
    for genre in genre_set:
        cat = genre_to_cat.get(genre.lower(), None)
        if cat:
            genre_list.add(cat)
    return list(genre_list)


genre_df['genre_lists'] = genre_df['genres'].apply(create_genre_list)

reviews_merged = pd.merge(reviews_df, genre_df, on='book_id', how='left')
```

After defining the genres we're interested in, we merge the table of genres with the table of reviews by book id so that we have a comprehensive DataFrame of reviews, with a 'genre_lists' variable labeling the review's subject genres.

```
reviews_merged = reviews_merged.dropna()
reviews_merged.isna().sum()
reviews_merged.head()
```

### Urban Dictionary Slang

Now we read in and clean the dataset of slang terms.

```
slang_df = pd.read_csv('urbandict-word-defs.csv',on_bad_lines='skip')
slang_df.head(10)
print(slang_df.shape)
```

```
#check where and what na values there are
slang_df.isna().sum()
```

```
#drop na terms and check that they are dropped
slang_df = slang_df.dropna()
slang_df.isna().sum()
```

A visual scan of the urbandict-word-defs.csv file revealed that some entries are duplicated in Urban Dictionary: several rows shared a 'word' value. 

```
#drop the duplicates
slang_df = slang_df.drop_duplicates(subset=['word'])
slang_df.shape
```

Now that the dataset of slang terms is free of NaN values and duplicate words, we can use the 'words' Series to calculate the frequencies of slang terms in each review. Slang frequency is defined by the number of slang terms divided by the total number of words in the review; we wrote a function that would take in a review and return this frequency. We then apply this function to the review text and store the calculated frequencies in a new column called 'slang_freq' within our DataFrame.

```
slang_words = set(slang_df['word'].str.lower())
def calc_slang_freq(row):
    words = row.lower().split()
    total_words = len(words)
    slang_count = sum(word in slang_words for word in words)
    return slang_count / total_words if total_words > 0 else 0
reviews_merged['slang_freq'] = reviews_merged['review_text'].apply(calc_slang_freq)
reviews_merged.head()
```

With the frequency of slang usage and genre of the book being reviewed for each review added to our DataFrame, we are ready to begin EDA to understand the trends captured!

## Exploratory Data Analysis

### General Slang Distribution

First, we want to visualize the distribution of slang usage in digital reviews. Visualizing this will help us understand whether digital reviews commonly contain slang or not in general, without distinction by genre; it provides a good baseline for our understanding of language formality in online reviews.

```
plt.figure(figsize=(10,6))

plt.hist(reviews_merged['slang_freq'], bins=50, color='blue')
plt.title('Histogram of Slang Frequency in Reviews')
plt.xlabel('Slang Frequency')
plt.ylabel('# of Reviews')
plt.grid(alpha=0.75)
plt.show()
```

```
#the mean frequency is quite high
reviews_merged['slang_freq'].describe()
```

The distribution of slang frequency indicates that on average, book reviews (with no distinction across genres) employ a high proportion of informal language. On average, book reviews have a slang frequency of 0.8 -so most reviews or predominantly slang. To learn more about the correlation of genres with this skew, we can investigate which genre categories most reviews belong to next.

### Review Count by Genre

We also want to get a sense of the distribution of reviews across genres. Below, we graph a bar chart of reviews by genre, so we can understand which genres are most popularly reviewed. Knowing the breakdown of reviews by genre gives insight into the genres that may be most heavily influencing the shape of the slang frequency distribution plotted above.

```
exploded_genres = reviews_merged.explode('genre_lists')
exploded_genres = exploded_genres.reset_index(drop=True)

plt.figure(figsize=(10,6))

sns.countplot(data=exploded_genres, y='genre_lists')
plt.title('Count of Books in each Genre')
plt.xlabel('Genre')
plt.ylabel('# of Books')
plt.show()
```

Most reviews are for fantasy, young adult, mystery, and romance books - all of which are fiction genres. Because these are the predominant genres being reviewed, if there is a correlation between genre and slang frequency, it's possible that these are the genres causing the skew towards high slang usage.

It's also possible that reviews across the board (in all genre categories) just have high slang frequencies regardless of genre. We need to compare slang frequencies across genres to see if this is the case.

### Comparing Slang Frequencies across Genres

```
sns.boxplot(x="genre_lists", y="slang_freq", data=exploded_genres)
```

Each genre visibly has many outliers. The datasets we use are very large, and we are looking to understand general trends, so we can drop the outliers to improve readability and focus on the averages over the data.

```
slang_by_genre = sns.boxplot(x="genre_lists", y="slang_freq", data=exploded_genres, showfliers = False)
slang_by_genre.set_xticklabels(slang_by_genre.get_xticklabels(), rotation=45)
```

We see that for all major genres, slang frequency if relatively high, with medians floating above 0.8. There is no significant difference in the median frequency across genres. It appears that the quartile sections of the boxplots for nonfiction and literary fiction are shifted slightly lower, so these two genres may have reviews with slightly less slang. The difference, however, is not large or obvious.

The similarity across the boxplots points to the fact that across all genres, slang frequency is generally high. There are not any particular genres that are causing the slang frequency to tend towards a high value, or pulling the average lower. Hence, there may not be a correlation between genre and slang frequency.

Next, we will perform linear regression to confirm this, after reviewing whether or not user perceived quality of a book (which is a part of their experience and understanding of literary content) may play a role in slang frequency. 

### Comparing Slang Frequencies across Ratings

```
sns.boxplot(x="rating", y="slang_freq", data=reviews_merged, showfliers = False)
```

We see that the median slang frequency is relatively similar across ratings as well; the lack of significant difference between the plots for all ratings indicates that specific ratings, or perceptions of content quality, are causing the trend towards high slang usage. This points towards the potential lack of relation between rating and slang use.

Linear regression will assist with confirming whether or not  this is the case as well.

### Linear Regression between Genre and Slang Frequency

```
formula = 'slang_freq ~ C(genre_lists)'
y, X = dmatrices(formula, data=exploded_genres, return_type='dataframe')

# Initialize the model
model = sm.OLS(y, X)

# Fit the model
result = model.fit()

# Print the summary results
print(result.summary())
```

The model shows that genres have very little impact on the slang frequencies in the reviews. We can see that there is a positive correlation between mystery/thriller, romance, and young adult and the slang frequency in the reviews. There is also a slight negative correlation between literary fiction and nonfiction to slang frequency. The p-value for each genre is very low, suggesting that genres are a good predictor of the slang frequencies in the reviews. However, the correlation coefficient and the individual coefficients for the genres are very small relative to the scale of the dependent variable, which indicates that the negative and positive correlations we see are not strong. We also know that literary fiction and non-fiction are the genres for which there are the fewest reviews, so scarcity of data calls these correlations into question. Furthermore, the R-squared value is zero, which indicates that the independent variable, genres, does not predict changes in slang frequency at all. Thus, the specific genre does not indicate if the readers are more likely to use slang or not in their reviews.

### Linear Regression between Rating and Slang

```
formula = 'slang_freq ~ rating'
y, X = dmatrices(formula, data=reviews_merged, return_type='dataframe')

# Initialize the model
model = sm.OLS(y, X)

# Fit the model
result = model.fit()

# Print the summary results
print(result.summary())
```

The p-value suggests that rating is a good predictor of slang usage, but this is not entirely the case. The relationship between rating and slang may be a little more complex than linear, as the r squared value is close to 0. This means the rating is not able to account for the variation in the slang frequency data. The coefficient of rating is also very small, indicating that there is only a slight positive impact toward the slang frequency based on the rating.

# Error
In the wrangling step where genre lists for each book are merged with the table of reviews, there are multiple methods to decide the genre, and results may vary for each; we chose one method. For the genre standardization function, we manually encoded which Goodreads tags to include in our definition of each main genre. These tags were selected for simplicity and because they were the most intuitively and commonly used tags associated with each general genre. We also could have opted to label each book in the Goodreads genre dataset with only one main genre based on its tags; however, we decided to label them with a list of all the main genres identified from the associated tags to reflect the real world. This allowed for each book to have multiple genres which influenced the way we performed our analysis. We acknowledge that other interpretations and methods can lead to different outcomes.

Furthermore, as the initial univariable visualization of review counts for each genre showed, there were significantly fewer data points for the literary fiction genre than other genres; data scarcity may thus mean that the estimates of slang frequency for literary fiction book reviews is not truly generalizable (lacks external validity).

# Discussion and Conclusion

Among the background resources we've consulted for this project include the article "To Slang or Not To Slang: Defending authentic language in YA and children's literature" by Marva Hinton and a study by Anabelle Mahoney regarding the reader attitudes towards literature depending on expectations set by the genre of the work in question.Reflecting on these background sources, there are a few conclusions we can draw through intuition. "To Slang or Not To Slang: Defending authentic language in YA and children's literature" covers how slang language allows younger audiences to connect to the setting of works by being relatable. However, YA and children's literature isn't confined to any particular genre so while age may incline slang usage, there isn't an expectation that it'll reflect between genres. Additionally, concerning the study done by Mahoney, genre expectations isn't something that particularly influences reception and its language. Someone partaking in a particular book is likely a person who is already inclined towards the genre to begin with. Any exceptional reaction to a book built off the expectations the study mentions would be the consequence of the individual qualities of said book rather than be a product of the genre. The same can also be said about rating's effects on review language. 

With these in mind, let's take a look at the conclusions of our analysis: 

While there is a substantially greater quantity of reviews for what could be considered popular genres, being fantasy, young adult, 'mystery/thriller, and romance, the general distribution of slangs amongst reviews for all genres is roughly similar. Common and efficient modes to communicate intent, that is slang, will naturally become ingrained in speech and writing with the natural progression of language. In a non-formal setting where there is no requirement to write in any particular way, as such with reviews, the readers on GoodReads favor the style of writing that they're most accustomed and comfortable with, and in most cases, roughly 80% across the board, that includes the usage of slang language. While this isn't unexpected, it is to be noted that there isn't any particular incentive to deviate from this even for reviews of nonfiction genres, where the formality of the content being read does not reflect in the review being given. Ultimately, genre doesn't have particular bearing on whether or not a reviewer uses slang. Additionally, neither does the rating given, as enthusiatic reviewers, negative reviewers, indifferent reviewers, and all those in between are inclined to digitally write as they naturally do regardless of circumstance. 

Thus, our initial hypothesis that there is a correlation between the content readers consume and their language formality is suggested to be false. Neither the content, nor their appreciation of it changes patterns of informal language in review writing.
