# Abstract

Further statistical analysis, including the employment of a basic classification model and the Spearman correlation coefficient, indicated a weak relationship between high-end hardware specifications and game ratings. Specifically, the model's predictive accuracy was low, and a slight negative correlation was observed between GPU advancements and average playtime. These outcomes underscore the limited influence of hardware requirements on game success metrics such as ratings and engagement levels.

This study highlights the complexity of factors contributing to the success of PC games, challenging the notion that higher hardware requirements equate to better player experiences or higher ratings. While acknowledging limitations such as the simplification of success metrics and the use of basic analytical models, our research paves the way for future inquiries. Further investigations could expand upon the variables considered, employ more sophisticated modeling techniques, and explore the nuanced interplay between technical specifications and other determinants of game success.

## Steam Requirements

In order to reach an optimal dataset, we need to identify what is a good PC. To do that, we designate a high end PC to be of the following criteria:
- **Processor (CPU)**: i7, i9, Ryzen 7, Ryzen 9
- **RAM**: >= 4GB
- **Graphics (GPU)**: >= XX(70, 80, 90) (Super, TI), >= AMD 5500
- **Bonus**: SSD Req

We leave the reviews in the format of a string for easy regex parsing.

```python
# Data load
requirements = pd.read_csv('./data/store/steam_requirements_data.csv')
```

```python
# Classify high End PC
def high_end_pc(requirements_text: str) -> bool:
    """ Classifies a high pc if it meets one of the requirements """
    processor_pattern = r"Minimum:.+(i7|i9|Rython 7|Rython 9)" 
    ssd_pattern = r"Minimum:.+SSD"
    ram_pattern = r"Minimum: .+4GB"
    gpu_pattern = r"Minimum:.+(GTX 70|GTX 80|GTX 90|RTX 70|RTX 80|RTX 90|Super|TI|RX 5500|RX 5600|RX 5700|5800|8000|XT)"

    if re.search(processor_pattern, requirements_text): # High End Processor 
        return True
    elif re.search(ram_pattern, requirements_text): # High End RAM 
        return True
    elif re.search(gpu_pattern, requirements_text): # High End Graphics Card 
        return True
    elif re.search(ssd_pattern, requirements_text): # Bonus/Additional Req 
        return True
    return False

high_end_pcs = requirements['pc_requirements'].progress_apply(high_end_pc)
```

```python
# Percentage of high end pcs
high_end_pcs.sum()/high_end_pcs.count()
```

```python
# Assign data
requirements['high_end_pc'] = high_end_pcs
```

## Exploratory Data Analysis (EDA)

### Part 1: Filtering down to show statistically significant reviews

One of the main things that I am interseted in is the distribution of how many games require `high_end_pc`'s.

```python
# Visualize Percentage of games with High PC Reqs
high_end_pc_dist = master['high_end_pc'].value_counts()

# Create the pie chart
high_end_pc_dist.plot(kind='pie', title='Proportion of High End PC Reviews')
plt.xlabel('Percentage of High End PCs: {:.2f}'.format(high_end_pc_dist[True]/high_end_pc_dist[False]))
plt.legend(labels=['Low End PC', 'High End PC'], loc='best')
plt.show()
```

The distribution looks standard, and as expected. Only a small subset of games are expected to have a high end PC requirement. This leaves me the question, what types of games are most reviewed in this dataset? This will give us the metric for how popular the game is.

```python
# Top 20 Most Reviewed Games
game_counts = master['app_name'].value_counts()
game_counts[:20].to_frame()
```

Lots of big names here. `nan` appears to top the charts but that appears to be just be because of the lack of data in the name. We can navigate around this by using the `app_id` instead. This leads to the next step of finding out the scores for each game.

```python
# Calculating averages scores
scores = master.groupby('app_name')['review_score'].mean()
scores.to_frame()
```

Looks pretty messy and many of the games may have over-inflated scores due to the lack of many reviews for it. We should clean up the dataset to remove these **outliers**. Lets limit the dataset to only include games with more than 1,000 reviews.

```python
# Remove games with only a small amount of reviews
threshold_for_count_of_reviews = 1000
top = master['app_id'].value_counts()
validated_games = top[top > threshold_for_count_of_reviews]
filtered_master = master[master['app_id'].isin(validated_games.index)]

# Recalculate average scores
new_scores = filtered_master.groupby('app_name')['review_score'].mean()
new_scores.sort_values(ascending=False).to_frame()
```

Much better, only two games have the perfect score. Those two games at the top may be an outlier because having a perfect score at that many reviews is very rare. Let's take another look at the high end PC distribution to make sure it didn't change too much.

```python
# Taking another look of high_end_pc after filtering
high_end_pc_dist = filtered_master['high_end_pc'].value_counts()

# Create the pie chart
pie_chart = high_end_pc_dist.plot(kind='pie', title='Proportion of High End PC Reviews after Filter')
plt.xlabel('Percentage of High End PCs: {:.2f}'.format(high_end_pc_dist[True]/high_end_pc_dist[False]))
plt.legend(labels=['Low End PC', 'High End PC'], loc='best')
plt.show()
```

Looks good! The PC distribution is relatively the same and will not affect the future analysis that we will be doing on both metrics (Score and Popularity).

### Classification

```python
# Use Pre-calculated averages
def classifcation(high_end_pc: bool) -> int:
    return average_review_counts[high_end_pc]
```

```python
# Get X values
matched_requirements = requirements[requirements['steam_appid'].isin(review_counts_per_game.index)]
high_end_pc_list = matched_requirements['high_end_pc'].tolist()

# Make preds
preds = [classifcation(x) for x in high_end_pc_list]

# Calculate Errors
true_y = review_counts_per_game\
        [review_counts_per_game.index.isin(matched_requirements['steam_appid'].values)]\
        .sort_index().values
print("MSE: ", metrics.mean_squared_error(preds, true_y))
print("MAE: ", metrics.mean_absolute_error(preds, true_y))
print("R2: ", metrics.r2_score(preds, true_y))
```

Using a basic classifier which predicts using mean of the training set, it does not do well. If attempting to perform some sort of classification or regression to predict success, one would need more information than just whether it has high minimum PC requirements.

Based on Spearman Correlation Coefficient value, The observed slight negative correlation between average playtime and GPU model suggests a technical significance, implying that as GPU models improve, there may be a slight decrease in playtime.
