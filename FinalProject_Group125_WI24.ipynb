## Dataset: McDonald’s Global Store Locations

```python
# Define a function to parse the 'runhours' column if needed
def parse_runhours(runhours_str):
    # Example placeholder function to convert runhours to a tuple of opening and closing times
    # This simple split can be modified for more complex scenarios
    if pd.isna(runhours_str):
        return None, None
    times = runhours_str.split('-')
    if len(times) == 2:
        return times
    else:
        return runhours_str, runhours_str  # Assuming same opening and closing time if only one is provided

# Load the dataset
mcdonalds_df = pd.read_csv('./data/dataset1/McDonalds.csv')

# Number of observations:
print(mcdonalds_df.shape[0])

# Number of variables:
print(mcdonalds_df.shape[1])

mcdonalds_df.head()
```

The "McDonald's Global Store Locations" dataset is a collection of data points representing individual McDonald's restaurant locations worldwide. The dataset includes the store name, an identifier, country and subdivision information, city, exact address, postcode, contact telephone number, operating hours (run hours), and the geographical coordinates (latitude and longitude) of each location.

From the output, we can see a sample of the dataset focused on McDonald's locations in Andorra, revealing specific details such as the store's operational times ranging from 8:00 to 23:00, among others. The DataFrame is structured to facilitate analysis of McDonald's presence in relation to geographical and operational variables, potentially to explore patterns or insights about its global distribution. The output indicates that the dataset has been successfully read into a pandas DataFrame, and the head of the DataFrame has been displayed, confirming the correct structuring of the data with all relevant columns.

After successfully loading the dataset, we will proceed to clean the McDonald dataset so we can use the contents. We start by replacing 'nodata' placeholders with NaN for better handling of missing values. Next, we use a function to split the 'runhours' column into 'opening_time' and 'closing_time' to separate the operating hours into more usable data. Afterward, the original 'runhours' column is dropped as it's no longer needed. The 'services' column is also transformed; each service type is converted into its own binary column, indicating the presence or absence of that service. Finally, any rows missing essential data like 'latitude' or 'longitude' are removed, ensuring the dataset's completeness for geographical analysis.

```python
# Replace 'nodata' with NaN
mcdonalds_df.replace('nodata', pd.NA, inplace=True)

# Parse 'runhours' into separate columns for opening and closing times
mcdonalds_df[['opening_time', 'closing_time']] = mcdonalds_df.apply(
    lambda row: pd.Series(parse_runhours(row['runhours'])), axis=1
)

# Drop the original 'runhours' column as it's now redundant
mcdonalds_df.drop('runhours', axis=1, inplace=True)

# Split 'services' into separate binary columns
services_expanded = mcdonalds_df['services'].str.get_dummies(sep='/')
mcdonalds_df = pd.concat([mcdonalds_df, services_expanded], axis=1)

# Drop original 'services' column as it's now redundant
mcdonalds_df.drop('services', axis=1, inplace=True)

# Removing rows where critical information is missing
mcdonalds_df.dropna(subset=['latitude', 'longitude'], inplace=True)

mcdonalds_df.head()
```

The output shows the first few rows of the modified DataFrame. Placeholder 'nodata' strings are replaced with NaN, and the 'runhours' column has been removed. Additional binary columns for services have been added to the end of the DataFrame, extending its width to 228 columns. This structured format enables easier and more robust analysis of the data.

## Dataset: GDP per Country

Next, we read in our GDP dataset and name as GDP_df.

```python
gdp_df = pd.read_csv('./data/dataset2/GDPPerCountry.csv', encoding='Windows-1252')
gdp_df
```

After successfully loaded the dataset, we clean the GDP dataset through several steps, so we can use the contents from this dataset for further analysis. 

Our intention is to separate a single column containing multiple data points into distinct columns for country names, nominal GDP, population, and GDP per capita. The process includes stripping away unnecessary quotation marks and commas, and converting strings to integers where appropriate. This transformation is done through custom functions that are applied to each row of the dataset. The data is then sorted by GDP per capita to provide a meaningful order.

```python
#Since we only start with one mixed column, we need to separate them, which we do here with assign
gdp_df = gdp_df.assign(country = gdp_df['Country,"GDP(Nominal, 2022)",Population (2022),GDP Per Capita'])
gdp_df = gdp_df.assign(nominal_GDP = gdp_df['Country,"GDP(Nominal, 2022)",Population (2022),GDP Per Capita'])
gdp_df = gdp_df.assign(population = gdp_df['Country,"GDP(Nominal, 2022)",Population (2022),GDP Per Capita'])
gdp_df = gdp_df.assign(GDP_Per_Capita = gdp_df['Country,"GDP(Nominal, 2022)",Population (2022),GDP Per Capita'])

#clean the string, split it by commas, and return the first entry, which is the country
def country_cleaner(string):
    string = string.strip()
    string = string.replace('"', '')
    output = string.split(',')
    return output[0]

#strip and clean the string, replacing the double quotes and commas, and return the string as an integer
#for the nominal GDP
def nom_cleaner(string):
    string = string.strip()
    output = string.split(',"')
    string = output[1].replace('"','')
    string = string.replace(',', '')
    return int(string)


#strip and clean the string again, get the third value from the split and process that, then return
def pop_cleaner(string):
    string = string.strip()
    output = string.split(',"')
    string = output[2].replace('"','')
    string = string.replace(',', '')
    return int(string)


#values less than 1000 had different quotation and comma situations, so different cases were needed
#for processing
def capita_cleaner(string):
    string = string.strip()
    output = string.split(',"')
    if len(output) == 4:
        string = output[3].replace('"','')
        string = string.replace(',', '')
        return int(string)
    else:
        string = output[2].replace('"','')
        string = string.replace(',', '')
        return int(string) 
#apply the functions to each column
gdp_df['country'] = gdp_df['country'].apply(country_cleaner)
gdp_df['nominal_GDP'] = gdp_df['nominal_GDP'].apply(nom_cleaner)
gdp_df['population'] = gdp_df['population'].apply(pop_cleaner)
gdp_df['GDP_Per_Capita'] = gdp_df['GDP_Per_Capita'].apply(capita_cleaner)

#drop the mixed column we started with
gdp_df = gdp_df.drop(columns = ['Country,"GDP(Nominal, 2022)",Population (2022),GDP Per Capita'])

#sort by GDP per capita
gdp_df.sort_values(by = 'GDP_Per_Capita')
```

The output shows the DataFrame after cleaning, with countries listed alongside their nominal GDP, population, and calculated GDP per capita. Countries with a lower GDP per capita, like Burundi, Central African Republic, and Sierra Leone, are listed alongside wealthier nations like Qatar, Switzerland, and Luxembourg for comparison. The output is sorted in ascending order of GDP per capita, indicating a focus on understanding the economic standing of various countries. The DataFrame is neatly organized into 177 rows and 4 columns, allowing for efficient analysis of the economic data.

Then, we creating a new column in the gdp_df DataFrame called GDP_in_billions. This new column is calculated by dividing the nominal_GDP column by 1 billion (10^9) to convert the GDP figures from absolute values into billions, and then rounding the result to two decimal places for readability and convenience. 

```python
gdp_df = gdp_df.assign(GDP_in_billions = round(gdp_df['nominal_GDP']/1000000000, 2))
gdp_df
```

Now, we aim to consolidate data from two different datasets, McDonald's store locations and GDP information by country. The first portion of the code below counts the number of McDonald's stores in each country and creates a new DataFrame with this count. It then renames the columns to 'country' and 'store_count' for clarity. Next, this DataFrame is merged with the GDP data on the 'country' column to align the store counts with the corresponding GDP data for each country. Finally, we sorted the combined DataFrame alphabetically by country names. This process helps us to prepare the data for comparative analysis between the number of McDonald's stores and the GDP metrics for each country.

```python
# Summarize McDonald's data at the country level for 2023 using value_counts
mcdonalds_per_country = mcdonalds_df['country'].value_counts().reset_index()
mcdonalds_per_country.columns = ['country', 'store_count']  # Rename columns appropriately

# Merge with GDP data for 2022
combined_df = pd.merge(mcdonalds_per_country, gdp_df, how='inner', left_on='country', right_on='country')

# Sort the dataframe by 'country' column in ascending order
combined_df.sort_values('country', ascending=True)
```

The output displays a summary of the McDonald's store counts by country, merged with the GDP data. It lists several countries along with the number of McDonald's stores (`store_count`), the nominal GDP in absolute terms (`nominal_GDP`), the population of each country (`population`), the GDP per capita (`GDP_Per_Capita`), and the nominal GDP expressed in billions (`GDP_in_billions`). This merged and sorted data allows us to do a comparative analysis of the relationship between the presence of McDonald's stores and various economic indicators within each country in the upcoming section.

### The OLS Between Store Counts, GDP, Population, and GDP per Capita

Building on the insights gleaned from our initial analysis, we now venture into a more nuanced exploration of the data. In the following section, we delve into the interdependencies between these variables to decipher their collective influence on the proliferation of McDonald’s stores. By introducing interaction terms into our regression model, we aim to uncover the subtleties of how these economic factors not only contribute individually but also how they interplay and compound each other's effects. This approach equips us to unravel the complexities that a mere univariate analysis might overlook. The ensuing Ordinary Least Squares (OLS) regression will thus enhance our understanding of the dynamics at play, providing a more robust framework to examine whether the number of McDonald's stores can indeed serve as an economic indicator of a country's wealth and economic stature. The code snippet below operationalizes this analytical step, preparing us to interpret the results in the context of our overarching research objective.

```python
# Create interaction terms
combined_df['GDP_Population'] = combined_df['nominal_GDP'] * combined_df['population']
combined_df['GDP_GDP_Per_Capita'] = combined_df['nominal_GDP'] * combined_df['GDP_Per_Capita']
combined_df['Population_GDP_Per_Capita'] = combined_df['population'] * combined_df['GDP_Per_Capita']

# Define the formula including interaction terms
formula = 'store_count ~ nominal_GDP + population + GDP_Per_Capita + \
           GDP_Population + GDP_GDP_Per_Capita + Population_GDP_Per_Capita'

# Fit the OLS model
model = ols(formula, data=combined_df).fit()

# Print the summary of the regression model
print(model.summary())
```

The regression analysis presents an intricate understanding of how the number of McDonald’s restaurants is associated with a nation's wealth indicators, and the results are compelling. An R-squared value of 0.991 indicates an exceptionally strong model, accounting for nearly all the variability in the number of McDonald's stores across countries. The significance of the overall model is confirmed by an essentially zero p-value for the F-statistic, underscoring the collective influence of the predictors used. Among the predictors, nominal GDP emerges as a positive and significant contributor, suggesting that as the GDP of a country increases, the presence of McDonald’s stores tends to rise accordingly. Surprisingly, GDP per capita alone does not show statistical significance as a predictor, indicating that it might not independently influence the number of stores as one might expect.

However, the story unfolds further with the significant interaction terms. The GDP_GDP_Per_Capita and Population_GDP_Per_Capita interactions are particularly revealing, suggesting that the effect of GDP per capita on the number of McDonald’s stores is nuanced, depending on the broader economic context shaped by the overall GDP and population. Essentially, these interactions indicate that a country's wealth per capita is more likely to reflect in the number of stores when viewed through the lens of the country's total economic output and its population dynamics. Nevertheless, the high condition number noted in the regression output cautions us about potential multicollinearity, where independent variables are highly interrelated, which can obscure the impact of individual predictors. In essence, while the analysis does support the notion that McDonald's stores could mirror economic vitality, this reflection is complex and multifaceted. It is not the GDP per capita alone but its interplay with other economic factors that shapes the McDonald’s store count. Hence, while indicative, the number of stores should be interpreted as part of a broader economic narrative rather than a standalone metric of economic well-being.
