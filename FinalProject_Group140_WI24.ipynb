# Abstract

Climate change and deforestation are recognized as significant contributors to the increasing frequency and severity of natural disasters worldwide. This study investigates the extent to which temperature fluctuations, deforestation rates, and a country's level of economic development, as measured by GDP, influence susceptibility to environmentally-induced humanitarian disasters in South American countries during the 21st century. Drawing upon datasets including Global Land Temperatures by Country, Tree Cover Loss By Region, WorldRiskIndex, and Gross Domestic Product by Countries, our analysis aims to elucidate the interplay between these variables. Our hypothesis posits that countries experiencing higher temperatures and greater deforestation rates will exhibit a heightened susceptibility to disasters, compounded by economic disparities. By employing statistical techniques and data integration methods, we seek to provide insights into the complex dynamics underlying disaster risk, thereby informing policymakers and stakeholders for more effective disaster mitigation and resilience strategies.

# Research Question

To what extent do temperature fluctuations and deforestation rates, along with a country's level of development as measured by GDP, affect susceptibility to environmentally-induced humanitarian disasters in South American countries in the 21st century?

## Background and Prior Work

Given the increase in frequency and severity of natural disasters in the past several decades caused in part by climate change and deforestation, we are attempting to analyze if these two variables have an outsized impact on the susceptibility of a country to natural disaster, using WorldRiskIndex (WRI), which shows how susceptible a country is to suffering a natural disaster.<a name="cite_ref-1"></a>[<sup>1</sup>](#cite_note-1)

Climate change has been and continues to be an ongoing global crisis. According to the United States Geological Survey (USGS), rise in global surface temperatures will lead to higher rates of atmospheric water evaporation, fuelling increasing rates and severities of droughts, floods, and tropical storms.<a name="cite_ref-2"></a>[<sup>2</sup>](#cite_note-2)

Moreover, reports by the World Resources Institution, point out that climate change is directly contributing to humanitarian emergencies from heatwaves, wildfires, floods, tropical storms and hurricanes and they are increasing in scale, frequency and intensity.<a name="cite_ref-3"></a>[<sup>3</sup>](#cite_note-3) Research shows that 3.6 billion people already live in areas highly susceptible to climate change.

Deforestation poses similar compounding risks. The Global Forest Watch has been tracking global forest tree cover loss and has found that between 2000 and 2020 there was a net change of -2.4% in forest cover on Earth.<a name="cite_ref-4"></a>[<sup>4</sup>](#cite_note-4) Institutions, such as the World Health Organization, have been studying the imminent effects that climate change and deforestation have on different regions of the world.<a name="cite_ref-5"></a>[<sup>5</sup>](#cite_note-5) Effects of forest cover lead to changes in precipitation, but it has been shown that globally, its impact is mostly related to the surface temperature change. 

According to The New Humanitarian, in addition to deforestation through logging being a large source of greenhouse gas emissions, deforestation leads to loss of watersheds which protect environments from erosion, loss of carbon sinks which slow down climate change, and loss of natural resources which sustain communities and help them recover in the event of a disaster.<a name="cite_ref-6"></a>[<sup>6</sup>](#cite_note-6)

While economically developed nations like the U.S. are the worlds biggest drivers of climate change through carbon emissions, the countries most at risk for disaster and harm tend to be the worlds poorest nations.<a name="cite_ref-7"></a>[<sup>6</sup>](#cite_note-7) Richer nations already have strong infrastructure that proctects against certain kinds of climate emergencies, as well as the capital to fund prevention projects and recovery efforts.<a name="cite_ref-8"></a>[<sup>6</sup>](#cite_note-8) Additionally, rich countries often financially benefit from the very same activities, like industry and mining, that fuel disaster in less wealthy countries. Therefore, economic development, as measured by GDP, has the capacity to provide strong insights into a particular country's risk of disaster and ability to recover quickly.

# Hypothesis


We hypothesize that countries experiencing both greater average temperature increases due to climate change and higher deforestation rates are associated with a higher overall disaster risk score in the WorldRiskIndex. We predict this because of decades of research pointing to climate change and rising temperatures as one of the main driving factors of increased disaster occurrence. Additionally, we predict deforestation as a major grouped causal factor due to the close interdependent relationship between climate change and deforestation, as well as the compounding negative effects of loss of forest cover leading to habitat erosion and resource depletion. South America, as a region with one of the most important and most threatened forest areas, the Amazon, is particularly at risk for exploitative deforestation, leading to catastrophic domino disasters, as ground stability, air quality, and biodiversity all stand to suffer greatly. We also predict that economic disparity will be a significant compounding factor, increasing the disaster risk score, as lower economic stability decreases a country's ability to quickly and approriately respond to disasters as well as engage in preventative measures. ### Initial Data Preprocessing

For consistency across our datasets, we create a function to convert alpha-3 country codes to their full country names:

```python
def code_to_country_name(code):
    try:
        country_name = pycountry.countries.get(alpha_3=code).name
    except AttributeError:
        country_name = 'Unknown'
    else: # Bolivia and Venezuela are named differently in this library so we treat them separately
        if code == 'VEN':
            country_name = 'Venezuela'
        elif code == 'BOL':
            country_name = 'Bolivia'
    return country_name

# Apply the function to the new DataFrame column
trees_df['Country'] = trees_df['CountryCode'].apply(code_to_country_name)

# Verifying the conversion
trees_df['Country'].unique()
```

Renaming the column of interest for clarity and keeping only the relevant information:

```python
trees_df.rename(columns={'TreeCoverLoss_ha': 'TreeCoverLoss'}, inplace=True)

trees_df = trees_df[['Country', 'Year', 'TreeCoverLoss']]
trees_df
```

## Preparing the Final Dataset 

### Temperature Data

First, we rename the temperature column name in the temperature dataset for South America since we calculate the annual average temperature for each country to achieve consistent format in the final DataFrame:

```python
sa_temps.rename(columns={'Temperature': 'AverageTemperature'}, inplace=True)
average_temps = sa_temps.groupby(['Country', 'Year'])['AverageTemperature'].mean().reset_index()
```

For the GDP dataset, we transform it to a long format to align with other datasets and ensure the 'Year' column is of type integer:

We merge the datasets in stages based on the country and year columns, starting with tree cover loss and temperature data, followed by GDP data, and finally incorporating the WRI dataset:

```python
first_merge = pd.merge(sa_trees, average_temps, on=['Country', 'Year'], how='outer')

second_merge = pd.merge(first_merge, gdps_long_df, on=['Country', 'Year'], how='outer')

final_combined_df = pd.merge(second_merge, sa_wris, on=['Country', 'Year'], how='left')

# Rearranging order of columns
final_df = final_combined_df[['Country', 'Year', 'WRI', 'TreeCoverLoss', 'AverageTemperature', 'GDP']] 
```

## Analysis #2: Exploring Variable Correlations with Heatmaps

Now onto how variables in our final processed dataset are related to each other. Do they have a positive correlation? Or a negative one? Are they even related?

```python
#2. Heatmaps: seeing if there might be any realtionship between variables
#2.a) AverageTemperature vs TreeCoverLoss
#2.b) TreeCoverLoss vs WRI
#2.c) TreeCoverLoss vs GDP

# Define the pairs of variables for the heatmaps
pairs = [
    ('AverageTemperature', 'TreeCoverLoss'),
    ('TreeCoverLoss', 'WRI'),
    ('TreeCoverLoss', 'GDP')
]

# for loop for each country
for country, group in grouped_by_country:
    
    fig, axes = plt.subplots(1, len(pairs), figsize=(15, 5))
    
    # iterate over each pair and create a heatmap
    for i, (var1, var2) in enumerate(pairs):
        # correlation matrix
        corr_matrix = group[[var1, var2]].corr()
        
        
        sns.heatmap(corr_matrix, annot=True, cmap='viridis', ax=axes[i], vmin=-1, vmax=1)
        axes[i].set_title(f'{var1} vs {var2} - {country}')
    
    plt.tight_layout()
    plt.show()
```

While there are some countries for which there exists a little negative correlation between TreeCoverLoss and other variables (such as Argentina, where there is almost no correlation between TreeCoverLoss and GDP), there are also countries such as Bolivia for which this correlation is positive and shows a larger relationship between TreeCoverLoss and GDP. This makes us wonder whether maybe Brazil, the country with the largest tree cover area of them all due to housing the majority of the Amazon Rainforest, would be facing a similar problem. 

By checking Brazil's heatmap we can see this correlation we just discussed is not only smaller for Tree Cover Loss with WRI Rankings and Average Temperature, but actually negative for GDP, which is an unexpected result for us based on the preliminary visual observations we took from our plots. Nevertheless, Suriname, Peru, Paraguay, Uruguay, Bolivia and Guyana show a positive correlation larger than 0.4 for TreeCoverLoss and GDP, so given these are almost half of the countries in the continent, we might be able to think there is a positive correlation between GDP and Tree Cover Loss, though one likely confounded by other factors.
In fact, all South American countries share a positive correlation between GDP and TreeCoverLoss, having as an excpetion only Brazil (with a negative correlation) and Argentina (almost no correlation).

This made us think, will we then have a relationship between this same TreeCoverLoss and WRI? It doesn't seem like we will have one, but we're still looing into them. ## Analysis #3: Looking into Outliers with Boxplots

Outliers, deviations from the typical pattern of a dataset, can significantly impact the analysis and interpretation of data. They may skew statistical measures, affect the accuracy of correlations, and distort conclusions drawn from the data. In this analysis, we examine the presence of outliers in our dataset comprising variables such as WRI, Tree Cover Loss, and GDP across South American countries. By identifying outliers and assessing their impact, we aim to determine whether they should be retained in our analysis or treated as anomalous data points.

```python
#Boxplot all variables of final dataset to identify outliers

#for loop to iterate over each country
for country, group in grouped_by_country:
    
    fig, axes = plt.subplots(1, len(group.columns) - 2, figsize=(15, 5))
    
    
    columns_to_plot = [col for col in group.columns if col not in ['Year', 'Country']]
    
    #for loop for each column
    for i, column in enumerate(columns_to_plot):
        sns.boxplot(y=group[column], ax=axes[i])
        axes[i].set_title(f'{column} - {country}')
    
    plt.tight_layout()
    plt.show()
```

The boxplots reveal variations in the distributions of WRI, Tree Cover Loss, and GDP across countries, while average temperatures appear relatively consistent. Despite observing differences in distribution shapes, the presence of outliers is minimal across variables. We can see several examples of extreme outliers in WRI and TreeCoverLoss for Bolivia and Venezuela, but the majority of variables for other countries do not show outliers. This suggests that extreme values are not pervasive within our dataset. However, these slight deviations in distribution could influence the correlations observed between variables, as indicated in our previous heatmap analysis. Nonetheless, considering the dynamic nature of the variables under examination and the potential long-term impacts of climate trends, we have opted not to remove outliers from our dataset. Removing outliers could lead to a loss of valuable information and may result in findings that do not accurately reflect real-world conditions. Therefore, we acknowledge the presence of outliers but choose to retain them in our analysis to maintain the integrity and reliability of our results. This decision ensures that our findings remain robust and applicable to the complex dynamics of environmental susceptibility in South American countries.
