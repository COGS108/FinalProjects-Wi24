### Hypothesis Test

To provide a rigorous statistical foundation for our research question, we utilize hypothesis testing. This approach allows us to evaluate the likelihood of observing the data we have if our null hypothesis (no significant difference in performance metrics across regions) were true.

**Why a Hypothesis Test?**
- **Objectivity**: Hypothesis testing gives us a systematic way to test our assumptions against observable data, reducing the influence of bias.
- **Quantitative Measure**: It offers a p-value, a quantitative measure of the strength of evidence against the null hypothesis.
- **Control of Error Rates**: By setting a significance level (commonly 0.05), we control the Type I error rate, the probability of incorrectly rejecting a true null hypothesis.
- **Comparison Across Groups**: Given multiple regions and years, hypothesis testing, especially ANOVA and subsequent Tukey HSD tests, allows us to compare means across these groups simultaneously.
- **Insights into Specific Differences**: When we find significant results, post-hoc tests like Tukey HSD help us pinpoint exactly which regions' player performance metrics differ.

By applying this method, we aim to discern not just apparent or anecdotal differences, but statistically significant trends in player performance across regions, thereby answering our research question with a high level of confidence.


**Observation:** The model's accuracy in predicting players' regions based on their performance metrics is not substantially better than random guessing.

**Null Hypothesis (H0):** There is no significant difference in the individual player performance metrics (KD, KDA, ACS/Map) across the different Valorant regions (Americas, EMEA, Pacific, China) over the years.

**Alternate Hypothesis (H1):** There is a significant difference in the individual player performance metrics across the different Valorant regions over the years.

**Test Statistic:** The F-statistic from the ANOVA test comparing player performance metrics across regions.

**Significance Level:** The hypothesis test is conducted at the 0.05 significance level.

**P-value:** The p-value obtained from the ANOVA test for KD in 2023 is 0.0052 showing a statistical difference in the EMEA and Pacific, and for KDA in 2023, it is 0.0115, indicating a statistically significant difference in performance metrics for these years, while other metrics and years showed no significant difference. For the KDS in 2023, the difference is in between Americas and Pacific and EMEA and Pacific.


```python
metrics = ['KD', 'KDA', 'ACS/Map']
years = valorant['Year'].unique()
regions = valorant['Region'].unique()

results = []

for metric in metrics:
    for year in years:
        print(f"\nTesting {metric} in {year}:")
        subset = valorant[valorant['Year'] == year][['Region', metric]].dropna()
        
        if subset.empty:
            continue
        
        f_val, p_val = stats.f_oneway(*(subset[subset['Region'] == region][metric] for region in regions))
        
        if p_val < 0.05: 
            print(f"Significant difference found for {metric} among regions in {year} (p={p_val:.4f})")
            results.append((metric, year, p_val))
            
            tukey = pairwise_tukeyhsd(endog=subset[metric], groups=subset['Region'], alpha=0.05)
            print(tukey)
        
            
        else:
            print(f"No significant difference found for {metric} among regions in {year} (p={p_val:.4f})")
```

For the significant tests, a Tukey HSD post-hoc test was conducted to determine which regions have statistically different performance metrics. The test results, particularly for 2023, show that some regions differ from each other. For example, the EMEA and Americas region has a statistically significant difference in metrics from the Pacific region, which could suggest that the level of play has evolved differently in these regions.

```python
tukey_summary = pd.DataFrame(data=tukey.summary().data[1:], columns=tukey.summary().data[0])


tukey_summary['range'] = tukey_summary['upper'] - tukey_summary['lower']

plt.figure(figsize=(12, 8))
plt.hlines(y=np.arange(tukey_summary.shape[0]), xmin=tukey_summary['lower'], xmax=tukey_summary['upper'], color='skyblue')
plt.plot(tukey_summary['meandiff'], np.arange(tukey_summary.shape[0]), 'o', color='red')
plt.yticks(np.arange(tukey_summary.shape[0]), tukey_summary['group1'] + ' vs. ' + tukey_summary['group2'])
plt.title('Pairwise Comparison (Tukey HSD Test Results)')
plt.xlabel('Mean Difference (with 95% CI)')
plt.grid(axis='x')

plt.show()
```

The plot above displays Tukey's Honest Significant Difference (HSD) test results, a post-hoc analysis used after an ANOVA to assess which specific group means (in this case, Valorant player performance metrics by region) are different. The test examines pairwise comparisons between regions' performance metrics:

- Horizontal lines represent the 95% confidence intervals (CI) for the mean differences between regions.
- Red dots denote the actual mean differences observed in the data.
- The vertical line at zero is the point of no effect, where a mean difference of zero implies no difference in metrics.

**Interpretation in the context of our research question:**
- The confidence intervals for comparisons between China vs. Pacific, Americas vs. China, China vs. EMEA, and Americas vs. EMEA all cross the zero line, indicating there are no statistically significant differences in player performance metrics between these regions.
- For Americas vs. Pacific and EMEA vs. Pacific, the confidence intervals **do not cross the zero line**, signifying **significant differences** in performance metrics. These results point to a deviation in the Pacific region's performance when compared to China and EMEA.

These findings suggest that, with the exception of the Pacific region, there's a relative consistency in player performance metrics between the other regions. This could imply that player performances across Americas, China, and EMEA have leveled over time, aligning with the hypothesis that regional disparities are diminishing.


**Conclusions from the Hypothesis Tests:**
- **ANOVA** results predominantly indicate no significant differences in player performance metrics across regions for **most** years, suggesting a trend towards uniformity in player performance.
- The **Tukey HSD** analysis reveals that significant differences do exist, particularly with the Pacific region exhibiting slightly worse performance metrics compared to other regions in certain years.

**Implications:**
These findings imply that while the player performance metrics across the Americas, China, and EMEA regions have largely leveled, the Pacific region has retained distinct characteristics, specifically a marginally lower performance level. This nuanced perspective underscores the importance of regional considerations in the analysis of player performance evolution in Valorant, leading to the conclusion that regional performance homogenization is prevalent but not absolute, with the Pacific region standing out in the latest seasons.


### Multiclass Classification

Now, our research question asks to see if individual player performance has changed over time to be more leveled throughout the regions. To test this we can create a multiclass classification model. How this model helps us is that, if a properly tuned model can accurately label the region for each player given their individual player performance, then the metrics show a clear disparity between player performance in certain regions to other regions. However, if a classifier is not able to accurately classify a players metric to their region, it just goes to show that the player performance metrics are not that different between regions than they once were. Obviously, there could be other reasons that the gap has changed like changes in the game. This would be a great test to see if player performance has actually become more level.

We select our metrics `KD`, `KDA`, `ACS/Map` as the our features for our model as it is the main three parametere for individual player performance. We then split our data and get ready to make our model.

```python
X_label = valorant[['KD', 'KDA', 'ACS/Map']]
label_encoder = LabelEncoder()
y_label = label_encoder.fit_transform(valorant['Region'])

X_train, X_test, y_train, y_test = train_test_split(X_label, y_label, test_size=0.2)
```

To preprocess the model, we standardize all the feeatues so there is a balance. Then we create a pipeline for a list of models. To get a baseline result we are trying 13 models below to see which one initially can get us the highest cross validation score. Then we will move to tune that model up individually. This well help us pick the best model to use to classify players metrics into regions.

```python
preproc = ColumnTransformer(
    transformers=[
        ('scale', StandardScaler(), ['KD', 'KDA', 'ACS/Map']) 
    ], remainder='drop'
)

pipeline = Pipeline([
    ('preprocessor', preproc),
    ('classifier', None)  
])


models = [
    RandomForestClassifier(),
    LogisticRegression(),
    SVC(),
    KNeighborsClassifier(),
    GradientBoostingClassifier(),
    DecisionTreeClassifier(),
    AdaBoostClassifier(),
    GaussianNB(),
    ExtraTreesClassifier(),
    XGBClassifier(eval_metric='logloss'),
    LGBMClassifier(verbosity=-1),
    CatBoostClassifier(verbose=0),  
    SVC(kernel='rbf')
]

for model in models:
    pipeline.set_params(classifier=model)  
    scores = cross_val_score(pipeline, X_label, y_label, cv=5) 
    model_name = model.__class__.__name__
    if hasattr(model, 'kernel'): 
        model_name += f" with {model.kernel} kernel"
    print(f"{model_name}: Average CV score = {np.mean(scores)}")
```

After baseline testing for each of these models, we can see that Logistic Regression does th best job of classifying performance metrics into regions. Lets see if we can hypertune this model up to get its cross val score higher. Right now, the score of 0.396 is not great for a model. This suggests that player perormance metrics are much closer and there is not that big of a skill gap from region to region to differentiate players. However, we will do more analysis within logistic regression to get a better understanding. 

The `GridSearchCV` process optimizes hyperparameters for a Logistic Regression model within a pipeline, using accuracy as the evaluation metric. It systematically tests combinations of parameters (`C`, `penalty`, `solver`) across a defined grid, employing 5-fold cross-validation to assess performance. This exhaustive search ensures the selection of optimal hyperparameters, aiming to enhance the model's prediction accuracy. The outcome is a model fine-tuned to our specific dataset, demonstrated by `grid_search.best_params_` and `grid_search.best_score_`, which reveal the most effective settings and their corresponding accuracy.

```python
from sklearn.model_selection import GridSearchCV

pipeline = Pipeline([
    ('preprocessor', preproc),
    ('classifier', LogisticRegression(max_iter=1000)) 
])

hyperparameters = {
    'classifier__C': [0.001, 0.01, 0.1, 1, 10, 100],
    'classifier__penalty': ['l1', 'l2'],
    'classifier__solver': ['liblinear', 'saga'] 
}

grid_search = GridSearchCV(pipeline, hyperparameters, cv=5, scoring='accuracy', verbose=1)

grid_search.fit(X_label, y_label)
print(grid_search.best_params_)
print(grid_search.best_score_)
```

Below, our code defines a machine learning pipeline that standardizes features and applies logistic regression for classification. The logistic regression uses an L2 regularization penalty with a very small `C` parameter value, indicating strong regularization, and the `liblinear` solver. After fitting the model to the training data and predicting on the test set, we generate a classification report , which indicates low performance across different classes, with particularly poor results for the minority class (class 1). The `support` column reflects the class imbalance present in the dataset.
