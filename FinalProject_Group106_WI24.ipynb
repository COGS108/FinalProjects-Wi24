### Data Cleaning
1. Cleaning the `Rainfall in San Diego` dataset :

The San Diego rainfall dataset is cleaned and transformed by narrowing the focus to specific rows, renaming columns for clarity, and replacing **T** rainfall amounts with zeros. 

```python
sd_rainfall = sd_rainfall.iloc[2:26]
sd_rainfall.columns = ['Year', 'Jan', 'Feb', 'Mar', 'Apr', 'May', 'Jun', 'Jul', 'Aug', 'Sep', 'Oct', 'Nov', 'Dec', 'Annual']
sd_rainfall.replace('T', 0, inplace=True)
```

Then we reshape the data from a wide format to a long format where each row represents a month's rainfall for a given year, and maps month names to numeric values for easier sorting.

```python
# Convert the dataset from wide format -> long format
sd_rainfall_rearrange = pd.melt(sd_rainfall, id_vars=['Year'], var_name='Month', value_name='Precipitation (inches)')

# Map month names to numeric
month_to_num = {'Jan': 1, 'Feb': 2, 'Mar': 3, 'Apr': 4, 'May': 5, 'Jun': 6, 'Jul': 7, 'Aug': 8, 'Sep': 9, 'Oct': 10, 'Nov': 11, 'Dec': 12}
sd_rainfall_rearrange['Month'] = sd_rainfall_rearrange['Month'].map(month_to_num)
```

 The dataset is sorted chronologically by **Year** and **Month**, with the **Month** column converted to integers and missing values filled with zeros to ensure completeness.

```python
# Sort the dataset by year and month
sd_rainfall_rearrange = sd_rainfall_rearrange.sort_values(by=['Year', 'Month'])

# Convert 'Month' to integer and handle any missing values by filling with 0
sd_rainfall_rearrange['Month'] = sd_rainfall_rearrange['Month'].fillna(0)
sd_rainfall_rearrange['Month'] = sd_rainfall_rearrange['Month'].astype(int)

# Ensure there are no rows with missing month values.
sd_rainfall_rearrange = sd_rainfall_rearrange[sd_rainfall_rearrange['Month'].notna()]
```

Finally, the index is reset for a tidy, sequential arrangement, preparing the dataset for subsequent analysis or processing.

```python
# Reset the dataframe index
sd_rainfall_rearrange = sd_rainfall_rearrange.reset_index(drop=True)
sd_rainfall_rearrange.head()
```

2. Cleaning the `Temperature in West` dataset:

We clean the temperature data by extracting year and month from a **Date** column formatted as YYYYMM, converting these into separate columns for clarity and numerical analysis.

```python
# Divide the 'Date' column by 100 and take the integer quotient to extract the year
west_temp['Year'] = west_temp['Date'].astype(str).str[:4] 

# Use the modulo operation on the 'Date' column to extract the month
west_temp['Month'] = west_temp['Date'].astype(str).str[4:6] 
west_temp['Month'] = west_temp['Month'].astype(int)
```

The temperature values initially labeled **Value**, are renamed to **Temperature(°F)** to clearly indicate the measurement in Fahrenheit.

```python
# Rename the 'Value' column to 'Temperature(F)' 
west_temp = west_temp.rename(columns={'Value': 'Temperature(°F)'})
```

 Then we reorganize the dataset to include only these newly created **Year** and **Month** columns along with the temperature readings, streamlining the data for easier analysis and interpretation.

```python
# Select the final columns in the cleaned dataset
west_temp = west_temp[['Year', 'Month', 'Temperature(°F)']]
west_temp.head()
```

3. Cleaning the `Erosion` dataset:

We process the erosion dataset by refining its structure: renaming columns for clarity, consolidating date and time data into datetime objects for two separate surveys and dropping unnecessary information.

```python
# Set new column names for clarify.
erosion.columns = [
    'Time', 'Year1', 'Month1', 'Day1', 'Hour1', 'Year2', 'Month2', 'Day2', 'Hour2', 
    'Interval_length_days', 'Erosion_Negative_Change_m3', 'Lower_cliff', 'Upper_cliff', 
    'Erosion_Positive_Change_m3', 'Duration_hours', 'Wave_Height_m', 
    'Sum_Squared_Wave_Impact_Height_m2', 'Rainfall_cm'
]

# Create a datetime object representing the survey's date and time
for survey in ['1', '2']:
    date_cols = [f'Year{survey}', f'Month{survey}', f'Day{survey}', f'Hour{survey}']
    # Convert Year, Month, Day, Hour columns to a datetime object for each survey
    erosion[f'Survey{survey}_Date'] = pd.to_datetime(
        erosion[date_cols].astype(int).rename(columns=lambda x: x[:-1])
    )

# List columns that are not needed for the analysis
columns_to_drop = [
    'Time', 'Year1', 'Month1', 'Day1', 'Hour1', 'Year2', 'Month2', 'Day2', 'Hour2', 
    'Lower_cliff', 'Duration_hours', 'Wave_Height_m', 
    'Sum_Squared_Wave_Impact_Height_m2', 'Interval_length_days', 'Rainfall_cm',
    'Erosion_Negative_Change_m3', 'Erosion_Positive_Change_m3'
]

# Drop the listed columns
erosion = erosion.drop(columns=columns_to_drop) 
```

We focus on the latest survey dates to analyze erosion patterns, specifically concerning upper cliff erosion, by dropping extraneous columns and summarizing the data into a more manageable form. The dataset is then grouped by **Year** and **Month**, with erosion measurements aggregated to reveal monthly trends. 

```python
# Determine the latest date between the two surveys for each record
erosion['Max_Date'] = erosion[['Survey1_Date', 'Survey2_Date']].max(axis=1).dt.to_period('M').dt.to_timestamp()
# Extract the year and month from the latest date
erosion['Year'] = erosion['Max_Date'].dt.year
erosion['Month'] = erosion['Max_Date'].dt.month

# Rename 'Upper_cliff' to 'Upper_Cliff_Erosion'
erosion = erosion.rename(columns={'Upper_cliff': 'Upper_Cliff_Erosion'})

# Select the final columns in the cleaned dataset
final_columns = ['Year', 'Month', 'Upper_Cliff_Erosion']
cleaned_erosion_data = erosion[final_columns]

# Group by 'Year' and 'Month' and aggregate 'Upper_Cliff_Erosion' values by summing them
erosion_monthly = cleaned_erosion_data.groupby(['Year', 'Month']).agg({'Upper_Cliff_Erosion': 'sum'}).reset_index()
```

We also define an **evaluate_expression()** function to handle the negative signs in the data, ensuring accurate erosion calculations. The final output only contains a concise monthly overview of upper cliff erosion, facilitating our analysis of temporal erosion trends.

```python
# Define a function to deal with '-' sign
def evaluate_expression(expr):
    # Split the expression on hyphens and filter out empty strings
    numbers = filter(None, expr.split('-'))
    # Convert the string numbers to integers and take their sum
    return sum(-int(num) for num in numbers)

erosion_monthly['Upper_Cliff_Erosion'] = erosion_monthly['Upper_Cliff_Erosion'].apply(evaluate_expression)

erosion_monthly.head()
```

4. Merging all the datasets to a combination dataset on **Year** and **Month**:

Since we cleaned up all the datasets above and filtered them by **Year** and **Month**, we merge rainfall, temperature, and erosion datasets into a single unified dataset by first ensuring the **Year** and **Month** columns across all datasets are of integer type for consistency.

```python
# Ensure 'Year' and 'Month' columns in all datasets are integer type.
sd_rainfall_rearrange['Year'] = sd_rainfall_rearrange['Year'].astype(int)
sd_rainfall_rearrange['Month'] = sd_rainfall_rearrange['Month'].astype(int)
west_temp['Year'] = west_temp['Year'].astype(int)
west_temp['Month'] = west_temp['Month'].astype(int)
erosion_monthly['Year'] = erosion_monthly['Year'].astype(int)
erosion_monthly['Month'] = erosion_monthly['Month'].astype(int)
```

 We begin by merging the rainfall and temperature data, then proceed to incorporate the erosion data, both steps using outer joins to capture all data points, even when there are date mismatches. 

```python
# Merge the rainfall and temperature datasets
merged_dataset = pd.merge(sd_rainfall_rearrange, west_temp, on=['Year', 'Month'], how='outer')

# Merge the erosion dataset into the combined dataset
merged_dataset = pd.merge(merged_dataset, erosion_monthly, on=['Year', 'Month'], how='outer')

# The merged dataset is sorted by 'Year' and 'Month'
merged_dataset = merged_dataset.sort_values(by=['Year', 'Month']).reset_index(drop=True)
```

After merging, we sort the dataset chronologically and remove any rows with a **Month** value of 0 to maintain accuracy. Finally, we filter the dataset to include only records from 2017 to 2020, producing a focused dataset that comprehensively covers rainfall, temperature, and erosion metrics within this specified period, ready for our further analysis.

```python
# Remove 0 value in 'Month' column
merged_dataset = merged_dataset[merged_dataset['Month'] != 0].reset_index(drop=True)

# Filter to display row from year 2017 to 2020
filtered_dataset = merged_dataset[(merged_dataset['Year'] >= 2017) & (merged_dataset['Year'] <= 2020)]

filtered_dataset.head()
```

### 4. Line plot for Precipitation, Temperature, and Erosion from 2017 to 2020:

In this step, we plot precipitation, temperature, and erosion data from 2017 to 2020 to compare and identify patterns and potential correlations. We expect to see whether increased precipitation coincides with heightened coastal erosion in San Diego and how temperature variations may influence this relationship. 

```python
filtered_dataset['Date'] = pd.to_datetime(filtered_dataset[['Year', 'Month']].assign(DAY=1))

filtered_dataset['Precipitation (inches)'] = pd.to_numeric(filtered_dataset['Precipitation (inches)'], errors='coerce')
filtered_dataset['Upper_Cliff_Erosion'] = pd.to_numeric(filtered_dataset['Upper_Cliff_Erosion'], errors='coerce')

fig, ax1 = plt.subplots(figsize=(12, 6))
line1, = ax1.plot(filtered_dataset['Date'], filtered_dataset['Precipitation (inches)'], label='Precipitation', color='blue')
line2, = ax1.plot(filtered_dataset['Date'], filtered_dataset['Temperature(°F)'], label='Temperature', color='red')

ax2 = ax1.twinx()
line3, = ax2.plot(filtered_dataset['Date'], filtered_dataset['Upper_Cliff_Erosion'], label='Erosion', color='green')

ax1.set_xlabel('Year')
ax1.set_ylabel('Precipitation (inches) and Temperature (F)', color='black')
ax2.set_ylabel('Erosion (cubic meters)', color='green')

lines = [line1, line2, line3]
labels = [l.get_label() for l in lines]
ax1.legend(lines, labels, loc='best')

plt.title('Trend of Precipitation, Temperature, and Erosion (2017-2020)')
plt.grid(True)
plt.tight_layout()
plt.show()
```

 Upon analyzing the plotted trends of precipitation, temperature, and erosion from 2017 to 2020, several notable observations arise. Additionally, there seems to be a potential relationship between precipitation and erosion, as periods of increased precipitation coincide with spikes in erosion levels. Moreover, while temperature variations exhibit some correlation with precipitation, their relationship with erosion levels might necessitate further investigation. These initial observations suggest intricate dynamics within the environmental system, hinting at the interplay between climate elements and erosion processes over the examined time frame.
