Interest Score/Interest Over Time Definition:  <br>
The "interest score" on Google Trends represents the relative popularity of a search query in a
specific region and time frame. It is indexed from 0 to 100, where 100 signifies the peak popularity
for the term. This score does not indicate the absolute search volume but rather shows the search term's popularity relative to the highest point on the chart for the given region and time. A higher score means more people are searching for that particular term at that time, while a lower score indicates lesser interest. The data is useful for identifying trends and understanding how interest in certain topics changes over time. <br>

```python
# Data cleaning to assign labels 'low', 'rising', 'popular', and 'trending' to each datapoint.
def bin_interest_score(num):
    if num < 25:
        return 'low'
    elif num < 50:
        return 'rising'
    elif num < 75:
        return 'popular'
    else:
        return 'trending'

df_all_trends['US-CA_Interest_Label'] = df_all_trends['US-CA_Interest_Score'].apply(bin_interest_score)
df_all_trends['US-NY_Interest_Label'] = df_all_trends['US-NY_Interest_Score'].apply(bin_interest_score)
df_all_trends.iloc[5200:5600]
```

```python
# Sample data
from datetime import datetime

dates_column = df_all_trends['date']

# Convert dates to the desired format
formatted_dates = []
for date_str in dates_column:
    try:
        # Try parsing the date as "%m/%d/%Y"
        original_date = datetime.strptime(date_str, "%m/%d/%Y")
        formatted_dates.append(original_date)
    except ValueError:
        try:
            # If parsing as "%m/%d/%Y" fails, try parsing as "%Y-%m-%d"
            original_date = datetime.strptime(date_str, "%Y-%m-%d")
            formatted_dates.append(original_date)
        except ValueError:
            # If both formats fail, handle the error or skip the date
            print(f"Error: Could not parse date '{date_str}'")
            continue
df_all_trends['date'] = formatted_dates
df_all_trends.insert(1, 'month', df_all_trends['date'].dt.month)
df_all_trends.insert(2, 'year', df_all_trends['date'].dt.year)
df_all_trends.head()
```

## Combined New York and California Trends Overtime


```python
all_trends = df_all_trends['trend'].unique()

for trend in all_trends:
    trend_data = df_all_trends[df_all_trends['trend'] == trend]
    trend_data = trend_data.sort_values(by='date')  # Sort by date

    # Convert dates to numeric values for regression
    trend_data['date_num'] = pd.to_numeric(trend_data['date'])

    plt.figure(figsize=(8, 3))

    # Plot NY data and trendline
    plt.plot(trend_data['date'].values, trend_data['US-NY_Interest_Score'].values, color="orange", label="New York")
    ny_slope, ny_intercept = np.polyfit(trend_data['date_num'], trend_data['US-NY_Interest_Score'], 1)
    plt.plot(trend_data['date'].values, ny_slope * trend_data['date_num'] + ny_intercept, color="red", linestyle="--")

    # Plot CA data and trendline
    plt.plot(trend_data['date'].values, trend_data['US-CA_Interest_Score'].values, color="lightskyblue", label="California")
    ca_slope, ca_intercept = np.polyfit(trend_data['date_num'], trend_data['US-CA_Interest_Score'], 1)
    plt.plot(trend_data['date'].values, ca_slope * trend_data['date_num'] + ca_intercept, color="green", linestyle="--")

    plt.xlabel('Date')
    plt.ylabel('Interest Score')
    plt.title(f'Fluctuations of {trend} with Trendlines in Overall Interest Score')
    plt.legend(loc="upper right")
    plt.grid(True)
    plt.show()
```
