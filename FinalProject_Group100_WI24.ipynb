# Research Question

How do remote work arrangements compared to traditional office-based work impact the mental health of employees in the tech industry?

## Background and Prior Work

Ever since the pandemic, the workflow of society has completely transformed. According to an article from the BBC, remote work has been heralded as the solution to our fast-paced, pre-pandemic lifestyles and allows for greater flexibility and convenience. With the option of working from home, employee’s have expressed improved job satisfaction and more quality time to spend with their families (2). However, recent research has shown that remote work has been taking more of a toll on our mental health than we may have anticipated. The Society of Human Resource Management (SHRM) details this shift in an article on potential downsides to remote work. They outline that despite the initial grandeur and comfortability of working in the comfort of one’s own home, the daily absence of social interaction has led to an increased sense of social isolation, social anxiety, social insecurity, and wasted and longer hours (3).

As society faces a variety of mental health crises in different settings, much research has been conducted on this relationship between the work-place and employee’s mental health. Researchers based in Canada from the Faculty of Health Sciences at Simon Fraser University, the Department of Psychology at Toronto Metropolitan University, and the Institute for Social Connection researched this topic on the impact of working from home on mental health - specifically examining Canadian workers’ mental health during the third-wave of the COVID-19 Pandemic (1). They sought to test the potential relationship between different work settings (remote, in-person, hybrid) on self-rated mental health, through which they modeled in the form of an online survey. They concluded that hybrid-work arrangements were most positively associated with positive self-rated mental health. In addition to conventional confounding variables (i.e. age, gender, origin, education, hours per week), they also examined how stresses from complying with COVID-19 guidelines affected an individual's mental health. Their main findings include the difficulties in establishing a work-life balance and increased feelings of isolation.

The earlier mentioned article from BBC explored the idea that remote work is worse for people’s well-being than many may think. While remote work has resulted in a general consensus that people feel physically and mentally more stable, the reality of recent findings is that this case may not be entirely accurate. One employee notes that “working alone all day every day…is tough,” and that they have found that “instead of taking breaks to chat to people, “ they “pick up [their] phone. All of the extra screen time has definitely had a negative impact on [their] wellbeing,” (2). While remote work may initially have been viewed as a utopia of breaks, multi-tasking, and working comfortably, research shows that the reality is a complicated picture of what wellbeing looks like in employees that work from home.

There is countless research that explores this relationship between mental health and the work-place, exploring a range of factors from slow internet connection, a house in disarray, constant interruptions, isolation, and a seemingly endless work day (3). As we live in a society that is consistently fueled by industries, we believe that analyzing the relationship between mental health and working life is critical to improving society as a whole. These research findings reveal that this is an issue worth tackling and analyzing for the betterment of interpersonal and societal relationships.

```python
sns.set_style("whitegrid")
plt.figure(figsize=(20, 8))

# Subplot 1: Benefits provided to the Employees
plt.subplot(1, 2, 1)
eda_percentage = df['benefits'].value_counts(normalize=True).rename_axis('benefits').reset_index(name='Percentage')
ax = sns.barplot(x='benefits', y='Percentage', data=eda_percentage, palette='copper_r')
for p in ax.patches:
    width = p.get_width()
    height = p.get_height()
    x, y = p.get_xy()
    ax.annotate(f'{height:.0%}', (x + width/2, y + height*1.01), ha='center', fontweight='bold', fontsize=12)

plt.title('Benefits provided to the Employees', fontsize=18, fontweight='bold')
new_labels = ['Yes', 'Don\'t know', "No"]
plt.xticks(ticks=[0, 1, 2], labels=new_labels, fontsize=16)
plt.yticks(fontsize=16)
plt.legend().remove()  # Assuming we don't need a legend here as it's straightforward

# Subplot 2: Benefits provided and Treatment sought, with count-based percentage annotations
plt.subplot(1, 2, 2)
ax = sns.countplot(x='benefits', hue='treatment', data=df, palette='Blues_r')
plt.title('Benefits provided to the Employees', fontsize=18, fontweight='bold')
count = 0
for p in ax.patches:
    if(count == 0 or count == 3):
      total = len(df[df['benefits']==0])
    elif(count == 1 or count == 4):
      total = len(df[df['benefits']==1])
    else:
      total = len(df[df['benefits']==2])
    height = p.get_height()
    percentage = 100 * (height / total)
    if(percentage == 0.0):
      continue
    count = count+1
    ax.annotate(f'{percentage:.1f}%', (p.get_x() + p.get_width() / 2., height),
                ha='center', va='bottom', fontsize=12, fontweight='bold')

plt.xticks(ticks=[0, 1, 2], labels=new_labels, fontsize=16)
plt.yticks(fontsize=16)
legend_labels = ['Didn\'t Seek Treatment', 'Sought Treatment']
plt.legend(title='Treatment Status', fontsize=16, labels=legend_labels)

plt.tight_layout()
plt.show()
```

Here we have two graphs that show the benefits provided to the employees. At 36%, there were more employees who were not provided benefits. Of those employees, most were likely to not seek treatment more than any other group. On the contrary, employees who were provided benefits were more likely to seek treatment than any other group. This could indicate that employees with benefits feel more inclined to seek out treatment because it is available to them.
