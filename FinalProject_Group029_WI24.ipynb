## Background and Prior Work

Since 1982, the NFL has held an annual Draft Combine for the upcoming college players eligible to be selected in that year’s NFL Draft. Each year, the league invites up to 335 prospects to participate in the many tests and drills of the Combine. Some of the main tests in the Combine include the 40-yard dash, broad jump, vertical jump, bench press, and 3-cone drill, which are designed to test a player’s speed, agility, strength, and jumping ability. Positive performance in the Combine has been known to increase the draft stock of a player, while a negative performance can decrease draft stock. Furthermore, the earlier a player gets drafted the more money a player is guaranteed to make in their rookie contracts. Therefore, many players look at the draft as a big opportunity to not only make it to the NFL but also to get a significant payday.

There have been many studies around the topic of NFL Draft Combine performance. Cal State Fullerton did a study on the effect of performance in Draft Combine drills on the length of players’ careers and postseason accolades.<a name="cite_ref-1"></a>[<sup>1</sup>](cite_note-1) The drills they used were the 40-yard dash, vertical jump, broad jump, bench press, three-cone drill, and 20-second shuttle, and they utilized data from all offensive and defensive players. The researchers also measured length of career with years played and postseason accolades as All-Pro selections, Pro Bowl selections, NFL individual awards and Super Bowls won. Fullerton then used Multivariate Analysis of Variance (MANOVA) and Multiple Regressions in order to measure correlation for multiple positions. They found that the 40-yard dash was a great indicator of NFL success for all positions using their chosen metrics. In addition, UC Berkeley did a study using the same drills as Fullerton, but focused on quarterbacks, running backs and receivers, and used draft position and in-game performance as their metrics.<a name="cite_ref-2"></a>[<sup>2</sup>](cite_note-2) They used a Random Forest model in this study in order to find a correlation between certain speed and agility drills and yards per game in the NFL. They found that the NFL Combine drills had less of an effect on in-game performance than expected.

Our project will be similar to the two projects listed above in the sense that we are measuring the effect of the NFL Combine. There will be some key differences between our projects though. For example, the Cal State Fullerton research encompassed players from 2004 to 2009 while UC Berkeley encompassed players from 2000 to 2010. Our research, however, plans to encompass players from 2000 to 2022. This could prove to cause significant differences in our findings compared to previous research as sports such as football have greatly changed in the last decade. New play styles, salary changes, position preferences, technology, etc could all cause new effects on NFL Combine data and subsequently draft positions and salary earnings. Furthermore, our project will also focus on career earnings. Although draft success and career earnings are likely to be positively correlated, our team believes that there can be a lot of interesting underlying data and analysis to be done. For example, earning a lot of money does not necessarily mean a player had a successful career. There could be the possibility that some Combine drills are “overhyped” and do not accurately depict the skill level of an individual. Our team hopes that through our research we will be able to pinpoint these differences and showcase the full effect of the NFL Combine on a player’s future.

# Hypothesis


Our team hypothesizes that the better an NFL player’s Combine performance is the higher their draft position will be and subsequently the higher their career earnings overall. This is because a higher draft position means more guaranteed money a rookie player will earn on their first contract alongside the understanding that earlier draft picks should mean higher quality players.
```python
# All Necessary Imports
import csv
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import matplotlib.ticker as mtick
import seaborn as sns
```

```python
stats_data_path = 'datasets/combine_stats_df.csv'
stats_data = pd.read_csv(stats_data_path)
stats_data.head()
```

```python
# Drop unnecessary unnamed column, School, and Shuttle
stats_data.drop(stats_data.columns[0], axis=1, inplace=True)
stats_data.drop(columns=['School','Shuttle'], inplace=True)

# Convert Height to inches
def feet_inches_to_inches(height):
    if isinstance(height, float):
        return height * 12 
    else:
        feet, inches = map(int, height.split('-'))
        total_inches = feet * 12 + inches
        return total_inches
    
stats_data['Ht'] = stats_data['Ht'].apply(feet_inches_to_inches)

# Rename columns for clarity
stats_data = stats_data.rename(columns={'Pos':'Position','Ht': 'Height (in)','Wt':'Weight (lb)','draft_year':'Draft Year', '40yd' : '40yd (s)', 'Vertical':'Vertical (in)', 'Bench': 'Bench (reps)', 'Broad Jump' : 'Broad Jump (in)', '3Cone': '3Cone (s)'})

stats_data.head()
```

```python
# grabs positions that we are interested in
stats_data = stats_data[(stats_data['Position'] == 'RB') | (stats_data['Position'] == 'WR') | (stats_data['Position'] == 'CB') | (stats_data['Position'] == 'DE')]
stats_data
```

#### Data Preparation

These steps are critical for data cleaning and preparation before analysis. By converting names and positions to lowercase, we ensure consistent formatting, preventing mismatches due to case differences. Dropping the original `Player` and `Position` columns after merging helps to avoid redundancy. The merge operation itself is aimed at combining performance metrics with draft and contract data based on player name, position, and draft year, creating a comprehensive dataset for analysis.

```python
# Created lowercase Player name and Position columns to prevent loss of data and make sure data is matched properly
stats_data['name_lower'] = stats_data['Player'].str.lower()
stats_data['pos_lower'] = stats_data['Position'].str.lower()
contract_draft_data['name_lower'] = contract_draft_data['Player'].str.lower()
contract_draft_data['pos_lower'] = contract_draft_data['Position'].str.lower()

# Drop Player and Position columns to prevent duplicate columns in the merged dataframe
contract_draft_data.drop(columns=['Player','Position'], inplace=True)

# Merge the two dataframes based on player name, position, and draft year
merged_df = pd.merge(stats_data, contract_draft_data, on=['name_lower','pos_lower','Draft Year'], how='inner')

# Drop duplicate/unnecessary columns
columns_to_drop = ['name_lower','pos_lower'] 
merged_df.drop(columns=columns_to_drop, inplace=True)
stats_data.drop(columns=columns_to_drop, inplace=True)
contract_draft_data.drop(columns=columns_to_drop, inplace=True)
print(merged_df.shape)
merged_df.head()
```

#### Data Filtering and Description

Here, the data is filtered to focus only on the positions of interest: Running Backs (RB), Wide Receivers (WR), Cornerbacks (CB), and Defensive Ends (DE). This step narrows the analysis to the specific groups we want to investigate, making subsequent analyses more relevant and targeted.

```python
Positions = ['RB', 'WR', 'CB', 'DE']
filtered_data = merged_df[merged_df['Position'].isin(Positions)]
# describing the data based on position and each skill test
yard_dash_distribution = filtered_data.groupby('Position')['40yd (s)'].describe()
vertical_distribution = filtered_data.groupby('Position')['Vertical (in)'].describe()
bench_distribution = filtered_data.groupby('Position')['Bench (reps)'].describe()
broad_jump = filtered_data.groupby('Position')['Broad Jump (in)'].describe()
cone = filtered_data.groupby('Position')['3Cone (s)'].describe()
filtered_data
```

#### Boxplots By Position

Boxplots are generated for each performance metric by player position. These visualizations help identify the distribution, median, and presence of outliers for each metric across different positions. From these plots, we might observe, for example, that certain positions tend to have faster 40-yard dash times or higher vertical jumps, reflecting the physical demands and performance expectations of those positions in the NFL.

```python
# get an idea of the outliers
plt.figure(figsize = (15,15)) 
Metrics = ['40yd (s)', 'Vertical (in)', 'Bench (reps)', 'Broad Jump (in)', '3Cone (s)']
descriptive_stats_total = filtered_data.groupby('Position')[Metrics].describe()
position_order = ['RB', 'WR', 'CB', 'DE']
colors = {
    'RB': 'blue',
    'WR': 'orange',
    'CB': 'green',
    'DE': 'red',
}

for i, metric in enumerate(Metrics, start=1):
    plt.subplot(2,3,i)
    sns.boxplot(x='Position', y = metric, data = filtered_data, hue='Position', palette=colors, legend=False, order = position_order)
    plt.title(f'Boxplot of {metric} by Position')
plt.tight_layout()
```

- These boxplots suggest that certain combine metrics may be more important for some positions than others. For example, speed metrics like the 40-yard dash appear to be critical for RBs, WRs, and CBs, as shown by the tighter IQRs for these groups. - Conversely, for DEs, strength metrics (bench press) seem to have a more central role, as indicated by the higher median and IQR. Meanwhile, agility and speed vary more widely among DEs, possibly because the position demands a combination of skills and attributes, and excellence in one particular metric may not be as strongly required as in others.

- Ultimately, these insights are valueable as they help us to identify which metrics we should prioritize when considering correlation between position and draft pick during the draft process.

#### Correlation Analysis

This step calculates the correlation between each performance metric and the draft pick number for each position. The correlations provide insights into how combine performance might influence draft outcomes. For instance, a positive correlation with the 40-yard dash time for WRs suggests that faster runners in this position might be picked earlier in the draft. This analysis reveals the varying importance of different metrics for different positions.

```python
correlation_data = filtered_data[['Position', '40yd (s)', 'Vertical (in)', 'Bench (reps)', 'Broad Jump (in)', '3Cone (s)', 'Draft Pick']]
for position in Positions:
    print(f"Correlation for {position}:")
    position_data = correlation_data[correlation_data['Position'] == position]
    position_data_numeric = position_data.drop(columns=['Position'])  # Exclude non-numeric column
    correlations = position_data_numeric.corr()['Draft Pick'].drop('Draft Pick')
    print(correlations)
    print("\n")
```

##### **Conclusions**

- The 40-yard dash seems to be a significant metric for RBs, WRs, and DEs, as expected for positions where speed is crucial.

- The vertical jump and bench press metrics show generally weak correlations across positions, hinting that these metrics may not be as highly valued for draft decisions, or their influence may be position-specific.

- The broad jump displays a consistent negative correlation for RBs and WRs, reinforcing the idea that explosive power is important for these positions.

- The 3Cone drill results are somewhat contradictory, with positive correlations where negative ones would be expected. This may warrant further scrutiny to understand the context or potential data issues.

#### Data Preparation

In these steps, we group the rows of the merged dataset by the player name, position, draft year, and all of their Combine metrics. The Contract Value and Guaranteed Value columns were summed, while the Contract Value Normalized and Guaranteed Value Normalized columns were averaged. This allows us to view each player’s total career earnings in one row. The reason our team grouped by so many variables is because we want to confirm that we are not grouping different players with the same name together.

```python
# groupby player and sum contract value, sum guaranteed value, mean contract value normalized, and mean guaranteed value normalized, include draft pick and position

combined_values = merged_df.groupby(['Player','Position','Draft Year','Draft Pick', '40yd (s)','Vertical (in)','Bench (reps)','Broad Jump (in)', '3Cone (s)']).agg({
    'Contract Value': 'sum',
    'Contract Value Normalized': 'mean',
    'Guaranteed Value': 'sum', 
    'Guaranteed Value Normalized': 'mean',
}).reset_index()

combined_values = combined_values.rename(columns={'Contract Value':'Total Contract Value','Contract Value Normalized':'Average Contract Value Normalized','Guaranteed Value': 'Total Guaranteed Value','Guaranteed Value Normalized': 'Average Guaranteed Value Normalized'})
combined_values
```

#### Boxplots By Position

Boxplots are generated for each contract metric by player position. These visualizations help identify the distribution, median, and presence of outliers for each metric across different positions. From these plots, we might observe, for example, that certain positions tend to have on average higher career earnings than others, showing that the NFL values those positions more highly.

```python
# boxplot (position specific) showing IQR of different measurements of earnings. Also indiciating outliers within the data
plt.figure(figsize = (15,15)) 
Metrics = ['Total Contract Value', 'Average Contract Value Normalized', 'Total Guaranteed Value', 'Average Guaranteed Value Normalized']
descriptive_contract_total = combined_values.groupby('Position')[Metrics].describe()
position_order = ['RB', 'WR', 'CB', 'DE']
colors = {
    'RB': 'blue',
    'WR': 'orange',
    'CB': 'green',
    'DE': 'red',
}

for i, metric in enumerate(Metrics, start=1):
    plt.subplot(2,2,i)
    sns.boxplot(x='Position', y = metric, data = combined_values, hue='Position', palette=colors, legend=False, order = position_order)
    plt.title(f'Boxplot of {metric} by Position')
plt.tight_layout()
```

- Finally, the plots conclusively show that RBs have the worst contract value, on average, out of the four chosen positions.

- Overall, this visualization shows that there is a big wage disparity among running backs, wide recievers, cornerbacks, and defensive ends. Most players make less than 25 million dollars in career earnings with some exceptions and outliers of players making over 200 million dollars in career earnings. This could depict how many players don't make it past their rookie contract or get cut early on in their careers. #### Scatter Plots for Each Earnings Metric by Every Combine Metric Difference

These scatter plots further explore the relationship between earnings metrics and combine metrics. Each plot shows the difference between the each player's combine metrics for each combine metric, calculated in such a way that the higher the difference, the better the combine performance. Each plot might show, for instance, whether the 40-yard dash or the broad jump tends to lead to better player earnings. These visualizations allow for a better understanding of how player performance in combine metrics can relate to their earnings and which metrics tend to lead to better contract values.

```python
Metrics = ['Total Contract Value', 'Average Contract Value Normalized', 'Total Guaranteed Value', 'Average Guaranteed Value Normalized']
Combine = ['40yd Difference (ds)','Vertical Difference (in)','Bench Difference (reps)','Broad Jump Difference (in)', '3Cone Difference (ds)']
colors = {
    '40yd Difference (ds)': 'blue',
    'Vertical Difference (in)': 'orange',
    'Bench Difference (reps)': 'green',
    'Broad Jump Difference (in)': 'red',
    '3Cone Difference (ds)': 'purple'
}

# Difference between values calculated. 40yd and 3Cone differences are calculated differently as they are timed values
max_40yd = combined_values['40yd (s)'].max()
min_vertical = combined_values['Vertical (in)'].min()
min_bench = combined_values['Bench (reps)'].min()
min_broad_jump = combined_values['Broad Jump (in)'].min()
max_3cone = combined_values['3Cone (s)'].max()

combined_values['40yd Difference (ds)'] = -10*(combined_values['40yd (s)'] - max_40yd)
combined_values['Vertical Difference (in)'] = combined_values['Vertical (in)'] - min_vertical
combined_values['Bench Difference (reps)'] = combined_values['Bench (reps)'] - min_bench
combined_values['Broad Jump Difference (in)'] = combined_values['Broad Jump (in)'] - min_broad_jump
combined_values['3Cone Difference (ds)'] = -10*(combined_values['3Cone (s)'] - max_3cone)

# scatterplot comparing the difference between each of the combine metrics across players to contract value. Combine performance is better as you go right
plt.figure(figsize=(15, 10))
for i, metric in enumerate(Metrics, start=1):
    plt.subplot(2, 2, i)
    for combine_metric in Combine:
        # Sort the data based on the combine metric before plotting
        sorted_data = combined_values.sort_values(by=combine_metric)
        plt.scatter(sorted_data[combine_metric], sorted_data[metric], label=f'{combine_metric}', alpha=0.5, color=colors[combine_metric])
    plt.title(f'{metric} vs Combine Metric Differences')
    plt.xlabel('Combine Metric Difference')
    if metric in ['Total Contract Value', 'Total Guaranteed Value']:
        plt.ylabel('Value (in Millions)')
        plt.gca().yaxis.set_major_formatter(mtick.FuncFormatter(lambda x, pos: f'{x / 1e6:.1f}'))
    else:
        plt.ylabel(metric)
    plt.legend()
    plt.grid(True)

plt.tight_layout()
plt.show()
```

**40-yard dash (40yd) and 3-cone drill (3Cone):**
- The difference values for 40-yard dash and 3-cone drill across players is less spread out than that of other metrics. This is due to the fact that they are calculated in time rather than as distance or reps.

##### **Conclusions**
- The lack of strong trends in these scatter plots indicates that there is no single combine metric that correlates to contract value.
- Outliers in the data suggest that exceptional combine results do not guarantee better contract value, but are more likely to lead to better contract value.

# Ethics & Privacy

In exploring the relationship between NFL Combine performance, draft position, and subsequent NFL career earnings, it's crucial to consider how ethics and privacy should be handled. There may be inherent biases in the dataset, both in terms of who it includes and how the data was collected. For instance, the dataset might primarily represent certain demographics or exclude underrepresented populations, skewing the analysis results. Sometimes certain players who become stars might not be invited to the Combine although they are highly skilled due to lack of access. Since the Combine invites players instead of having any player be allowed to show up, some players who play for lesser-known schools or lower divisions may not be fully represented in this dataset. Furthermore, it is very possible that a significant number of these underrepresented players actually make it to the NFL and have successful careers. Additionally, biases in data collection methods, such as subjective evaluations or incomplete records, could also impact the findings. From a preliminary look at our dataset, it seems that the “Combine Stats 2000-2022 with Draft Year” is missing some data for some observations. A possible reason for this is that a player chose to only do specific drills that they thought would benefit them. This in turn could cause some potential biases within our results as we may have a lot of data from certain drills for certain positions compared to other drills and positions.

Other ethical factors to consider include potential cheating or even a player focusing on a specific skill to stand out based on the results of our data. According to our hypothesis, a player with a great 40-yard dash is the best indicator of how well a player will perform. If that is the case, then Combine players may solely focus on the 40-yard dash to move up draft spots instead of focusing on improving other areas critical to their positions and roles.

When performing our data analysis and research, it is vital that our team keeps these ethical and bias concerns in mind. For example, if we are seeing trends that a significant portion of running backs are only doing the 40-yard dash, we would need to disclose that information to our readers and work on steps to find the effect that has on our data and analysis. Although our team as of now cannot say for sure how these potential biases will affect our results, we will actively work to identify these biases and clearly state their effects. Overall, It is essential to be transparent about the limitations and potential biases inherent in the dataset when communicating the analysis results. It is important to emphasize that our results should not be considered an “end all be all” as there are likely countless other factors that contribute to a player’s success that are not analyzed within our research scope. By openly acknowledging these issues and explaining the steps taken to mitigate them, we can uphold ethical standards and ensure the integrity and fairness of our analysis.
