# Abstract

Our project investigated whether NBA referees show bias in the way that they make foul calls in official league games, specifically whether certain referees call fouls in a way that favors a specific team or away/home team advantage. The way referees call games can impact the outcome of who wins or loses since foul calls can result in basket opportunities, player ejection, changes of possession, and influence the way players or teams will play. Our group hypothesized that certain NBA referees exhibit a pattern in their foul-calling which favors: the home team, which can influence game outcomes of giving home teams more of an advantage to winning. As humans, there may be biases or tendencies that referees have that can have a role in game outcomes and player/team performance.

Through our exploratory data analysis, we found that the majority of fouls called in NBA games are shooting and personal fouls (when a player is fouled while shooting the ball and when illegal contact is made with another player), which, like many foul calls, are subjective and up to interpretation by the referee. These types of fouls are called more on away teams than home teams. In addition, we saw a pattern that overall there are more fouls called on the away/visiting team (the away team has committed the illegal contact) than the home team, but the difference is not significant. Furthermore, we found that specifically for won games by the home team compared to lost home games, there is a bigger difference in the amount of fouls calls made.

In order to address this question, we first took data from the NBA about game, player, and referee statistics and extracted foul calls for which referees, games, and teams or players as well as the outcomes of the games. We compared each team’s win rate under each referee, and compared them across the NBA. We also compared the type and distribution of fouls called and how they correlated with the game’s outcome. Finally, we passed all the data through a machine learning classifier. We repeatedly found that individual referees did not have an impact on deciding the winner of the game, but that the number of fouls committed did, and that home field advantage was a real phenomenon. We were also able to eliminate referee bias as a cause of this advantage.

## Dataset #1: Games/Game Info
Contains rundowns of the teams, seasons, and outcome (who won/lost) for each game

```python
games_df = pd.read_csv('nbadata/csv/game.csv', parse_dates=['game_date'])

# Select games from our desired date range, and select columns of interest
games_df = games_df[(games_df['game_date'] >= MIN_DATE) 
                    & (games_df['game_date'] <= MAX_DATE) 
                    & (games_df['team_id_home'].isin(team_ids))
                    & (games_df['team_id_away'].isin(team_ids))]
games_df = games_df[['game_id', 'game_date', 'team_id_home', 'team_id_away', 'team_name_home', 'team_name_away', 'wl_home']]
games_df
```

## Dataset #2: Foul Plays

The `play_by_play` dataset contains one row for every play that occurs. We will filter these plays to find only plays in which a foul occurred.

```python
# Get data for all foul plays only
plays_df = pd.read_csv('nbadata/csv/play_by_play.csv', usecols=['game_id', 'eventnum', 'homedescription', 'neutraldescription', 'visitordescription'])
plays_df = plays_df.merge(games_df, how='inner', on='game_id')
plays_df = plays_df[(plays_df['homedescription'].str.contains('FOUL', na=False, regex=False)) | 
                    (plays_df['neutraldescription'].str.contains('FOUL', na=False, regex=False)) |
                    (plays_df['visitordescription'].str.contains('FOUL', na=False, regex=False))]
plays_df
```

# Exploratory Data Analysis

### Section 1 of EDA - Comparing home/vistor fouls and win rate

We would like to see if there is any kind of relationship between fouls committed and win rate, and if a home or visitor team is more likely to be fouled. Since different teams commit different amount/type of fouls, we will look at average over the whole NBA.

```python
def get_teams_fouls(subdf):
    n_home_fouls = (~subdf['homedescription'].isna()).sum()
    n_visitor_fouls = (~subdf['visitordescription'].isna()).sum()
    return pd.Series(data=[n_home_fouls, n_visitor_fouls, subdf['wl_home'].iloc[0]], index=['home_fouls', 'visitor_fouls', 'wl_home'])

total_fouls_df = plays_df.groupby('game_id').apply(get_teams_fouls)
total_fouls_df
```

### Section 1.1: Compare fouls committed vs. win rate

In other words, plot number of fouls for teams that won and number of fouls for teams that lost

```python
def get_fouls(row, outcome='W'):
    # get the number of fouls for teams based on outcome (W/L)
    if row['wl_home'] == 'W':
        if outcome == 'W':
            return row['home_fouls']
        else:
            return row['visitor_fouls']
    else:
        if outcome == 'W':
            return row['visitor_fouls']
        else:
            return row['home_fouls']
```

```python
fouls_winning = total_fouls_df.apply(get_fouls, axis='columns', args=('W',))
fouls_losing = total_fouls_df.apply(get_fouls, axis='columns', args=('L',))
```

```python
plt.figure('fouls vs win rate', figsize=(15, 8))
plt.hist(fouls_winning, bins=range(min(fouls_winning.min(), fouls_losing.min()), max(fouls_winning.max(), fouls_losing.max()) + 1), alpha=0.5, label='Wins', color='g')
plt.hist(fouls_losing, bins=range(min(fouls_winning.min(), fouls_losing.min()), max(fouls_winning.max(), fouls_losing.max()) + 1), alpha=0.5, label='Losses', color='r')
plt.title('Distribution of Fouls in Winning vs Losing Games')
plt.xlabel('Number of Fouls')
plt.ylabel('Frequency')
plt.legend()
plt.show()
```

```python
# Basic descriptive statistics
display(fouls_winning.describe())
display(fouls_losing.describe())
```

As we can see from the histograms and the descriptive statistics, it's a very slight relationship, but it appears that the winning team commits less fouls than the losing team overall. This makes sense, as more fouls means more free throws / change of possessions for the other team.

### Section 2: Exploring the most common types of fouls

Certain fouls are much more common, and will have a larger impact on who wins the game. We explore the overall distribution here.


```python
home_fouls_types = plays_df['homedescription'].str.extract(r'([A-Z.]+\.FOUL)').dropna().value_counts()
visitor_fouls_types = plays_df['visitordescription'].str.extract(r'([A-Z.]+\.FOUL)').dropna().value_counts()
```

```python
fouls_types = home_fouls_types.add(visitor_fouls_types, fill_value=0)
plt.figure('fouls types', figsize=(15, 8))
fouls_types.plot(kind='bar', color='blue')
plt.title('Frequency of Different Types of Fouls')
plt.xlabel('Foul Type')
plt.ylabel('Frequency')
plt.legend(['Total Fouls'])
plt.show()
```

We see that two foul types, personal fouls and shooting fouls, are by far the most common. Personal fouls are often very subjective, and is an area that a referee could have a serious impact on. We will keep this in mind during main analysis.

### Section 3: Exploring the distribution in number of fouls per game per referee

```python
fouls_by_referee = pd.merge(officials_df, plays_df[['game_id', 'eventnum']], on='game_id', how='inner')

# Count the number of fouls called by each referee
fouls_count = fouls_by_referee.groupby(['official_id', 'first_name', 'last_name'])['eventnum'].count().reset_index(name='foul_count')
# Merge this count with the total games officiated by each referee
games_by_referee = officials_df.groupby(['official_id', 'first_name', 'last_name'])['game_id'].nunique().reset_index(name='games_officiated')
fouls_analysis = pd.merge(fouls_count, games_by_referee, on=['official_id', 'first_name', 'last_name'])
# Calculate fouls per game
fouls_analysis['fouls_per_game'] = fouls_analysis['foul_count'] / fouls_analysis['games_officiated']
plt.figure(figsize=(10, 6))
sns.boxplot(x='fouls_per_game', data=fouls_analysis)
plt.title('Distribution of Fouls Called per Game by Referees')
plt.xlabel('Fouls per Game')
plt.show()
```

We can see that the vast majority of referees call around 27-35 fouls in a game, but there are many outliers. This is most likely due to insufficient sample size, as many referees have only officiated a few games.

```python
# Calculate the standard deviation of fouls per game for each referee
std_dev_fouls = fouls_by_referee.groupby(['official_id', 'first_name', 'last_name'])['game_id'].nunique()
std_dev_fouls = std_dev_fouls.reset_index(name='std_dev_fouls_per_game')

# Merge with fouls_analysis to get standard deviation alongside average fouls per game
fouls_analysis = pd.merge(fouls_analysis, std_dev_fouls, on=['official_id', 'first_name', 'last_name'])
plt.figure(figsize=(12, 7))
sns.scatterplot(x='fouls_per_game', y='std_dev_fouls_per_game', data=fouls_analysis)
plt.title('Referee Consistency: Average Fouls per Game vs. Variability')
plt.xlabel('Average Fouls per Game')
plt.ylabel('Standard Deviation of Fouls per Game')
plt.show()
```

### Section 4: Exploring how fouls are called on home vs away teams

As our background research showed, referees often have a bias to call fouls in favor of the home team. We will do some surface-level exploration here.

```python
# Add a column to indicate if the foul was called on the home or away team
def foul_location(row):
    if pd.notna(row['homedescription']):
        return 'Home'
    elif pd.notna(row['visitordescription']):
        return 'Away'

plays_df['foul_on'] = plays_df.apply(foul_location, axis=1)
# Count the fouls for home and away
fouls_home_away = plays_df.groupby('foul_on')['eventnum'].count().reset_index(name='count')
plt.figure(figsize=(8, 5))
sns.barplot(x='foul_on', y='count', data=fouls_home_away)
plt.title('Number of Fouls Called on Home vs. Away Teams')
plt.xlabel('Team Type')
plt.ylabel('Number of Fouls')
plt.show()
```

```python
# Count fouls against home and away teams
fouls_home = plays_df[plays_df['foul_on'] == 'Home'].groupby('game_id')['eventnum'].count().reset_index(name='fouls_against_home')
fouls_away = plays_df[plays_df['foul_on'] == 'Away'].groupby('game_id')['eventnum'].count().reset_index(name='fouls_against_away')
# Merge the counts back into a single DataFrame
fouls_summary = pd.merge(fouls_home, fouls_away, on='game_id', how='outer').fillna(0)
# Merge this summary with games_df to include game outcomes
games_df['home_win'] = games_df['wl_home'].apply(lambda x: 1 if x == 'W' else 0)
game_fouls = pd.merge(fouls_summary, games_df[['game_id', 'home_win']], on='game_id')
display(game_fouls.head())
```

```python
avg_fouls_outcome = game_fouls.groupby('home_win').agg(
    avg_fouls_against_home=('fouls_against_home', 'mean'),
    avg_fouls_against_away=('fouls_against_away', 'mean')
).reset_index()
```

```python
positions = np.arange(len(avg_fouls_outcome['home_win']))
width = 0.25

# Plotting
plt.figure(figsize=(10, 6))
plt.bar(positions - width/2, avg_fouls_outcome['avg_fouls_against_home'], width, label='Fouls Against Home')
plt.bar(positions + width/2, avg_fouls_outcome['avg_fouls_against_away'], width, label='Fouls Against Away')
plt.xlabel('Game Outcome for Home Team')
plt.ylabel('Average Number of Fouls')
plt.title('Average Number of Fouls by Game Outcome')
plt.xticks(positions, ['Lost', 'Won'])
plt.legend()
plt.show()
```

In the two plots above, we can see that winning teams usually commit less fouls. This is in-line with our intuition.

```python
X = sm.add_constant(game_fouls['fouls_against_away'])
Y = game_fouls['home_win']

model = sm.Logit(Y, X).fit()

model.summary()
```

The basic logistic regression shown above affirms the belief that more fouls committed by the opposing teams means a higher chance of winning. In the "coef" column for "fouls_against_away", the value of +0.0383 confirms this positive relationship.

## Analysis Section 1: Determine if Certain Referees Impact the Outcome
We would like to find out if certain referees have a hand in raising/lowering a certain team's win rate when officiating. We will determine this by comparing a team's overall win rate to their win rate when officiated by a certain referee.

We would now like to detect any biases that referees have in calling fouls on home/away teams. Our EDA section showed that referees called slightly more fouls on away teams, and that the team with more fouls tended to lose. Therefore, a referee can have a significant impact on who wins the game.

We also limit the number of referees to the 100 most active, since many referees have only officiated one or two of a team's games, which is too small a sample size.


### Analysis 1.1: Find Win Rate of Referee-Team Pairing

We will first count the wins and losses for all times a referee was present at a team's game. In other words, we list all games for Team Y that referee X officiated.

```python
games_refs = games_df.merge(officials_df[['game_id','official_id']], on='game_id')
games_refs
```

```python
# Convert the dataframe to long format on team IDs, so each row represents one team rather than one game
melt_cols = ['team_id_home', 'team_id_away']
keep_cols = [c for c in games_refs.columns if c not in melt_cols]

games_refs_long = pd.melt(games_refs, id_vars=keep_cols, value_vars=melt_cols, var_name='team_type', value_name='team_id')
```

```python
def get_all_winloss_rate(sub_df):
    def get_winloss(row):
        return (row['wl_home'] == 'W' and row['team_type'] == 'team_id_home') or (row['wl_home'] == 'L' and row['team_type'] == 'team_id_away')
    
    rate = sub_df.apply(get_winloss, axis='columns').mean()
    return pd.Series(index=['num_games', 'win_rate'], data=[len(sub_df), rate])
        
games_refs_overall = games_refs_long.groupby(['team_id', 'official_id']).apply(get_all_winloss_rate)
games_refs_overall
```

```python
top100_ref = officials_df['official_id'].value_counts()[:100].index.sort_values()
top100_outcomes = games_refs_overall[games_refs_overall.index.get_level_values('official_id').isin(top100_ref)].sort_index(level=1)
top100_outcomes_str = top100_outcomes.reset_index().astype({'official_id': 'str'}).set_index(['team_id', 'official_id'])
```

```python
# Visualize 5 of the variation of win rates for the top 100 referees
fig, ax = plt.subplots()
for team_id in top100_outcomes_str.index.levels[0][:5]:
    top100_outcomes_str.loc[team_id].plot(kind='line', ax=ax, y='win_rate', legend=False, figsize=(15,8))
```

In the plot above, each line represents one team; the x-axis represents the referees, and the y-axis represents a team's win rate under a given referee. We can see that win rates can fluctuate a lot, however it is unclear whether this is due to a referee or some other factor. 

Below, we will conduct pairwise T-tests that compare a team's win rate under a certain referee to the team's win rate with all other referees. This can give us an indication if certain referees have an impact on the winner.

```python
# Conduct many pairwise t-tests to find if certain referees have a tendency to have a biased effect on a game's outcome
test_results = []
for ref_id in top100_ref:
    ref_winrates = top100_outcomes.loc[(slice(None), ref_id), :].droplevel(1)
    other_winrates = top100_outcomes[top100_outcomes.index.get_level_values('official_id') != ref_id] \
                                        .reset_index()

    result = ttest_ind(ref_winrates['win_rate'], other_winrates['win_rate'])
    test_results.append(result)
    print(f'Referee ID {ref_id}:', result)
```

```python
fig, ax = plt.subplots(figsize=(15,8))
ax.plot(range(len(test_results)), [t.pvalue for t in test_results])
fig.suptitle('P-values Comparing Win Rates Across All Referees')
ax.set_xlabel('Referee index')
ax.set_ylabel('Average Team Win Rate')
ax.axhline(0.05, color='red')
```

Examining the p-values above, we can see that none of them are even below 0.1. This suggests that the invidiual referees have little  to no influence on who wins the game. Although, some p-values are rather low, such as referee ID 1628951 with a p-value of 0.1899, suggesting that maybe this particular referee has a slight impact on the game's winner. 

Keep in mind that games are officiated by groups of referees, with one head referee who makes the final call. As we saw above, this limits the influence of individual referees. It's likely that the variation in win rates (as seen by the Atlanta Hawks) cannot be explained by individual referee bias. 

## Analysis Section 2: Determine if Certain Referees Call More Fouls on Home or Away Teams

Expanding on the previous analysis section, we would like to see how a difference in foul-calling impacts the game's outcome. That is, do referees prefer to call certain types of fouls on home team as opposed to away teams?

### Analysis 2.1: Group by referee ID, team ID, and court (home or away) and average the number of each type of foul called on the team by the referees

```python
home_fouls_types = plays_df['homedescription'].str.extract(r'(\w+\.FOUL)').dropna().value_counts()
visitor_fouls_types = plays_df['visitordescription'].str.extract(r'(\w+\.FOUL)').dropna().value_counts()
```

```python
# Gets the distribution of types of fouls committed by a certain team
def get_fouls_info(plays, team_id, court):
    if court == 'home':
        fouls_df = plays[(plays['team_id_home'] == team_id) & (plays['foul_on'] == 'Home')].copy()
        fouls_df['foul_type'] = fouls_df['homedescription'].str.extract(r'([A-Z.]+\.FOUL)')
    elif court == 'away':
        fouls_df = plays[(plays['team_id_away'] == team_id) & (plays['foul_on'] == 'Away')].copy()
        fouls_df['foul_type'] = plays['visitordescription'].str.extract(r'([A-Z.]+\.FOUL)')

    def get_fouls_distribution(game_df):
        return game_df.groupby('foul_type').count()[['eventnum']].rename(columns={'eventnum': 'count'})

    return fouls_df.groupby('game_id').apply(get_fouls_distribution).assign(team_id=team_id, court=court)
```

```python
foul_types_list = []
for team_id in team_ids:
    for court in ['home', 'away']:
        foul_types_list.append(get_fouls_info(plays_df, team_id, court))
```

```python
# foul_types is a dataframe indexed by (game ID, team ID, court) that counts the number and type of fouls committed by that team in that game
foul_types = pd.concat(foul_types_list).reset_index().pivot(columns=['foul_type'], values=['count'], index=['game_id', 'team_id', 'court']).fillna(0)
foul_types.columns = foul_types.columns.levels[1]
foul_types
```

### Analysis 2.2: Perform T-test to Determine Whether Home/Away Teams Commit the Same Amount and Type of Fouls

Across all teams and games, we examine how home teams and away teams are different in the amount and type of fouls they commit. Our t-tests will use a alternative hypothesis that claims "home teams commit less fouls on average than away teams".

```python
team_home_fouls = foul_types.xs('home', level='court')
team_away_fouls = foul_types.xs('away', level='court')
result = ttest_rel(team_home_fouls, team_away_fouls, alternative='less')

pd.DataFrame(data={'foul_type': foul_types.columns, 'statistic': result.statistic, 'pvalue': result.pvalue})
```

In the dataframe above, we can see that many p-values are close to 0, implying that away teams commit more of these fouls than home teams, on average. Interestingly, the p-values for shooting fouls (S.FOUL) and personal fouls (P.FOUL) have p-values essentially equal to 0. This means that home teams commit much less of these fouls than away teams. Since these fouls, especially personal fouls, are very subjective, this suggests that referees have an impact on the fouls committed. 

**However, we cannot claim that referees are the cause of this disparity of foul-calling.**

## Analysis Section 3: Machine Learning Classifier
Finally, to detect a correlation between referee and game outcome, we will train a simple machine learning classifier. First, we will train the model with all game information, and the model will try to predict which team will win. Then, we will drop the referee ID from the dataset and re-train. If the model gets worse, then we can say that there was information in the referee ID that had an impact on predicting the winner or loser. We also repeat this process while also dropping fouls data, then dropping all data.

We will limit the referees to the top 100 (i.e., the 100 referees that officiated the most games) to simplify the problem and remove referees with a low sample size.

```python
foul_col_names = [c for c in foul_types.columns if 'FOUL' in c]
foul_types_long = foul_types.reset_index().pivot(columns=['court'], values=['team_id'] + foul_col_names, index='game_id').reset_index()
foul_types_long.columns = foul_types_long.columns.map('_'.join)
foul_types_long = foul_types_long.rename(columns={'game_id_': 'game_id'})
foul_types_long
```

```python
all_data = games_refs.merge(foul_types_long, on=['game_id', 'team_id_home', 'team_id_away'])
```

```python
data_foul_col_names = [c for c in all_data if 'FOUL' in c]
```

**MODEL 1: Train with both official ID and foul information**

```python

x_cols = ['team_id_home', 'team_id_away', 'official_id'] + data_foul_col_names
y_col = 'wl_home'
X = all_data[x_cols]
y = all_data[y_col]

clf1 = GradientBoostingClassifier(n_estimators=200)
clf1.fit(X, y)
print('Summary of model trained on data including official ID')
print(f'Mean accuracy: {clf1.score(X, y)}')
print('Feature Importances:')
display(pd.Series(data=clf1.feature_importances_, index=x_cols).sort_values(ascending=False))
print('Confusion matrix')
confusion_matrix(y, clf1.predict(X))
```

**MODEL 2: Train with only foul information**

```python
x_cols = ['team_id_home', 'team_id_away'] + data_foul_col_names
y_col = 'wl_home'
X = all_data[x_cols]
y = all_data[y_col]

clf2 = GradientBoostingClassifier(n_estimators=200)
clf2.fit(X, y)
print('Summary of model trained on data without official ID')
print(f'Mean accuracy: {clf2.score(X, y)}')
print('Feature Importances:')
display(pd.Series(data=clf2.feature_importances_, index=x_cols).sort_values(ascending=False))
print('Confusion matrix:')
confusion_matrix(y, clf2.predict(X))
```

**MODEL 3: Train with only official ID**

```python
x_cols = ['team_id_home', 'team_id_away', 'official_id']
y_col = 'wl_home'
X = all_data[x_cols]
y = all_data[y_col]

clf3 = GradientBoostingClassifier(n_estimators=200)
clf3.fit(X, y)
print('Summary of model trained on data without foul numbers')
print(f'Mean accuracy: {clf3.score(X, y)}')
print('Feature Importances:')
display(pd.Series(data=clf3.feature_importances_, index=x_cols))
print('Confusion matrix:')
confusion_matrix(y, clf3.predict(X))
```

**MODEL 4: Baseline model, trained with only team IDs and no other information**

```python
x_cols = ['team_id_home', 'team_id_away']
y_col = 'wl_home'
X = all_data[x_cols]
y = all_data[y_col]

clf4 = GradientBoostingClassifier(n_estimators=200)
clf4.fit(X, y)
print('Summary of model trained on data without foul numbers')
print(f'Mean accuracy: {clf4.score(X, y)}')
print('Feature Importances:')
display(pd.Series(data=clf4.feature_importances_, index=x_cols))
print('Confusion matrix:')
confusion_matrix(y, clf4.predict(X))
```

We can see that, including the official ID and the types/counts of fouls, a simple gradient boosted decision tree algorithm can achieve a modest 62% accuracy on the dataset (model 1). Interestingly, when we drop the official ID from the dataset (model 2), the accuracy does not change. However, dropping the foul information and using only the official's ID (model 3) reduces accuracy to 58%. Dropping both official ID and foul information (model 4) also resulted in a 58% accuracy, implying that knowing the official ID does not help the model at all.

This implies that just foul information is moderately important to predicting the winner. Indeed, models 1 and 2 found that knowing the number of personal/shooting fouls from each team was most important to predicting the outcome. In Analysis 2, we found that personal/shooting fouls are most likely called on away teams more than home teams, however Analysis 1 showed that referees had little to no impact on who wins the game. 

From this, we may conclude that, while away teams are moderately more likely to lose than home teams (due to committing more fouls), this may not be because of the referee. The cause is most likely due to other factors, such as crowd noise, jet lag, etc.

# Discusison and Conclusion

We find that the number and type of fouls called on each team certainly has an effect on who wins the game. In particular, shooting fouls and personal fouls are the most indicative. However, even though these two fouls can be very subjective in play-calling, our study finds that the referee who calls them does not have a significant influence on who wins.

Our EDA confirmed the prior belief that teams who commit more fouls tend to lose more often. We also found that the number of fouls per game is fairly consistent, around 27 to 35. We also found that away teams committed slightly more fouls than home teams.

In Analysis 1, we tried to determine if certain referees affected a team’s win rate, by computing a team’s win rate under a referee and comparing to their win rate under all other referees with a t-test. However, all p-values were large (the smallest was around 0.19), failing to reject the null hypothesis that referees do not have an impact on a team’s win rate. We concluded that all of the top 100 most active referees are relatively unbiased.

In Analysis 2, we expand on our EDA by examining the distribution of types/counts of fouls called in the context of home vs away teams. Again, using a t-test for each type of foul, we find that away teams commit more fouls than home teams overall. Among these are personal fouls, which are very subjective, and is an area in which referees may have an impact. However, Analysis 2 found that referees did not have an impact overall, so this disparity of fouls is likely due to some other factor.

In Analysis 3, we put all the data on fouls and official IDs into a gradient-boosted decision tree classifier that tries to predict the winner of the game. The classifier was only able to predict with 62% accuracy, however this decreased to 58% when we removed foul data. Interestingly enough, removing official IDs from the dataset did not affect accuracy at all, which further emphasizes the finding that individual referees have little to no impact on who wins the game. 

We conclude that referees do not have a significant impact on deciding the game’s outcome. We did find strong evidence that away teams commit more fouls and lose more often than home teams, but we were not able to attribute this “home field advantage” to the referees. Home field advantage is most likely due to some other factor, such as jet lag on the away team, opposing crowd noise, or unfamiliarity of the facility.

Limitations of our analysis are that we specifically looked at foul calls only to see any correlations between game outcome and home and away teams, but there are many other factors that can contribute to team and player performance and referee calls. In addition, we can only compare the amount of foul calls made, who the fall calls were made against, and what teams won from the datasets available to determine whether there are possible biases; therefore, from our analysis alone, it can be hard to say that there is definitely no bias because we cannot compare aspects like whether the observed foul calls by referees would not be called by others or if there would be any disagreement from others. Referees also make many other calls throughout games that may impact outcomes, sp saying referees have no impact on who wins the game based on foul calls alone may not be accurate. Overall, datasets do not capture the full picture, such as emotions, context of whether the games are high stakes, type of play style of teams, whether players are performing to their usually potential, and much more, so drawing conclusion about possible biases from these datasets can only give patterns or generalizations.
