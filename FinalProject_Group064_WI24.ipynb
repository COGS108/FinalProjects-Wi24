## Background and Prior Work



### Introduction:
Modern e-cigarettes were introduced to the United States in 2006 and were quickly embraced by a very broad demographic. Within a few years, vape users could be found across generations and among former, current, and new smokers. The CDC reported in 2018 approximately 8.1 million Americans were using e-cigarettes and approximately 49.1 million people were using tobacco products in general. Interestingly, prevalence in the use of smoking cessation products increased between 2009 and 2018—however, only vaping and smokeless tobacco use continued to increase between 2017 and 2018..(https://www.cdc.gov/mmwr/volumes/68/wr/mm6845a2.htm?s_cid=mm6845a2_w).

Furthermore, E-cigarettes and smokeless tobacco, due to their nicotine salt content and method of ingestion, lead to different addiction patterns and health outcomes compared to traditional tobacco products. Specifically, e-cigarettes and smokeless tobacco are hypothesized to be associated with lower immediate and long-term health risks compared to traditional tobacco products. 

Despite the rapid and widespread adoption of vapes, relatively little is known about their short and long term health effects. Partly due to the range of variability in their ingredients, it is unclear whether smoking e-cigs is a reasonable alternative to traditional tobacco products (https://academic.oup.com/ntr/article/24/8/1141/6425099). In addition to the unknown health concerns, it is worth noting behavioral side effects are similarly not well understood. One study examined the difference between freebase nicotine (comparable to traditional tobacco use) and nicotine salts (found in e-cigarettes), and found that while the freebase form produced greater anxiety associated withdrawals, nicotine salts led to greater “reinforcement-related behaviors” (https://www.ncbi.nlm.nih.gov/pmc/articles/PMC10687399/). This suggests the pharmacokinetics and behavioral outcomes associated with different forms of nicotine use are worth examining further.

## Analysis:

The analysis of smoking cessation method preferences demonstrates a pronounced preference for e-cigarettes as the most popular alternative, likely due to their ease of use and familiarity. Following e-cigarettes, nicotine gums or lozenges and patches are preferred, indicating a favor for non-invasive options. The comparative popularity of Chantix/Varenicline and Zyban/Bupropion/Wellbutrin suggests a moderate acceptance of pharmaceutical aids. Conversely, nicotine nasal sprays or inhalers are the least favored, hinting at limitations in user preference or awareness. ## EDA Section3:
### If you ingest nicotine do you suffer more from mental health?
This section of our EDA aims to answer our research question where we are trying to see if an increased use in nicotine would lead to a decrease in mental health levels.

```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np



mental_health.replace(value_mappings, inplace=True)

def plot_mental_health_status(condition):
    fig, ax = plt.subplots(figsize=(10, 6))
    categories = ['A little', 'Somewhere in between', 'A lot']
    width = 0.35
    colors = ['skyblue', 'lightcoral']

    for i, user_status in enumerate(['Current user', 'Past user']):
        condition_counts = mental_health[mental_health['e-cigarette use status'] == user_status][condition].value_counts().reindex(categories).fillna(0)
        ax.bar(np.arange(len(categories)) + i * width, condition_counts, width, label=user_status, color=colors[i], edgecolor='grey')

    ax.set_xticks(np.arange(len(categories)) + width / 2)
    ax.set_xticklabels(categories)
    ax.set_xlabel(f'Level of {condition}', fontweight='bold')
    ax.set_ylabel('Frequency', fontweight='bold')
    ax.set_title(f'{condition} Levels in Current vs Past E-cigarette Users')
    ax.legend()
    plt.xticks(rotation=45)
    plt.tight_layout()
    plt.show()

plot_mental_health_status('level of anxiety')
plot_mental_health_status('level of depression')
```

Our findings reveal that past e-cigarette users display higher instances of anxiety and depression compared to those currently using e-cigarettes. This suggests a potential link between discontinuing e-cigarette use and increased mental health issues. However, the underlying causes remain to be fully understood. These results are particularly relevant to the study of nicotine ingestion methods and their impact on mental health.

### Note:
Cigarrete and E-cig smokers consistently have higher mean scores in all anxiety metrics, however this does not illustrate how the distributions may be different. To further explor these group differences, the Kruskal-Wallis test will be used to test more generally if there are significant differences between groups. This is a non-parametric statistical test for comparing group differences when assumptions about the distributions cannot be made.

```python
# Group data by 'nicotine' column
grouped_data = anx_nicotine.groupby('nicotine')

# Define the anxiety variables to compare
anxiety_vars = ['anxiety_disorder_ever', 'anxiety_frequency', 'anxiety_level', 'anxiety_severity_cat']

# Iterate over each anxiety variable
for var in anxiety_vars:
    print(f"\nKruskal-Wallis Test for {var}:")
    print("=========================================")
    
    # Extract data for the current variable
    data_groups = [group[var] for _, group in grouped_data]
    
    # Perform Kruskal-Wallis test
    stat, p_value = kruskal(*data_groups)
    
    # Print results
    print(f"Kruskal-Wallis Statistic: {stat}")
    print(f"P-value: {p_value}")
    
    # Interpret the results
    if p_value < 0.05:
        print("The Kruskal-Wallis test is statistically significant, indicating that there are significant differences among the groups.")
    else:
        print("The Kruskal-Wallis test is not statistically significant, indicating that there are no significant differences among the groups.")
```

### Note:
The Kruskal-Wallis results show that statistically significant differences exist between all of groups with various relationships to nicotine. However, to further explore these differences, tests will need to be done comparing each individual group to every other and across each anxiety variable. For this, the Mann-Whitney U test can be used. This is another non-parametric test that determines whether statistically significant differences exist between two independent groups. Additionally, because multiple comparisons are being done the significance of the Mann-Whitney U tests will be ran through Bonferroni Correction calculation, which adjusts significance levels in a way that reduces Type I errors.

```python
# Define the anxiety variables
anxiety_vars = ['anxiety_disorder_ever', 'anxiety_frequency', 'anxiety_level', 'anxiety_severity_cat']

# Create an empty list to store the pairwise comparison results
anx_pairwise_results = []

# Iterate over each anxiety variable
for var in anxiety_vars:
    
    # Extract data for the current variable
    data_groups = [group[var] for _, group in grouped_data]
    
    # Perform pairwise Mann-Whitney U tests
    pairs = combinations(range(len(data_groups)), 2)
    for i, j in pairs:
        group1 = data_groups[i]
        group2 = data_groups[j]
        stat, p_value = mannwhitneyu(group1, group2, alternative='two-sided')
        
        if p_value < 0.05:
            anx_pairwise_results.append((var, nicotine_order[i], nicotine_order[j], p_value))
        else:
            pass
            
# Perform multiple testing correction (e.g., Bonferroni correction)
corrected_p_values = multipletests([result[3] for result in anx_pairwise_results], method='bonferroni')[1]

# Print significant pairwise comparisons after multiple testing correction
print("\nSignificant Pairwise Comparisons (Bonferroni Corrected):")
print("===========================================================")
for i, result in enumerate(anx_pairwise_results):
    if corrected_p_values[i] < 0.05:
        print(f"{result[0]}: {result[1]} vs {result[2]} (p-value: {corrected_p_values[i]})")
```

### Note:
The results of this test show that there are many significant differences between the various nicotine-use groups. To better understand these relationships, the distribution of responses for each group and variable will be plotted, with markers to indicate which groups were significantly different. It is meaningful to note that these tests, because normal distributions were not assumed, use both the median and variance of the distributions to calculate the test statistic. A plot showing the median value as well as the distribution will help to better illustrate theses data.


```python
# Define the order of nicotine groups for consistent plotting
nicotine_order = ['smoker', 'vaper', 'smokeless', 'no nicotine']

# Define the order of anxiety variables for consistent plotting
anxiety_vars = ['anxiety_disorder_ever', 'anxiety_frequency', 'anxiety_level', 'anxiety_severity_cat']
anxiety_vars_dict={'anxiety_disorder_ever': 'Anxiety Disorder Ever', 
                   'anxiety_frequency': 'Anxiety Frequency', 
                   'anxiety_level': 'Anxiety Level', 
                   'anxiety_severity_cat': 'Severity Category'}


# Group data by 'nicotine' column
grouped_data = anx_nicotine.groupby('nicotine')

# Create an empty list to store the pairwise comparison results
pairwise_results = []

# Iterate over each anxiety variable
for var in anxiety_vars:
    
    # Extract data for the current variable
    data_groups = [group[var] for _, group in grouped_data]
    
    # Perform pairwise Mann-Whitney U tests
    pairs = combinations(range(len(data_groups)), 2)
    for i, j in pairs:
        group1 = data_groups[i]
        group2 = data_groups[j]
        stat, p_value = mannwhitneyu(group1, group2, alternative='two-sided')
    
        if p_value < 0.05:
            pairwise_results.append((var, nicotine_order[i], nicotine_order[j], p_value))
        else:
            pass
                    
# Perform multiple testing correction (e.g., Bonferroni correction)
corrected_p_values = multipletests([result[3] for result in pairwise_results], method='bonferroni')[1]

# Extract significant pairwise comparisons after multiple testing correction
significant_pairs = [pair for i, pair in enumerate(anx_pairwise_results) if corrected_p_values[i] < 0.05]


# Create a figure and axis object
fig, axes = plt.subplots(2, 2, figsize=(9, 6), sharey=False)

# Flatten axes for easier iteration
axes = axes.flatten()

# Iterate over each anxiety variable
for i, anxiety_var in enumerate(anxiety_vars):
    ax = axes[i]
    sns.violinplot(x='nicotine', y=anxiety_var, data=anx_nicotine, ax=ax, order=nicotine_order, width=.7, palette='husl')
    ax.set_title(f'{anxiety_vars_dict[anxiety_var]}')
    if i == 0 or i == 2:
        ax.set_ylabel('Response')  # Add x-axis label only for the lower two plots
    else:
        ax.set_ylabel('')
    ax.set_xlabel('')  # Add x-axis label
    
    if i==0:
        ax.set_ylim(top=3, bottom=0)
        ax.set_yticks([1, 2])  # Set the positions of the ticks
        ax.set_yticklabels(['No', 'Yes'])
    else:
        ax.set_ylim(top=5, bottom=0)

    # Remove outline
    ax.spines['top'].set_visible(False)
    ax.spines['right'].set_visible(False)

    # Add grid to the plot
    ax.grid(linestyle='-', alpha=0.1, color='black')

    # Get y-position for each nicotine group
    y_pos = anx_nicotine.groupby('nicotine')[anxiety_var].median()

    # Highlight significant differences
    offset_factor = 0.15  # Adjust this value for spacing between lines
    offset = 0
    sig_pairs_for_var = [pair for pair in significant_pairs if pair[0] == anxiety_var]
    for pair in sig_pairs_for_var:
        group1 = pair[1]
        group2 = pair[2]
        p_value = pair[3]
        if p_value < 0.05:
            ax.hlines(y_pos[group1] + offset, nicotine_order.index(group1), nicotine_order.index(group2), colors='black', linewidth=1.5)
            offset += offset_factor  # Increase offset for the next line

# Add a single legend entry indicating statistical significance
custom_legend = [Line2D([0], [0], color='black', lw=1.5, label=' p < .05')]

# Add legend outside the subplots and position it in the center
fig.legend(handles=custom_legend, loc='upper center', bbox_to_anchor=(0.515, .49), ncol=1, fontsize='large')

fig.suptitle('Nicotine-Group Differences in Anxiety', fontsize=16, fontweight='bold')

# Adjust layout
plt.tight_layout()
fig.subplots_adjust(top=.86, hspace=.4)

# Show plot
plt.show()
```

Analysis of Anxiety and Nicotine Use:
- Many significant differences between the nicotine-use groups exist, but there is a consistent trend: while smokers and vapers' anxiety metrics differ from smokeless tobacco users and those who abstain from nicotine altogether, it seems that there is little to no significant difference in anxiety *between* smokers and vapers. This is surprising because the mode of nicotine ingestion is very different (one uses freebase nicotine and the other uses nicotine salts).

```python
grouped_data = mhm_nicotine.groupby('nicotine')

for var in mhm_vars:
    
    data_groups = [group[var] for _, group in grouped_data]
    
    # Perform Kruskal-Wallis test
    stat, p_value = kruskal(*data_groups)
    
   
# Create an empty list to store the pairwise comparison results
pairwise_results = []

# Iterate over each anxiety variable
for var in mhm_vars:
    
    # Extract data for the current variable
    data_groups = [group[var] for _, group in grouped_data]
    
    # Perform pairwise Mann-Whitney U tests
    pairs = combinations(range(len(data_groups)), 2)
    for i, j in pairs:
        group1 = data_groups[i]
        group2 = data_groups[j]
        stat, p_value = mannwhitneyu(group1, group2, alternative='two-sided')
        
        if p_value < 0.05:
            pairwise_results.append((var, nicotine_order[i], nicotine_order[j], p_value))
        else:
            pass
            
# Perform multiple testing correction (e.g., Bonferroni correction)
corrected_p_values = multipletests([result[3] for result in pairwise_results], method='bonferroni')[1]

# Print significant pairwise comparisons after multiple testing correction
for i, result in enumerate(pairwise_results):
    if corrected_p_values[i] < 0.05:
        print(f"{result[0]}: {result[1]} vs {result[2]} (p-value: {corrected_p_values[i]})")



# Define the order of nicotine groups for consistent plotting
nicotine_order = ['smoker', 'vaper', 'smokeless', 'no nicotine']


# Group data by 'nicotine' column
grouped_data = mhm_nicotine.groupby('nicotine')

# Create an empty list to store the pairwise comparison results
pairwise_results = []

# Iterate over each anxiety variable
for var in mhm_vars:

    # Extract data for the current variable
    data_groups = [group[var] for _, group in grouped_data]
    
    # Perform pairwise Mann-Whitney U tests
    pairs = combinations(range(len(data_groups)), 2)
    for i, j in pairs:
        group1 = data_groups[i]
        group2 = data_groups[j]
        stat, p_value = mannwhitneyu(group1, group2, alternative='two-sided')
        if p_value < 0.05:
            pairwise_results.append((var, nicotine_order[i], nicotine_order[j], p_value))
        else:
            pass
            
# Perform multiple testing correction (e.g., Bonferroni correction)
corrected_p_values = multipletests([result[3] for result in pairwise_results], method='bonferroni')[1]

# Extract significant pairwise comparisons after multiple testing correction
significant_pairs = [pair for i, pair in enumerate(pairwise_results) if corrected_p_values[i] < 0.05]


# Create a figure and axis object
fig, axes = plt.subplots(2, 2, figsize=(9, 6), sharey=True)

# Flatten axes for easier iteration
axes = axes.flatten()

# Iterate over each anxiety variable
for i, mhm_var in enumerate(mhm_vars):
    ax = axes[i]
    sns.violinplot(x='nicotine', y=mhm_var, data=mhm_nicotine, ax=ax, order=nicotine_order, width=.7, palette='colorblind', saturation=.6, linewidth=1)
    ax.set_title(f'{mhm_vars_dict[mhm_var]}')
    if i == 0 or i == 2:
        ax.set_ylabel('Response')  # Add x-axis label only for the lower two plots
    else:
        ax.set_ylabel('')
    ax.set_xlabel('')  # Add x-axis label

    # if i==0:
    #     ax.set_ylim(top=3, bottom=0)
    #     ax.set_yticks([1, 2])  # Set the positions of the ticks
    #     ax.set_yticklabels(['No', 'Yes'])
    # else:
    ax.set_ylim(top=4.5, bottom=.5)

    # Remove outline
    ax.spines['top'].set_visible(False)
    ax.spines['right'].set_visible(False)

    # Add grid to the plot
    ax.grid(linestyle='-', alpha=0.1, color='black')

    # Get y-position for each nicotine group
    y_pos = mhm_nicotine.groupby('nicotine')[mhm_var].median()

    # Highlight significant differences
    offset_factor = 0.15  # Adjust this value for spacing between lines
    offset = 0
    sig_pairs_for_var = [pair for pair in significant_pairs if pair[0] == mhm_var]
    for pair in sig_pairs_for_var:
        group1 = pair[1]
        group2 = pair[2]
        p_value = pair[3]
        if p_value < 0.05:
            ax.hlines(y_pos[group1] + offset, nicotine_order.index(group1), nicotine_order.index(group2), colors='black', linewidth=1.5)
            offset += offset_factor  # Increase offset for the next line

# Add a single legend entry indicating statistical significance
custom_legend = [Line2D([0], [0], color='black', lw=1.5, label=' p < .05')]

# Add legend outside the subplots and position it in the center
fig.legend(handles=custom_legend, loc='upper center', bbox_to_anchor=(0.515, .49), ncol=1, fontsize='large')

fig.suptitle('Nicotine-Group Differences in Mental Health', fontsize=16, fontweight='bold')

# Adjust layout
plt.tight_layout()
fig.subplots_adjust(top=.86, hspace=.4)

# Show plot
plt.show()
```
