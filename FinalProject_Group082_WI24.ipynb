# Abstract

This project aimed to understand how UCSD's Basic Needs Center (BN) helps students who face food insecurity. Our study involved analyzing UCOP Basic Needs Dashboard Data and conducting a survey to identify how various BN services, including the Calfresh office, Food Recovery Network (FRN), and Triton Food Pantry (TFP), affect students' experiences with food insecurity. Our initial findings revealed that while many students were aware of these services, usage varied across different demographic groups. ## Background and Prior Work

42% of UCSD students experienced food insecurity in 2022<a name="cite_note-1"></a>[<sup>1</sup>](#cite_note-1). The UCSD Basic Needs Center seeks to assist students with food, housing, and financial insecurity. Services such as the Triton Food Pantry (TFP), Food Recovery Network (FRN), and the Calfresh office have offered services focusing on reducing food insecurity within the student population through a variety of methods. Triton Food Pantry offers a free, point-based system where students can pick out a variety of newly-purchased food and ingredients each week. Food Recovery Network offers a similar system, but with recovered food items from local grocery stores, with no limits on what students can take. The Calfresh office offers a different valuable resource, where students can get assistance with their California EBT (aka food stamps) application.

The Basic Needs Center has collected data on how their services assist thousands of students every year, and the demographic breakdown of those students. However, they have not been able to analyze this data in combination, viewing how students might use one resource in addition to another. Additionally, they have yet to see how students may be benefiting from their services after they have interacted with them directly (i.e., how many students are actually being approved for Calfresh after getting help applying).

It is easy to find research on rates of food insecurity within (and outside of) higher education, such as the UCOP’s Basic Needs Initiative<a name="cite_note-2"></a>[<sup>2</sup>](#cite_note-2), which includes a dashboard displaying rates of food and housing insecurity<a name="cite_note-1"></a>[<sup>1</sup>](#cite_note-1) broken down by campus and identities. However, there has been less research on how Basic Needs services on college campuses have assisted this food insecurity. This is especially relevant since the implementation of California Assembly Bill 132, which gave one-time funding to California Community Colleges to provide comprehensive basic needs services<a name="cite_note-3"></a>[<sup>3</sup>](#cite_note-3). Externally from universities, there has been data collected on similar services, such as surveys on EBT effectiveness<a name="cite_note-4"></a>[<sup>4</sup>](#cite_note-4) and research on the impact of food banks<a name="cite_note-5"></a>[<sup>5</sup>](#cite_note-5). It is necessary to analyze the impact of these campus services on a smaller, more focused, level. This will assist in finding out what services work, what doesn’t, what needs more funding, and what problems need to be solved from a different avenue than Basic Needs Centers. This information will especially be essential in advocating for more funding and resources to be put towards centers such as UCSD BN.

### Service Awareness Usage

We will now look at the general usage of different services.

```python

def plot_service_responses(df, column_name):
    # Define the order of the responses
    order = ["I have heard of this service, but never used it.", "I've never heard of this service.", "I have heard of this service and used it."]
    # Define colors for each response
    colors = {"I have heard of this service, but never used it.": "blue", "I've never heard of this service.": "red", "I have heard of this service and used it.": "green"}
    
    # Calculate the percentage of each response type
    response_percentages = df[column_name].value_counts(normalize=True) * 100
    response_percentages = response_percentages.reindex(order).fillna(0)  # Ensure all categories are present and fill missing values with 0
    
    # Convert Series to DataFrame for easier plotting
    response_percentages = response_percentages.reset_index()
    response_percentages.columns = [column_name, 'Percentage']
    
    # Plot
    plt.figure(figsize=(10, 6))
    sns.barplot(x='Percentage', y=column_name, data=response_percentages, hue=column_name, legend=False, order=order, palette=colors)
    plt.title(f'Response Distribution for {column_name}')
    plt.xlabel('Percentage')
    plt.ylabel('Responses')
    plt.show()

# Columns related to Basic Needs services
service_columns = ['Basic Needs Hub', 'TFP', 'FRN', 'Calfresh Team']

for column in service_columns:
    plot_service_responses(Group82_df, column)
```

### Relationships

We will create contingency tables to explore relationships between groups and service usage. As Race and Other Identities are the two major characteristics of concern with differing levels of food security between grounds, we will look at those two.

```python
def plot_relationship_heatmap(df, demographic_variable, service_variable):
    # Creating a contingency table
    crosstab = pd.crosstab(df[demographic_variable], df[service_variable])
    
    # Normalize the crosstab to show proportions rather than counts
    crosstab_normalized = crosstab.div(crosstab.sum(axis=1), axis=0)
    
    # Plotting the heatmap
    plt.figure(figsize=(12, 8))
    sns.heatmap(crosstab_normalized, annot=True, cmap="Blues", fmt=".2f")
    plt.title(f'Relationship between {demographic_variable} and {service_variable}')
    plt.ylabel(demographic_variable)
    plt.xlabel(service_variable)
    plt.xticks(rotation=45, ha='right')
    plt.yticks(rotation=0)
    plt.show()

demographic_variables = ['Race', 'Other Identities', 'Major', 'Year']
service_variables = ['Basic Needs Hub', 'TFP', 'FRN', 'Calfresh Team']
```

```python
our_survey = pd.read_csv('group82SurveyNew.csv')
ucop = pd.read_csv('UCUES_survey_data.csv')

our_survey = our_survey.drop(columns=['Name', 'Last modified time'])

our_survey = our_survey.drop(columns=['Email', 'Start time', 'Completion time'])
ucop = ucop.drop(columns=['Unnamed: 0'])

our_survey.columns = [
    'ID', 'Gender', 
    'Race', 'Other Identities', 'Major', 'Year', 'Basic Needs Hub', 
    'TFP', 'FRN','Calfresh Team','TFP2','FRN2','Calfresh Team2', 'Applied for Calfresh', 
    'Approved for Calfresh', 'Calfresh Team Assistance', 'Other Feedback'
]

#While comparing data to our survey, we will focus on the most recent dataset, 2022. 
ucop2022 = ucop[ucop['Year'] == 2022]
#Since we did not measure levels of food security, we will condense into just "food security" or "food insecurity"
ucop2022['Food Security'] = ucop2022['Food Security'].replace({'Very low food security':'Food insecurity','Low food security':'Food insecurity'})
#now, we can drop rows that say 'Food Security', as they can be inferred
ucop2022 = ucop2022[ucop2022['Food Security'] != 'Food security']
#then, we can drop columns without useful info to us
ucop2022 = ucop2022.drop(columns=['Food Security', 'Year'])
ucop2022 = ucop2022.groupby('Characteristics').sum().reset_index()
#to match ucop data later on, we will change years in our survey to upper/lower div 
our_survey['Year'] = our_survey['Year'].replace({'1st year':'Lower division', '2nd year':'Lower division', '3rd year ':'Upper division', '4th year':'Upper division','5+ year':'Upper division'})

#Parse out our data to match the format of the UCOP data

def set_bool(col, value):
    our_survey[value] = our_survey[col].str.contains(value)
    
    
 #splitting out race/ethnicity and other identities so we can look at them seperately   
races = ['Asian','Native Hawaiian or Pacific Islander', 'Black or African American','Hispanic or Latino',
         'American Indian or Alaskan Native','White','Other']
identities = ['Transfer Student','International Student','LGBTQIA+','First Generation Student','Undocumented Student']
genders = ['Man','Woman','Non-binary','Other']
years = ['Upper division','Lower division','Graduate Student']

allIdentities = races + identities + genders + years + ['Grand Total']

our_survey['Grand Total'] = 1

for identity in races:
    set_bool('Race',identity)
    
for identity in identities:
    set_bool('Other Identities',identity)
    
for gender in genders:
    set_bool('Gender', gender)
    
for year in years:
    set_bool('Year', year)
    
raceDict = {}
combdf = pd.DataFrame(columns=['ID', 'Basic Needs Hub', 'TFP','FRN','Calfresh Team','TFP2','FRN2','Calfresh Team2', 'Applied for Calfresh', 'Approved for Calfresh', 'Calfresh Team Assistance','identity'])
for identity in allIdentities:
   raceDict[identity] = our_survey.loc[our_survey[identity] == 1]
   raceDict[identity] = raceDict[identity][['ID', 'Basic Needs Hub', 'TFP','FRN','Calfresh Team','TFP2','FRN2','Calfresh Team2', 'Applied for Calfresh', 
    'Approved for Calfresh', 'Calfresh Team Assistance']]
   raceDict[identity]['identity'] = identity
   combdf = pd.concat([combdf,raceDict[identity]])

   
evalcols = ['Basic Needs Hub', 'TFP','FRN','Calfresh Team','TFP2','FRN2','Calfresh Team2', 'Applied for Calfresh', 'Approved for Calfresh', 'Calfresh Team Assistance']
counts = {}
for col in evalcols:
   counts[col] = combdf.groupby([col, 'identity']).size().reset_index()
   counts[col].columns = ['Source', 'Characteristics', 'number']
  
for key in counts:  
  race_totals = counts[key].groupby('Characteristics')['number'].transform('sum')
  counts[key]['Percentage'] = (counts[key]['number'] / race_totals) * 100
  
  hearddf = counts[key][counts[key]['Source'].isin(['I have heard of this service and used it.', 'I have heard of this service, but never used it.'])]
  #here, we will condense heard of and used into one category that represents awareness
  hearddf['Source'] = 'Heard of'
  hearddf = hearddf.groupby(['Source','Characteristics']).sum().reset_index()
  counts[key] = pd.concat([counts[key], hearddf])
```

Now all we have to do is fix the names of a few things...

```python
ucop2022['Characteristics'] = ucop2022['Characteristics'].replace({'African American':'Black or African American','Asian/Pac Isl':'Asian', 'International':'International Student','Hispanic/ Latinx':'Hispanic or Latino', 'American Indian':'American Indian or Alaskan Native', 
                                                                   'Male':'Man', 'Female':'Woman','Nonbinary':'Non-binary','LGBTQ':'LGBTQIA+','First Generation':'First Generation Student','Transfer':'Transfer Student'})
ucop2022['Source'] = 'Food Insecure'
ucop2022 = ucop2022.rename(columns={'% of Total Number of Records':'Percentage'})
#now, merge ucop and our survey data
combinedTFP = pd.concat([counts['TFP'],ucop2022])
combinedTFP['Source'] = combinedTFP['Source'].replace({'I have heard of this service and used it.':'Used TFP', 'Heard of':'Heard of TFP'})
combinedFRN = pd.concat([counts['FRN'],ucop2022])
combinedFRN['Source'] = combinedFRN['Source'].replace({'I have heard of this service and used it.':'Used FRN', 'Heard of':'Heard of FRN'})
combinedCF = pd.concat([counts['Calfresh Team'],ucop2022])
combinedCF['Source'] = combinedCF['Source'].replace({'I have heard of this service and used it.':'Used Calfresh Team', 'Heard of':'Heard of Calfresh Team'})
combinedBN = pd.concat([counts['Basic Needs Hub'],ucop2022])
combinedBN['Source'] = combinedBN['Source'].replace({'I have heard of this service and used it.':'Used Basic Needs Hub', 'Heard of':'Heard of Basic Needs Hub'})


combined = {'TFP':combinedTFP, 'FRN':combinedFRN, 'Calfresh Team':combinedCF, 'Basic Needs Hub':combinedBN}
racePlots = {}
genderPlots = {}
otherPlots = {}
yearPlots = {}
```

### Plotting the data
Ok! Now we can plot our data and see some results.

```python
for combination in combined:
    #plot race + ethnicity
    raceData = combined[combination][combined[combination]['Characteristics'].isin(('Grand Total','Hispanic or Latino','Asian','White'))]
    fig = plt.figure()
    ax = sns.barplot(data=raceData, x='Characteristics', y='Percentage', hue='Source', hue_order=['Food Insecure','Heard of ' + combination,'Used ' + combination], order=['Grand Total','Hispanic or Latino','Asian','White'])
    plt.legend(loc='lower right', fontsize='small')
    plt.title(combination + ' Usage by Race and Ethnicity')
    racePlots[combination] = fig
    for i in ax.containers:
        ax.bar_label(i,)
    plt.close()
    
    #plot gender
    genderData = combined[combination][combined[combination]['Characteristics'].isin(('Grand Total','Woman','Man','Non-binary'))]
    fig = plt.figure()
    ax = sns.barplot(data=genderData, x='Characteristics', y='Percentage', hue='Source', hue_order=['Food Insecure','Heard of ' + combination,'Used ' + combination], order=['Grand Total','Woman','Man','Non-binary'])
    plt.legend(loc='lower right', fontsize='small')
    plt.title(combination + ' Usage by Gender')
    genderPlots[combination] = fig
    for i in ax.containers:
        ax.bar_label(i,)
    plt.close()
    
    #plot other
    otherData = combined[combination][combined[combination]['Characteristics'].isin(('Grand Total','International Student','Transfer Student','LGBTQIA+', 'First Generation Student'))]
    fig = plt.figure()
    ax = sns.barplot(data=otherData, x='Characteristics', y='Percentage', hue='Source', hue_order=['Food Insecure','Heard of ' + combination,'Used ' + combination], order=['Grand Total','International Student','Transfer Student','LGBTQIA+', 'First Generation Student'])
    plt.legend(loc='lower right', fontsize='small')
    plt.title(combination + ' Usage by Other Identities')
    plt.xticks(rotation=45)
    otherPlots[combination] = fig
    for i in ax.containers:
        ax.bar_label(i,)
    plt.close()
    
    #plot upper/lower div
    yearData = combined[combination][combined[combination]['Characteristics'].isin(('Grand Total','Lower division', 'Upper division'))]
    fig = plt.figure()
    ax = sns.barplot(data=yearData, x='Characteristics', y='Percentage', hue='Source', hue_order=['Food Insecure','Heard of ' + combination,'Used ' + combination], order=['Grand Total','Lower division', 'Upper division'])
    plt.legend(loc='lower right', fontsize='small')
    plt.title(combination + ' Usage by Year')
    yearPlots[combination] = fig
    for i in ax.containers:
        ax.bar_label(i,)
    plt.close()
```

# Are food insecurity and service awareness correlated?

```python
TFPAnalyze = combinedTFP.pivot_table(index='Characteristics', columns='Source', values='Percentage', aggfunc='first').reset_index()
TFPAnalyze = TFPAnalyze.drop(columns=['I have heard of this service, but never used it.', "I've never heard of this service.","Used TFP"])
TFPAnalyze = TFPAnalyze.dropna()

FRNAnalyze = combinedFRN.pivot_table(index='Characteristics', columns='Source', values='Percentage', aggfunc='first').reset_index()
FRNAnalyze = FRNAnalyze.drop(columns=['I have heard of this service, but never used it.', "I've never heard of this service.","Used FRN"])
FRNAnalyze = FRNAnalyze.dropna()

CFAnalyze = combinedCF.pivot_table(index='Characteristics', columns='Source', values='Percentage', aggfunc='first').reset_index()
CFAnalyze = CFAnalyze.drop(columns=['I have heard of this service, but never used it.', "I've never heard of this service.","Used Calfresh Team"])
CFAnalyze = CFAnalyze.dropna()

BNAnalyze = combinedBN.pivot_table(index='Characteristics', columns='Source', values='Percentage', aggfunc='first').reset_index()
BNAnalyze = BNAnalyze.drop(columns=['I have heard of this service, but never used it.', "I've never heard of this service.","Used Basic Needs Hub"])
BNAnalyze = BNAnalyze.dropna()
```

```python
sns.scatterplot(x='Food Insecure', y='Heard of TFP', data=TFPAnalyze, s=100)
```

```python
sns.scatterplot(x='Food Insecure', y='Heard of FRN', data=FRNAnalyze, s=100)
```

```python
sns.scatterplot(x='Food Insecure', y='Heard of Calfresh Team', data=CFAnalyze, s=100)
```

```python
sns.scatterplot(x='Food Insecure', y='Heard of Basic Needs Hub', data=BNAnalyze, s=100)
```

Visually, it does not seem that there is any correlation between awareness of services and food insecurity rates. Let's check Pearson correlation to make sure.

```python
print(TFPAnalyze.corr(numeric_only=True))
print(FRNAnalyze.corr(numeric_only=True))
print(CFAnalyze.corr(numeric_only=True))
print(BNAnalyze.corr(numeric_only=True))
```

Overall, it seems like our analysis of the scatter plots is right: correlations between food insecurity and knowledge of services is weak. However, some interesting notes can still be made. All of the services do have a positive correlation. Correlation between food insecurity and awareness of TFP is almost 0; this is unsurprising, as awareness of TFP is good accross the board. The Basic Needs Hub and FRN have higher correlation, at about .3 for both. Calfresh awareness has the highest correlation with food insecurity, but is still fairly low at .4.

# Discussion and Conclusion:

## Synthesis of Findings:
Our investigation found that students are aware of the availability of Basic Needs (BN) services; however, the usage of these services varies across different demographic groups. Specifically, Asian and Hispanic/Latino students showed different levels of engagement with services such as Triton Food Pantry (TFP) and Food Recovery Network (FRN), while non-binary students showed higher usage rates in some cases. This variance highlights the importance of creating tailored outreach programs that take into account the unique needs and barriers of diverse student populations.

Our correlation analysis between service awareness and food insecurity levels revealed a more nuanced picture. While some services, such as the Calfresh Team, showed a positive association with awareness among food-insecure students, overall, the impact of service awareness on reducing food insecurity was limited. This suggests that awareness alone may not be enough to address the multiple factors that contribute to food insecurity, including stigma, accessibility, and the adequacy of aid provided.

## Extension of Previous Research:
Our study builds on existing research on campus food insecurity by providing a focused analysis of how specific Basic Needs Center services interact with student experiences of food insecurity at UCSD. By analyzing service awareness and usage within various demographic segments, we contribute nuanced insights into the effectiveness of these interventions. Additionally, our study responds to the need for more detailed investigations into the impact of legislative actions such as California Assembly Bill 132 on college campuses by offering empirical data to inform future policy and funding decisions.

## Limitations and Future Directions:
While our study provides valuable insights, it has its limitations. The self-reported nature of our survey data could introduce bias, and the specific context of UCSD may limit the generalizability of our findings. Future research should consider longitudinal studies to track changes over time and explore the causal impacts of service interventions on food insecurity. Additionally, qualitative research could further elucidate the barriers to service usage and identify unmet needs within the student population.
