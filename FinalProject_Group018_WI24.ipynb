# Research Question

What is the correlation between fluctuating gas and electric prices and electric vehicle sales in the United States?

## Background and Prior Work

The automotive industry in the United States in recent years has taken a turn which favors the emergence of electric vehicles as a favorable alternative to traditional gas cars. This is likely attributed to their lower operating costs, developed technology, and reduced environmental impact. Concurrently with the increasing interest in electric vehicles, there also happens to be a noticeable increase of gas and electricity prices. Gasoline prices have historically been on the rise due to international factors including political tensions, global oil supplies, and changes in global demand. Similarly, electricity prices have also increased over recent years due to upgrades to electricity infrastructures, a transition to renewable energy sources, and greater environmental awareness. In our research, we aim to see whether or not there is a correlation between the fluctuations in gas/electric prices and electric vehicle sales in the United States. 

In a 2017 research paper done at the University of California, Los Angeles, researchers find data and draw conclusions about the factors affecting plug-in electric vehicle sales in California from 2015 to 2015. Specifically, in Chapter 9 of their paper, they describe the correlation of gas price with plug-in electric vehicle sales. They explore this question by examining the descriptive statistics of gas prices in twenty different counties in California and monthly electric vehicle sales. From their research, with the use of fixed-effects regression models, they concluded that “the current gas price is positively associated with purchase propensity.” In their data, they discovered that a $1/gallon increase in gas price is associated with 366 greater monthly PEV purchases.” <a name="cite_ref-1"></a>[<sup>1</sup>](#cite_note-1) In regards to the effects of electricity prices, a research paper done by James B. Bushnell in 2022 describes a potential correlation between the energy prices and electric vehicle adoptions. In this paper, he researches data about residential electric prices in California and the decision to adopt EV technology. By exploring data that hones in on households charging an electric vehicle at home, he concludes that “the interpretation is that the customer places roughly one eighth the weight on electricity prices as she does on the price of gasoline.” <a name="cite_ref-2"></a>[<sup>2</sup>](#cite_note-2) The conclusion drawn from this paper may give us hints about the strength of a potential correlation between gas/electric prices and the sales of electric vehicles in the United States. 

We find that the research and conclusions reached by these research papers are heavily related to the research question we are trying to answer. We plan to extend the idea of this research by opening up our data to the entirety of the United States from 2016-2022. By generalizing the research to the entirety of the U.S., we feel that we are better able to provide insights about regional dynamics and state policies. It also gives us the opportunity to make broader conclusions about nationwide trends and potentially make inferences from events that affected the country as a whole. Furthermore, we believe that electric vehicles have only begun moving to the forefront of the automotive industry in recent years so extending our research beyond 2015 may give us more relevant information about the correlation between electric/gas prices and electric vehicle sales in the present.

## Dataset #3: EV, PHEV, and HEV Registrations

```python
sales_2016 = pd.read_csv("data/sales_2016.csv")
sales_2017 = pd.read_csv("data/sales_2017.csv")
sales_2018 = pd.read_csv("data/sales_2018.csv")
sales_2019 = pd.read_csv("data/sales_2019.csv")
sales_2020 = pd.read_csv("data/sales_2020.csv")
sales_2021 = pd.read_csv("data/sales_2021.csv")
sales_2022 = pd.read_csv("data/sales_2022.csv")
```

```python
all_sales = [sales_2016, sales_2017, sales_2018, sales_2019, sales_2020, sales_2021, sales_2022]
state = sales_2016['State']

#Pure EVs
ev_sales = pd.concat([df['Electric (EV)'] for df in all_sales], axis = 1)
ev_sales.index = state
ev_sales.columns = ['2016', '2017', '2018', '2019', '2020', '2021', '2022']
ev_sales.drop(ev_sales.index[-1], inplace=True)
ev_sales = ev_sales.replace(',', '', regex=True).astype(int)

#Plug-in Hybrids
phev_sales = pd.concat([df['Plug-In Hybrid Electric (PHEV)'] for df in all_sales], axis = 1)
phev_sales.index = state
phev_sales.columns = ['2016', '2017', '2018', '2019', '2020', '2021', '2022']
phev_sales.drop(phev_sales.index[-1], inplace=True)
phev_sales = phev_sales.replace(',', '', regex=True).astype(int)

#Hybrids
hev_sales = pd.concat([df['Hybrid Electric (HEV)'] for df in all_sales], axis = 1)
hev_sales.index = state
hev_sales.columns = ['2016', '2017', '2018', '2019', '2020', '2021', '2022']
hev_sales.drop(hev_sales.index[-1], inplace=True)
hev_sales = hev_sales.replace(',', '', regex=True).astype(int)
```

```python
# Adding a region column to the dataframes, for the 5 regions of the US and California
region = []
for i in annual_energy.index:
    if i in ['Arizona', 'New Mexico', 'Oklahoma', 'Texas']:
        region.append('Southwest')
    elif i in ['Alaska', 'Colorado','Hawaii', 'Idaho', 'Montana', 'Nevada', 'Oregon', 'Utah', 'Washington', 'Wyoming']:
        region.append('West')
    elif i in ['Illinois', 'Indiana', 'Iowa', 'Kansas', 'Michigan', 'Minnesota', 'Missouri', 'Nebraska', 'North Dakota', 'Ohio', 'South Dakota', 'Wisconsin']:
        region.append('Midwest')
    elif i in ['Alabama', 'Arkansas', 'Florida', 'Georgia', 'Kentucky', 'Louisiana', 'Mississippi', 'North Carolina', 'South Carolina', 'Tennessee', 'Virginia', 'West Virginia']:
        region.append('Southeast')
    elif i in ['California']:
        region.append('California')
    else:
        region.append('Northeast')
annual_energy['Region'] = region
include2022['Region'] = region
ev_sales['Region'] = region
phev_sales['Region'] = region
hev_sales['Region'] = region
```

### Linear Regression: Checking for Correlation


```python
region_df = {}
for region in avg_gas.columns:
    df = pd.DataFrame()
    df['Gas_Price'] = avg_gas[region]
    df['Electricity_Price'] = avg_electricity[region] 
    df['EV_Sale'] = total_sales_sum[region] 
    region_df[region] = df
```

```python

for i, region in enumerate(avg_gas.columns):
    outcome, predictors = patsy.dmatrices('EV_Sale ~ Gas_Price', region_df[region])
    model = sm.OLS(outcome, predictors)
    results = model.fit()
    print("Regression Results for " + region)
    print(results.summary())
    print("\n")
```

Looking at the OLS Regression Results for each region shows us that there is indeed a positive correlation between gas prices (the independent variable) and electrified vehicle sales (dependent variable). The p-values for each region are as follows: 
- California: 0.003
- Midwest: 0.012
- Northeast: 0.011
- Southeast: 0.011
- Southwest: 0.008
- West: 0.004

Using a base p-value of 0.05, each region is clearly below this threshold indicating that we can reject the null hypothesis.

```python
slope_intercept = {
    'California' : [-2.188e+05, 5.658e+05],
    'Midwest' : [7139.1349, 3.526e+05],
    'Northeast' : [-3.01e+05, 4.741e+05],
    'Southeast' : [-1.857e+05, 5.026e+05],
    'Southwest' : [-1.862e+05, 2.729e+05],
    'West' : [-2.603e+05, 3.113e+05]
}
# Scatter Plots for Gas Prices v EV Sales
fig, axs = plt.subplots(2, 3, figsize=(10, 8))
axs = axs.flatten()
for i, region in enumerate(avg_gas.T.index):
    axs[i].scatter(avg_gas.T.loc[region], total_sales_sum.T.loc[region])
    axs[i].set_title(region)
    axs[i].set_xlabel('Gas Prices ($/gallon)')
    axs[i].set_ylabel('EV Sales')
    axs[i].grid(True)
    xs = np.arange(region_df[region]['Gas_Price'].min(), region_df[region]['Gas_Price'].max()+ 1)
    ys = slope_intercept[region][1] * xs + slope_intercept[region][0]
    axs[i].plot(xs, ys, '--k', linewidth=1, label='Model')

plt.tight_layout()
plt.show()
```

The figure above plots the line of best fit found by the OLS Regression results (slope and intercept) for each region in order to visualize the strong correlation between gas prices and electrified vehicles sales.

## Discussion and Conclusion ##

To reiterate the purpose of our study, our research questions asks about the correlation between fluctuating gas/electricity prices and electric vehicle sales in the United States. In the beginning of our research project, we hypothesized that there is likely an existing positive correlation between gas prices and electric vehicle sales where an increase in gas prices would likely result in an increase in electric vehicle sales. In regards to electricity prices, we hypothesized that there would be an inverse relationship where an increase in electricity prices would ultimately lead to a decrease in gas prices. Upon analyzing our datasets, we conclude that there is a positive correlation between both gas and electricity prices with EV sales in the United States from 2016 to 2022.

During our linear regression analysis, we discovered that for gas and electricity prices throughout all of our regions, we find that all p-values fall under the statistical significance threshold of 0.05 meaning that we can reject the null hypothesis and conclude that the positive correlation found between our independent and dependent variables were not simply up to chance. A potential shortcoming of our findings may be reflected in the high R-squared values we found. On average, the R-squared value for gas sales vs. EV sales was 0.788 and the R-squared value for electricity sales vs. EV sales were 0.922. This means that a high amount of the variance in the dependent variable (EV sales) was a result of our independent variables (gas/electricity prices) and that there is a stronger positive correlation when looking at electricity prices than gas prices. Though this may show a strong correlation, we believe that this value may not truly be an accurate representation of the strength of the correlation due to the existence of confounding variables.

Before discussing the existence of confounding variables, it is important to note that our hypothesis inaccurately predicted that we would have an inverse relationship between electricity prices and EV sales. We were under the assumption that higher electricity prices would stray consumers away from switching to an EV alternative. A stronger correlation displayed in our found R-squared values show the opposite and brings up the possibility of existence of confounding variables. 

## Limitations ##
We believe that a potential explanation for the abnormally high correlation metrics may be a result of time. Over time, gas, electricity prices, and EV sales have all traditionally gone up due to numerous factors including global demand for oil, developing technology with EV sales, a higher concern for environmental impact, a transition to renewable energy sources, and many other potential factors. In order to most accurately find the correlation between gas/electricity prices and EV sales, we would have to take into account all of these confounding variables and more. Due to time limitations and limited access to data concerning these variables, we were unable to completely account for this limitation.

Another potential limitation we had was a lack of data from years before 2016. During our data collection process, the earliest year we could find for all three of our datasets was 2016. This means that for each of our scatter plots, we only had 7 points describing each of our regions. Initially, we felt that this would be a sufficient amount of data but the immense rise of prices for gas/electricity and increase in EV sales in recent years may give results that portray higher correlations than reality. 
 	
In order to increase the validity of our studies, something we could potentially do is to open up the range of our data to extend before 2016 and after 2022. Doing this would likely weaken the strength of the confounding variables and help us find a more accurate strength of correlation between gas/electricity prices and EV sales. If we were to extend our studies, we could also do further research on potential confounding variables and incorporate them into a multivariable OLS regression model in order to find a clearer answer for our research question.


Overall, though we were able to find a correlation and ultimately answer our research question, we recognize that there are a lot of limitations to our research project that were ultimately not addressed in our data. A larger dataset and greater emphasis on other potential confounding variables may have served as a helpful tool in more confidently answering the research question at hand. Ultimately, we conclude that although a correlation may exist between gas/electricity prices and EV sales,  our study can be extended thoroughly to research the strength of this correlation.
