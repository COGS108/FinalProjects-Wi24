## Background and Prior Work

Beginning in the late 1900s, superhero movies began to slowly gain popularity and traction, a different medium for the stories that had existed in comic books. Movies like Superman (1978) and Batman (1989) paved the way for Sam Raimi’s Spiderman movies and the X-men movies to emerge in the 2000s. From this, the Marvel Cinematic Universe (MCU) was launched in 2008 with the first Iron Man movie, the success of which incentivized the emergence of the DC Universe. <a name="cite_ref-1"></a>[<sup>1</sup>](#cite_note-1) Through various movies and franchises, the superhero genre moved from niche comic books to a profitable, industry-changing driver of pop-culture. <a name="cite_ref-2"></a>[<sup>2</sup>](#cite_note-2) Superhero movies dominated the movie box office for a decade, peaking around 2018. However, since then, there has been a general downward trend, culminating in superhero movies being ‘knocked off their pop-culture throne’ in 2023. <a name="cite_ref-3"></a>[<sup>3</sup>](#cite_note-3)

This growing disinterest in the superhero genre after its prolific success leads us to consider the factors that contributed to the box office success of the films and how that has changed over time. Work done by Daniel Parris visualized the relationship between critic scores and box office sales. Overall, he found that a 1-point increase in average critic score corresponded to a  ~$5M increase in domestic box office sales and overall superhero scores were decreasing. Parris attributed this to the ‘diminishing novelty of comic book movies’ along with ‘degradations in storytelling’. <a name="cite_ref-4"></a>[<sup>4</sup>](#cite_note-4) Parris’ work does not expand upon what constitutes degradations in storytelling, requiring further work examining specific factors that vary across movie to movie. With the lack of clearly defined objective measures or standards for 'good storytelling', we decided to focus specifically on superhero protagonist character traits to see if there are any that predict box office success and movie reception to help explain the success and slow decline of this genre. The traits and the effectiveness of their communication rely in part in the quality of storytelling, while also relying on actor performance, direction, etc. Additionally, diminishing novelty could be partially identified through certain traits occuring consistently over the years across a variety of superheros. However, it is also worth acknowledging that examining traits will not lend itself to an exhaustive explaination of trends as it does not include things like the COVID-19 pandemic.

````python
df = pd.read_csv('data/title.basics.tsv/data.tsv', sep="\t") # Load the title.basics data.tsv file into a dataframe (seperated by \t instead of ,)
df.head()                                                    # display the first few data points in this dataset
````

This is the data cleaning process where we attempt to fill in missing values to define them as missing instead of keeping them as string litereals and also select a subset of the whole dataset to use in our research. 

````python
df = df.replace('\\N', np.nan)                                                                          # Data Cleaning, every missing value is filled by a string literal '\N' so we replace it with np.nans
movies = df[(df['titleType'] == 'movie') & ~df['startYear'].isna()]                                     # Grab only the movies from this IMDb Dataset
movies = movies[(movies['startYear'].astype(int) > 2000) & (movies['startYear'].astype(int) < 2025)]    # We want to look at movies that are produced after 2000 and before 2025
movies                                                                                                  # Display the cleaned dataframe
````

````python
columns_to_drop = ['titleType', 'isAdult']       # List of columns that we want to drop
movies = movies.drop(columns=columns_to_drop)    # Dropping the columns
movies                                           # Display resulting dataset
````

## Marvel Box Office Dataset

This dataset contains information on DCEU movies and their box office performance.

````python
marvel_boxoffice  = pd.read_csv('data/marvel_box_office.csv')  # Load the marvel_box_office.csv file into a dataframe
marvel_boxoffice.rename(columns = {'Movie': 'primaryTitle'}, inplace= True)     # Rename columns for a later merge
marvel_boxoffice.head()                                        # Display first few columns of the Marvel box office dataset
````

Like our DC dataset, since we're only interested in the box office performance, we can drop all other columns aside from the Domestic Box Office and Movie to merge on later.

## Merging Datasets

To get the movies and their ratings into one singular dataframe, we will need to merge the ratings and movie information dataframes into one on the column 'tconst' since this is what represents different movies. All tconst should be unique in these two datasets.

````python
movies_ratings = movies.merge(ratings, on='tconst')      # Merge the movie info data with the ratings for each title Id
movies_ratings.head()                                    # Display the first few data points of this merged dataset
````

To get the traits for each superhero from our custom dataset with the data of the movies from the IMDb dataset we also need to merge the merged ratings and movie info dataset with our custom traits and personalities dataset to get the personalities of each hero.

````python
movie_with_traits = movies_ratings.merge(traits, on='tconst')   # Merge the final dataset of movies with the ratings with our custom traits dataset for each superhero
movie_with_traits.head()                                        # Dispaly first few datapoints and columns
````

We will then combine the DC box office dataset with our Marvel box office dataset:

````python
box_office = pd.concat([marvel_boxoffice, dc_boxoffice])
box_office.head()
````

Then, we can merge this with our movie_trait_df dataset to get our final merged dataset. 

````python
movie_trait_df = movie_with_traits.merge(box_office, how = 'left', on = 'primaryTitle')
movie_trait_df
````

## The Predictive Accuracy of our Predictors for Box Office Performance and Movie Rating

````python
print('Mean Squared Errors for Testing Predictions of Box Office Performance')
print(mean_squared_error(Y_test, predicted_test)) # Model MSE report for testing data
````

#### Root Mean Squared Error

The root mean squared errors of our predictions tell us the average of the absolute values of our difference between our predictions and our true values for box office performance without any squaring.
````python
print('Root Mean Squared Errors for Testing Predictions of Box Office Performance')
print(mean_squared_error(Y_test,predicted_test)**(1/2))
````

#### Adjusted R Squared

A simple R squared value measures what proportion of the variance in our dependent variable can be attributed to our independent variables. The adjusted R squared value that we are running does the same while taking into account the number of independent variables included as you can see through the formula below. Based on results, 51.80% of the variance in box office performance in our training data can be attributed to ethnicities (Polynesian and African American), character traits (humble, charismatic, altruistic), with the covariates of DC and runtime.

We can't run this for the testing data, since: number of observations = number of predictors 
(which is a constraint for this formula)

````python
N_train = 28 #Count observations in training dataset
P_boxoffice = 7 #Count of independent variables in model
R_squared_train_boxoffice = r2_score(Y_train, predicted_train) #R squared of training data predictions
````

````python
print('Adjusted R Squared for Training Predictions of Box Office Performance')
1 - ((1 - R_squared_train_boxoffice) * (N_train - 1) / (N_train - P_boxoffice - 1)) #Adjusted R Squared Formula
````
