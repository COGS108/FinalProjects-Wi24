# **Background and Prior Work**

On the 4th of February 2024, Grammy winners were announced for the past year. Taylor Swift took home the prestigious award for Album of the Year, becoming the most frequent recipient of the award in the process. While data-driven studies on this topic were hard to come by, there were numerous articles that used anecdotal evidence to describe the “Grammy bump” in an artist's sales directly following the awards ceremony. The official website of the Grammys tracked the sales of award-winning albums and found that nearly everyone saw a significant increase in sales in the weeks following the award show<a name="#1"><sup>1</a>. In the long term, we found a Stanford study that tracked the long-term effects of a Grammy win or nomination. The results were fascinating; while winners would tend to release music that differed from the typical sounds of their genre after a Grammy win, losers would do the exact opposite, releasing music that strayed towards the typical genre sounds<a name="#2"><sup>2</a>. The article suggests that Grammy winners were given more creative leeway after gaining the notoriety that comes from a win, while the same effect was not extended to those who were just nominated.

## Does Winning A Grammy Cause A Bump In Billboard Hot 100 Positioning?


The Billboard Hot 100 is a chart that uses data such as online streams, physical sales, and radio plays.  Therefore, if our hypothesis is correct, we believe that songs that won the grammy will place higher the week following the awards show as compared to the week before.

```python
# we need to get a list of all the values under "name" in the grammy songs dataframe
# then we can use that list to filter the billboard dataframe

# get a list of all the unique song names in the grammy songs dataframe, and then set all the values to lowercase
song_names = list(grammysongs_df['name'].unique())
song_names = [x.lower() for x in song_names]

# create a new column in the billboard dataframe that is True if the song is in the grammy songs dataframe
billboard_df['winner'] = billboard_df['name'].apply(lambda x: x.lower() in song_names)

grammy_winners = billboard_df[billboard_df["winner"] == True]
```

```python
## Next, we need to determine the weeks directly preceeding and following the awards show
import datetime

grammy_dates_df = pd.read_csv('data/dates.tsv', sep='\t')
grammy_dates_df['date'] = pd.to_datetime(grammy_dates_df['date'])
grammy_dates_after = [x - datetime.timedelta(days=1) for x in list(grammy_dates_df['date'])]
grammy_dates_before = [x - datetime.timedelta(days=8) for x in list(grammy_dates_df['date'])]

# create two new columns in the billboard dataframe, one for the week before the grammys and one for the week after
billboard_df['WeekBeforeGrammys'] = billboard_df['week'].apply(lambda x: pd.to_datetime(x) in grammy_dates_before)
billboard_df['WeekAfterGrammys'] = billboard_df['week'].apply(lambda x: pd.to_datetime(x) in grammy_dates_after)

billboard_df_of_interest = billboard_df[((billboard_df['WeekBeforeGrammys'] == True) | (billboard_df['WeekAfterGrammys'] == True)) & (billboard_df['winner'] == True)]
billboard_df_of_interest.head()
```

All of the numbers are very similar, so we can conclude that the Grammys has no positive effect on a song's placement in the Billboard Charts.

## Does Winning A Grammy Cause A Bump In Spotify Streams?

```python
winners = tuple(grammy_winners['artist'])
```

```python
#get spotify streams but with only grammy winners
spotify_df_filtered = spotify_df[spotify_df['artist'].isin(winners)]
print('There are ' + str(len(spotify_df_filtered['artist'].unique())) + ' artist that won Grammys in our Spotify dataset')
```

```python
spotify_df_filtered.head()
```

```python
grouped_df = spotify_df_filtered.groupby(['artist','week']).sum()
```

```python
dfp = grouped_df.pivot_table(index='week', columns='artist', values='streams', aggfunc='sum')

# plot only top 10 artist
top_artists = dfp.sum().nlargest(10).index.tolist()

ax = dfp[top_artists].plot(figsize=(10, 6), linewidth=2)

plt.title('Streams Over Time for Top 10 Artists', fontsize=16)
plt.xlabel('Week', fontsize=14)
plt.ylabel('Streams', fontsize=14)

plt.xticks(rotation=45)

plt.legend(loc='upper left', fontsize=10)

plt.tight_layout()
plt.show()
```

Here are the top artist streams on Spotify that have won Grammys. Now we want to check if winning a Grammy affects an artist streams.

```python
grouped_df = grouped_df.reset_index()
```

```python
february_data = grouped_df[grouped_df['week'].dt.month == 2]
#the mean of each artist in february
artist_feb_mean = february_data.groupby('artist').mean()
march_data = grouped_df[grouped_df['week'].dt.month == 3]
#the mean of each artist in march
artist_march_mean = march_data.groupby('artist').mean()
```

```python
merged = pd.merge(artist_feb_mean, artist_march_mean, on='artist', how='inner')
merged.rename(columns={'streams_x': 'feb_mean','streams_y': 'mar_mean'}, inplace=True)
merged.head()
```

```python
from scipy.stats import ttest_rel
t_statistic, p_value = ttest_rel(merged['feb_mean'], merged['mar_mean'])

alpha = 0.05

# Check if p-value is less than significance level
if p_value < alpha:
    print("The difference in mean streams between February and March is statistically significant.")
else:
    print("The difference in mean streams between February and March is not statistically significant.")
```

Here we did a T-test to check if the average streams difference between February and March is statistically significant. The difference was proven to not be statistically significant. This means we can infer that winning a Grammy doesn't increase an artist streaming numbers statistically.
