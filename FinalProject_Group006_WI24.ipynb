## Background and Prior Work

Over the years, San Diego has become notorious for its poor road infrastructure, especially the large number of potholes littering roads that don’t seem to be getting fixed. While potholes and poor road conditions cost drivers money, repairing the source of the problem is even more expensive for the city. However, poor planning, lack of transparency, and seemingly disproportionate impacts on certain communities and residents contribute to a generally negative perception of the city’s efforts. San Diego Mayor Todd Gloria has focused on this problem, saying in a tweet on January 10th, 2024 that “Potholes are a problem, but we're working harder than ever and #GettingItDone. In 2023, our crews filled 61,305 potholes. That's a 94% increase over the prior year and an astonishing 122% increase over 2021.” <a name="cite_ref-4"></a>[<sup>4</sup>](#cite_note-4) However, it is still unclear whether the sentiment towards the city’s government has changed regarding these issues.

Recent works have also highlighted the impact of road maintenance on traffic, however, there is little investigation on how road maintenance can influence residents' living satisfication. Research based on transportation by Western Virginia University economists showed causal evidence that poorly maintained roads reduce traffic speed and safety. <a name="cite_ref-5"></a>[<sup>5</sup>](#cite_note-5) The statistical analysis determined that decreasing pavement quality impacted vehicle crash rates. Similarly, research funded by the National Transportation Center @ Maryland, emphasizes that studies focus on traffic accidents mainly caused by poor road maintenance, but takes a step further to utilize roughness and rut-depth thresholds below which pavement related accidents can be reduced. <a name="cite_ref-6"></a>[<sup>6</sup>](#cite_note-6) They found out that the rut-depth values did not vary between areas with and without accidents, implying that a combination with other factors such as traffic volume, other pavement distresses, can also impact vehicle incidents. These works help us learn that road maintenance has an impact on communities and their wellbeing through vehicle safety and help us narrow down the problem to an objective that is related to our lives, and explore other ways road maintenance impacts communities.

#### Text Preprocessing for Description Feature

```python
# how many nonresponses
pothole_clean['public_description'].isnull().sum()
# Convert all words in the 'public_description' to lowercase
pothole_clean['public_description'] = pothole_clean['public_description'].str.lower()
```

```python
# Function to remove punctuation and tokenize text
def remove_punct_and_tokenize(text):
    # Convert input to string
    text = str(text)
    # Tokenize the text
    tokens = word_tokenize(text)
    # Filter out punctuation and stopwords
    tokens = [word for word in tokens if word.isalnum()]
    return ' '.join(tokens)
```

```python
# Remove punctuation from the 'public_description'
pothole_clean['public_description'] = pothole_clean['public_description'].apply(remove_punct_and_tokenize)
pothole_clean['public_description'].to_frame().head()
```

```python
# check for frequency of words of interest

#define function to find word frequency
def word_frequency(df, column_name, word):
    """Calculate the frequency of a word within a pandas DataFrame column."""
    
    # Tokenize the text in the specified column
    tokens = df[column_name].apply(word_tokenize).sum()

    # Calculate frequency distribution
    freq_dist = FreqDist(tokens)

    # Return frequency of the word
    return freq_dist[word]

# try out frequency function for a random word
word_frequency(pothole_clean, 'public_description', 'carmel')
```

#### 5. Sentiment Analysis
To get a measure of overall sentiment in our text, we'll compare our text to the VADER lexicon.

compound score - metric that calculates sum of all the lexicon ratings and normalizes between -1 (most extreme negative) and +1 (most extreme positive)

positive: compound >= 0.05

neutral: -0.05 < compound < 0.05

negative : compound <= -0.05

```python
# get list of the 'sentences' (responses) from each individual
public_description_list = list(pothole_clean['public_description'].values)
```

```python
# create function that will output dataframe 
# that stores sentiment information
def get_sentiments(input_list):
    
    output = pd.DataFrame()

    for sentence in input_list:
        ss = analyser.polarity_scores(sentence)
        ss['sentence'] = sentence
        output = pd.concat([output, pd.DataFrame([ss])], ignore_index=True)


    return output
```

```python
# get sentiment measures
public_description_sentiments = get_sentiments(public_description_list)
```

```python
# removing no response values here
public_description_sentiments = public_description_sentiments[public_description_sentiments['sentence'] != 'No response']
```

```python
# take a look at the output (negative first)
public_description_sentiments.sort_values(by='compound', ascending=True).head(10)
```

From this glance at the data categorized as most negative, we can see that for the most part the sentiment analysis was accurate in labeling the negative sentiments.

### Multiple Linear Regression

```python
#extracting dates and months of requests
pothole_clean['Month'] = pd.to_datetime(pothole_clean['date_requested_clean']).dt.month
pothole_clean['Year'] = pd.to_datetime(pothole_clean['date_requested_clean']).dt.year
# replacing month to numerical value
rainfall_merge = rainfall_df.replace({'Jan': 1, 'Feb': 2, 'Mar': 3, 'Apr': 4, 'May': 5, 'Jun': 6, 'Jul': 7,
                    'Aug': 8, 'Sep': 9, 'Oct': 10, 'Nov': 11, 'Dec':12})
#convert data type to integer
rainfall_merge['Year'] = rainfall_merge['Year'].astype('int')
rainfall_merge['Month'] = rainfall_merge['Month'].astype('int')
```

```python
# count requests for each month of each year
request_count = pothole_clean.groupby(['Year', 'Month']).count()['service_request_id'].reset_index()
request_count.rename(columns = {'service_request_id': 'count'}, inplace = True)
request_count.head()
```

```python
#merging to view monthly rainfall and request count
merged_df = request_count.merge(rainfall_merge, how = 'inner', left_on = ['Year', 'Month'], right_on = ['Year', 'Month'])
merged_df.head()
```

```python
import patsy
import statsmodels.api as sm

import warnings
warnings.filterwarnings('ignore')

#multiple linear regression on month, year, and rainfall
outcome, predictors = patsy.dmatrices('count ~ Month + Year + Rainfall', merged_df)
mod = sm.OLS(outcome, predictors)
res = mod.fit()
print(res.summary())
```

The results from this linear regression indicate that month, year, and rainfall are strong predictors of the number of repair requests. An R-squared value of 0.731 indicates that approximately 73.1% of the variance in the number of requests can be determined by the features month, year, and rainfall. Additionally, all the features have low p-values, suggesting they are significant features.

We will conduct a pearson correlation test to see if there is a significant relationship between the amount of rainfall and the number of requests. The null hypothesis states that the correlation coefficient for rainfall and number of requests is zero, meaning there is no relationship. The alternative hypothesis states that the correlation coefficient is not equal to zero, or that there could be a relatonship.

```python
# Pearson correlation hypothesis test between rainfall and request count
from scipy.stats import pearsonr
corr_coefficient, p_value = pearsonr(merged_df['Rainfall'], merged_df['count'])

# Define significance level
alpha = 0.05

data = {'Correlation Coefficient': corr_coefficient, 'p-value': p_value}

# Convert the dictionary into a DataFrame
result_df = pd.DataFrame(data, index=['Rainfall (in) and Request Count'])
result_df
```

```python
if p_value < alpha:
    print("Reject the null hypothesis. There is evidence to suggest there is a significant relationship between monthly rainfall and repair requests.")
else:
    print("Fail to reject the null hypothesis. There is not enough evidence to conclude there is significant relationship between monthly rainfall and repair requests.")
```

The results from the pearson correlation suggest that there is a small, positive linear relationship between rainfall and the number of requests, and a p-value of 0.01 indicates that there is a significant evidence to suggest there is a relationship between the two variables.
