Main factors influencing stock prices include:

1. Market Dynamics: Broader economic indicators and global financial trends often overshadow the effects of industry-specific developments on stock prices.
2. Technological Innovations: NVIDIA's valuation reflects a range of innovations beyond new machine learning models, including advancements in AI, gaming, and data centers.
3. Competition and Market Share: The actions of competitors and shifts in market share can influence investor sentiment and stock prices as much as, if not more than, technological advancements within NVIDIA.
4. Regulatory Changes: Legislative and regulatory environments affecting trade, technology, and intellectual property can impact NVIDIA's operations and investor perceptions.
5. Investor Sentiment: Psychological factors, news, and market speculation can sway stock prices independently of a company's actual performance or technological developments.

For the machine learning libraries, we will be looking at PyTorch. Again, we will be looking at PyTorch because of its popularity and because PyTorch and TensorFlow (the two most popular libraries) were released close in time to each other. We will be running a t-test to check whether or not the releases of these two softwares will affect the stock price by looking at the average stock price as well as volume of stock traded before and after their releases. We choose to use a t-test because the test will prove to us whether or not there is a significant difference between the means of two groups, in this case being the stock price and volume of stock traded before the release of a major software and after the release of a major software. 

```python
# want to find the number of days since the release of each software
days_since_ChatGPT = (dt.datetime(2024, 2, 23) - dt.datetime(2020, 6, 11)).days
days_since_PyTorch = (dt.datetime(2024, 2, 23) - dt.datetime(2016, 8, 31)).days
print('Days since the release of Chat-GPT 3.5: ', days_since_ChatGPT)
print('Days since the release of PyTorch: ', days_since_PyTorch)
```

```python
# create function to split dataframe into equal parts given a date
def Split_DF(df, split_date, num_dates):
    
    split_date = pd.to_datetime(split_date)
    
    before = df[df['Date'] < split_date][-num_dates:]
    after = df[df['Date'] >= split_date][:num_dates]
    
    return before, after

before_ChatGPT, after_ChatGPT = Split_DF(NVDA_close, '2020-06-11', days_since_ChatGPT)
before_PyTorch, after_PyTorch = Split_DF(NVDA_close, '2016-08-31', days_since_PyTorch)

# check number of rows to make sure they are the same
print(before_ChatGPT.shape[0], after_ChatGPT.shape[0])
print(before_PyTorch.shape[0], after_PyTorch.shape[0])
```

It seems as though the days since the release of each software does not correctly coincide with the amount of dates that are on NVIDIA's stock price history. We forgot to account for the fact that the stock market is not open every day! We will split the dataframe again, this time using the smaller number in the previous calculations to correctly make sure that the amount of days before and after the release of Chat-GPT and PyTorch are the same.

```python
# redo the split, taking into account days that the stock market is closed
before_ChatGPT, after_ChatGPT = Split_DF(NVDA_close, '2020-06-11', 932)
before_PyTorch, after_PyTorch = Split_DF(NVDA_close, '2016-08-31', 1882)

# check number of rows to make sure they are the same
print(before_ChatGPT.shape[0], after_ChatGPT.shape[0])
print(before_PyTorch.shape[0], after_PyTorch.shape[0])
```

Now we, want to take the average of the prices and volume traded before and after the release dates and do a t-test to see if there is any significance between the two sets.

We will start off with Chat-GPT.

```python
import scipy.stats as stats
from scipy.stats import ttest_ind

# run a t-test on close price and volume of stock traded
print(ttest_ind(before_ChatGPT['Close'], after_ChatGPT['Close']))
print(ttest_ind(before_ChatGPT['Volume'], after_ChatGPT['Volume']))
```

Given the p-values of these t-tests, there is indication of a strong significance of the release dates of Chat-GPT and the stock price of NVIDIA changing (in this case going up). Lets take a look at a t-test for the release of PyTorch.

```python
# run a t-test on close price and volume of stock traded
print(ttest_ind(before_PyTorch['Close'], after_PyTorch['Close']))
print(ttest_ind(before_PyTorch['Volume'], after_PyTorch['Volume']))
```

Given that the p-values of both of these t-tests were very low (<0.05), we can conclude that there is a significant difference between the means of the data set before and after the release of each respective software. However, this might not paint the entire story of why NVIDIA's stock price might have gone up so much. We must look at other factors. We want to normalize the prices in order to chart the movement of price on one graph. We will do this by converting all prices to a base currency. In this case, we will be using Bitcoin as our base currency. This means that instead of USD, we will be looking at a price that is normalized to Bitcoin. Additionally, we will need to make sure that there are the same number of dates in each list. In this case, we will use the first date that the price of Ethereum was tracked in our dataset as it is the newest of the three. We will then normalize all data using Min-Max scaling.

```python
# filter out the rows before 2016
NVDA_filtered = NVDA.loc[(NVDA['Date'] >= '2017-11-09')]
NVDA_filtered.reset_index(drop=True)

bit_filtered = bitcoin.loc[(bitcoin['Date'] >= '2017-11-09')]
bit_filtered.reset_index(drop=True)

# check to see if the data shapes match
print(NVDA_filtered.shape)
print(bit_filtered.shape)
print(eth.shape)
```

Here we see that we have a discrepency on the amount of dates we have because the stock market is not open everyday. In order to fix this, we will resample NVIDIA's stock price using linear interpolation to match the amount of dates as the other two.

```python
from scipy.interpolate import interp1d

# resample the data
NVDA_interp = interp1d(np.linspace(0, 1, 1581), NVDA_filtered['Close'].values, kind='linear')
NVDA_resampled = NVDA_interp(np.linspace(0, 1, 2324))

NVDA_resampled.shape
```

Now that all of our data have the same lengths, we will proceed with the normalization.

```python
# use bitcoin as base
bitcoin_as_base = {
    'bitcoin': bit_filtered['Close'],
    'ethereum': [eth / btc for eth, btc in zip(eth_close, bit_filtered['Close'].values)],
    'nvidia': [nvda / btc for nvda, btc in zip(NVDA_resampled, bit_filtered['Close'].values)]
}

# create function to do min-max scaling
def Min_Max_Normalization(data):
    minimum = min(data)
    maximum = max(data)
    
    return [(val - minimum) / (maximum - minimum) for val in data]

# use the function to normalize all prices
normalized_prices = {crypto: Min_Max_Normalization(prices) for crypto, prices in bitcoin_as_base.items()}
```

Now that we have the normalized data, its time to plot it.

```python
plt.figure(figsize=(12, 6))
for crypto, prices in normalized_prices.items():
    if crypto == 'nvidia':
        plt.plot(eth['Date'].values, prices, label=crypto, color='blue')
    elif crypto == 'bitcoin':
        plt.plot(eth['Date'].values, prices, label=crypto, color='green', alpha=.6)
    else:
        plt.plot(eth['Date'].values, prices, label=crypto, color='orange', alpha=.6)



plt.title('Normalized Price History of NVIDIA, Bitcoin, and Ethereum')
plt.xlabel('Years')
plt.ylabel('Normalized Price in Bitcoin')
plt.legend()
plt.grid(True)
plt.show()
```

Looking at the graph for the movement of prices, it looks like there does not seem to be a pattern between the two cryptocurrencies and NVIDIA. We will run a Spearman Rank correlation test to statistically prove this. Unlike the previous t-test, we will be using a Spearman test in order to check the strength and the direction of the relationship between these cryptocurrencies and NVIDIA's stock price.

```python
# perform Spearman test for Bitcoin vs NVIDIA
print('Bitcoin vs NVIDIA spearman correlation: \n', stats.spearmanr(normalized_prices['bitcoin'], normalized_prices['nvidia']))
```

```python
# perform t-test for Ethereum vs NVIDIA
print('Ethereum vs NVIDIA spearman correlation: \n', stats.spearmanr(normalized_prices['ethereum'], normalized_prices['nvidia']))
```

Interestingly enough, the Spearman correlation test provided a very low p-value when comparing Bitcoin and NVIDIA but a relatively high p-value when comparing Ethereum and NVIDIA. This could possibly mean that there is a correlation between the price of Bitcoin and the price of NVIDIA's stock. Given that the test statistic was negative, this implies that there was a negative correlation, meaning when the price of Bitcoin went down, the price of NVIDIA went up. For Ethereum, the p-value is high enough where we cannot confidently state that any correlation between the prices was more than mere chance.

Looking at the chart, there seems to be a large correlation between the release dates and the increase of stock price. Just like before, we want to run a t-test on this data.

For each release of a GPU, we will be looking at the prices roughly 6 months (2 quarters of a year) before the release and 6 months after the release of the GPU to see if there is a significant change in price. We chose 6 months as our timeframe because we felt that it was long enough such that we could see if there was any effect due to the release without being too long such that other factors could be at play. However, because there is not enough data for the last two releases, we will be omitting them as they released too recently.

```python
# iterate through GPU and create before and after sets

data = []

for date in GPU['NumericalDate']:

    gpu_name = GPU.loc[GPU['NumericalDate'] == date]
    gpu_name = gpu_name.iloc[0, 0]
    
    if gpu_name not in ['GeForce RTX 4070 Super', 'GeForce RTX 4080 Super']:
        before, after = Split_DF(NVDA_close, date, 180)
        
        # run the t-test
        statistic = ttest_ind(before['Close'], after['Close']).statistic
        pval = ttest_ind(before['Close'], after['Close']).pvalue
        
        data.append([gpu_name, statistic, pval])

# create dataframe of values
GPU_ttest = pd.DataFrame(data, columns=['GPU Name', 'T-test Statistic', 'P-value'])
GPU_ttest
```

Now that we have a dataframe of all of our t-test statistics and p-values, we want to check if any of the GPUs have a p-value of greater thatn 0.05.

```python
# see if there are any rows have a p-value of >0.05
if (GPU_ttest[GPU_ttest['P-value'] > 0.05].empty):
    print('No P-value greater than 0.05!')

plt.figure(figsize=(12,6))
ax = GPU_ttest.plot.bar(x='GPU Name', y='P-value', rot=0)
plt.xticks(rotation=90)

plt.title('P-values of T-tests on GPU release Dates')
plt.xlabel('GPU')
plt.ylabel('P-value')
plt.grid(True)
plt.show()
```

Given that all of the GPU releases have a p-value of less than 0.05, we can confidently say that they contribute a significant change in the prices of NVIDIA's stock. This may imply that the stock price of the company increased or decreased because the change in the product that they were releasing, yet another factor.
