# Hypothesis

We hypothesize that crime rates will fluctuate with the seasons, influenced by societal habits and climatic conditions. In the summer, we anticipate a spike in both home and car burglaries as well as shootings. This can be attributed to more residences being vacated for vacations and an increase in outdoor activities, which presents greater opportunities for such crimes. Urban areas, especially those bustling with tourists, might witness more petty thefts due to the denser crowds and vibrant social scene. Conversely, winter is expected to see a reduction in these outdoor crimes due to bad weather forcing many to stay indoors with the exception of a possible rise in domestic abuse incidents. As the climate becomes colder and the days become shorter, people remain indoors thus potentially heightening domestic tensions.

This seasonal crime pattern is also expected to vary geographically. For instance, when comparing New York City (NYC) and Los Angeles (LA), LA's warmer summers could lead to higher outdoor crime rates, whereas NYC's harsh winters might contribute to an increase in indoor crimes like domestic abuse. This hypothesis thus integrates the dynamics of seasonality and geography to offer a nuanced understanding of how and why crime rates might vary throughout the year.

## Los Angeles Weather Station Records (2020 - 2023)

```python
# The code of this function is partially generated by Github Copilot from the prompt "fix the bug for reading the csv file"
# Assign correct headers based on the observed structure and reload the dataset
column_headers = ['Station', 'Date', 'Measurement', 'Value', 'MFlag', 'QFlag', 'SFlag', 'Time']
# Read the CSV file with the correct headers
df_weather = pd.read_csv('USW00093134.csv', names=column_headers, skiprows=1, low_memory=False)

# Convert 'Date' to datetime
df_weather['Date'] = pd.to_datetime(df_weather['Date'], format='%Y%m%d')

# Filter the dataframe for dates between 2020-01-01 and 2023-11-09 as same as Crime data
df_weather_filtered = df_weather[(df_weather['Date'] >= '2020-01-01') & (df_weather['Date'] <= '2023-11-09')]

# Continue with the previous steps to extract TMAX and TMIN
# Extract TMAX and TMIN from Measurement
df_tmax = df_weather_filtered[df_weather_filtered['Measurement'] == 'TMAX']
df_tmin = df_weather_filtered[df_weather_filtered['Measurement'] == 'TMIN']

# Merge the TMAX and TMIN dataframes on 'Date' and 'Station'
df_merged = pd.merge(df_tmax[['Station', 'Date', 'Value']], df_tmin[['Station', 'Date', 'Value']], on=['Date', 'Station'], suffixes=('_TMAX', '_TMIN'))

# Calculate TAVG in tenths of Fahrenheit
df_merged['TAVG'] = (df_merged['Value_TMAX'] + df_merged['Value_TMIN']) / 2

# Convert TAVG from tenths of degrees to degrees Fahrenheit
df_merged['TAVG_C'] = df_merged['TAVG'] / 10


la_temps = df_merged[['Date', 'TAVG_C']].copy()
la_temps

```

```python
# Load New York Arrest 2006-2022 Data
URL = 'https://data.cityofnewyork.us/api/views/8h9b-rp9u/rows.csv?accessType=DOWNLOAD'
NY_Crime_past = pd.read_csv(URL)
NY_Crime_past.head()
```

```python
# Load New York Arrest 2023 Data
URL = 'https://data.cityofnewyork.us/api/views/uip8-fykc/rows.csv?accessType=DOWNLOAD'
NY_Crime_2023 = pd.read_csv(URL)
NY_Crime_2023.head()
```

```python
# Merge New York Arrests 2006-2022 data with 2023 data
NY_Crime = pd.concat([NY_Crime_past, NY_Crime_2023], ignore_index=True)
pd.to_datetime(NY_Crime['ARREST_DATE']).dt.year.unique()
```

```python
# Convert to 'ARREST_DATE' to datetime for consistency
NY_Crime['ARREST_DATE'] = pd.to_datetime(NY_Crime['ARREST_DATE'])

# Filter data to only 2020 to 2023
start_date = '2020-01-01'
end_date = '2023-12-31'
NY_Crime = NY_Crime[(NY_Crime['ARREST_DATE'] >= start_date) & (NY_Crime['ARREST_DATE'] <= end_date)]

# Drop not needed columns
NY_Crime.drop(columns=['PD_CD', 'KY_CD', 'LAW_CODE', 'ARREST_PRECINCT', 'JURISDICTION_CODE'], inplace=True)
pd.to_datetime(NY_Crime['ARREST_DATE']).dt.year.unique()
```

```python
def convertToAreaName(arrest_boro):
    if arrest_boro == 'B':
        return 'Bronx'
    elif arrest_boro == 'S':
        return 'Staten Island'
    elif arrest_boro == 'K':
        return 'Brooklyn'
    elif arrest_boro == 'M':
        return 'Manhattan'
    elif arrest_boro == 'Q':
        return 'Queens'
    else:
        return None

NY_Crime['AREA_NAME'] = NY_Crime['ARREST_BORO'].apply(convertToAreaName)
```

```python
replacements = {
    'CRIMINAL MISCHIEF & RELATED OF': 'CRIMINAL MISCHIEF & RELATED OFFENSES',
    'OFF. AGNST PUB ORD SENSBLTY &': 'OFF. AGNST PUB ORD SENSBLTY & RGHTS TO PRIV',
    'OTHER OFFENSES RELATED TO THEF': 'OTHER OFFENSES RELATED TO THEFT',
    'OFFENSES AGAINST PUBLIC ADMINI': 'OFFENSES AGAINST PUBLIC ADMINISTRATION'
}

# Replace values in the specified column
NY_Crime['OFNS_DESC'] = NY_Crime['OFNS_DESC'].replace(replacements)
NY_Crime.head()
```

```python
nyc_weather_cleaned = nyc_weather_filtered.dropna()
```

```python
nyc_weather_cleaned['time'] = pd.to_datetime(nyc_weather_cleaned['time'])

# Extract the date from the datetime column
nyc_weather_cleaned['date'] = nyc_weather_cleaned['time'].dt.date
nyc_weather_cleaned['date'] = pd.to_datetime(nyc_weather_cleaned['date'])

# Group by date and calculate the mean temperature for each day
daily_avg_temperature = nyc_weather_cleaned.groupby('date')['temperature_2m (°C)'].mean()

# Make daily average temperature column in df for every corresponding date
nyc_weather_cleaned['daily_avg_temp_2m (°C)'] = nyc_weather_cleaned['date'].map(daily_avg_temperature)
nyc_weather_cleaned
```

Indoor Crime Correlation: The negative value of the correlation coefficient suggests that there is an inverse relationship between temperature and indoor crime counts. In other words, as the temperature increases, the number of indoor crimes tends to decrease. The p-value is very small, which indicates that the result is statistically significant.

Outdoor Crime Correlation: The positive value of the correlation coefficient for outdoor crime suggests that there is a direct relationship between temperature and outdoor crime counts. This means that as the temperature rises, the number of outdoor crimes also tends to increase. The p-value here is also very small, indicating a statistically significant result.

```python
NY_top_20_crimes = NY_Crime['OFNS_DESC'].value_counts().head(20).index.tolist()

# Define which of the top 20 crimes are considered indoor and outdoor
NY_indoor_crimes = [
     'BURGLARY', 'FORGERY', 'DANGEROUS DRUGS', 'POSSESSION OF STOLEN PROPERTY', 
     'GAMBLING', 'RAPE', 'GRAND LARCENY', 'SEX CRIMES', "BURGLAR'S TOOLS", 'ALCOHOLIC BEVERAGE CONTROL LAW'
]

NY_outdoor_crimes = [
    'ROBBERY', 'VEHICLE AND TRAFFIC LAWS', 'INTOXICATED & IMPAIRED DRIVING', 'OTHER TRAFFIC INFRACTION',
    'PROSTITUTION & RELATED OFFENSES', 'UNAUTHORIZED USE OF A VEHICLE', 'CRIMINAL TRESPASS', 'GRAND LARCENY OF MOTOR VEHICLE',
    'DANGEROUS WEAPONS', 'CRIMINAL MISCHIEF & RELATED OFFENSES'
]

def create_crime_category(crime):
    if crime in NY_indoor_crimes:
        return 'Indoor'
    elif crime in NY_outdoor_crimes:
        return 'Outdoor'
    else:
        return None

# Categorize the crimes in the Simplified_LA dataframe
NY_Crime['CRIME_CATEGORY'] = NY_Crime['OFNS_DESC'].apply(create_crime_category)

# Remove 'Other/Unknown' category (None values)
NY_Crime = NY_Crime.dropna(subset=['CRIME_CATEGORY'])

# Plotting the count of indoor vs outdoor crimes from the top 20
NY_crime_category_counts = NY_Crime[NY_Crime['OFNS_DESC'].isin(NY_top_20_crimes)]['CRIME_CATEGORY'].value_counts()
NY_crime_category_counts.plot(kind='bar', color=['lightblue', 'grey'])
plt.title('Count of Indoor vs Outdoor Crimes for Top 20 Crimes in New York')
plt.xlabel('Crime Category')
plt.ylabel('Counts')
plt.xticks(rotation=0)
plt.show()
```
