# Abstract

Toward the goal of determining which factors are most indicative of a student's future success, we analyzed several variables—admission rate, STEM major percentage, standardized test scores, location, and ranking—and their relationship with median earnings 1 year after graduation. # Research Question

How do different university factors (such as acceptance rate, graduation rate, major breakdown, and location) predict graduates' earnings after a bachelor’s degree? We are aiming to determine the strength of correlation between selected university factors and the median earnings of graduates immediately after graduation. We will also be mindful of confounding factors; for example, a school located in a wealthier area likely produces graduates with higher earnings by attracting students with more affluent backgrounds to begin with.

## Background and Prior Work

This study aims to investigate the hypothesis that certain university characteristics, particularly acceptance rate, graduation rate, ranking, major breakdown, and location, have a statistically significant effect on graduates' wages. Understanding how such factors influence graduates' earning potential in their careers is crucial for students and employers. Larger institutions may have extensive networks and resources, potentially offering more opportunities for internships and job placements. On the other hand, the cost of education can influence graduates' financial situations, impacting their decisions about job offers and career paths.

A study by Dale and Krueger (2011) explored the earnings of graduates from various colleges and universities in the United States, finding that adjusting for pre-college characteristics decreased the differences in wages linked to the selectivity of institutions.<a name="cite_ref-1"></a>[<sup>1</sup>](#cite_note-1) This suggests that other factors beyond the university’s prestige may contribute to graduates’ earnings. Hence why we have decided to also focus on other factors such as acceptance rate, graduation rate, major breakdown, and location.

We also observed a study conducted by Bound, Lovenheim, and Turner (2009), which analyzed the intricate relationship between college selectivity and earnings.<a name="cite_ref-2"></a>[<sup>2</sup>](#cite_note-2) While Bound et al. primarily focused on selectivity, their approach involved considering acceptance rates as a pivotal variable, laying the groundwork for our investigation into the predictive power of acceptance rates on post-graduation earnings.

Additionally, Carnevale et al. (2013) investigated the economic value of college majors with a focus on how majors shape careers post-graduation.<a name="cite_ref-3"></a>[<sup>3</sup>](#cite_note-3) Their findings highlighted significant differences in earning based on the chosen field of study, which aligns with our hypothesis that universities with a predominant focus on STEM majors may produce graduates with higher wages compared to those from liberal arts universities. 

Hillman and Weichman (2016) explored the influence of location on the economic outcomes of graduates.<a name="cite_ref-4"></a>[<sup>4</sup>](#cite_note-4) Their study considered how the regional location of a university might impact graduates' earnings. This aligns with our emphasis on the geographical aspect and its potential role in shaping post-graduation financial success.

In summary, our investigation into the relationship between university factors and post-graduation earnings is informed by studies that investigate ranking (Dale and Krueger, 2011), acceptance rate (Bound, Lovenheim, & Turner, 2010), major breakdown (Carnevale et al., 2013), and location (Hillman & Weichman, 2016). By focusing on these variables, we aim to contribute valuable insights to the research conducted on graduates' earnings and what affects them.

# Hypothesis



We hypothesize that acceptance rates will have a strong inverse correlation with graduates’ earnings. A university with a lower acceptance rate means that many students apply for that college. Students apply and choose colleges based on which they believe will offer them the best education and opportunities to grow their career. Thus, we believe that low acceptance rates mean that the university offers better opportunities and education translating into higher wages post-graduation.

We expect college major breakdown to have the strongest correlation with post-graduation wages. A STEM-focused school such as MIT will produce graduates who end up working in higher paying fields and careers compared to a liberal arts college, whose prospective fields tend to be less lucrative on average than STEM jobs.

We also expect high university ranking and incoming SAT/ACT scores to both have a moderate correlation with higher wages. Highly ranked universities, as well as those with high average incoming test scores, likely have a higher density of students with higher socioeconomic status and backgrounds conducive to academic pressure, so we anticipate that these universities may produce graduates who are motivated to advance in their careers.

On the other hand, we don’t expect location to have a strong correlation. It is likely that most of the correlation we find between location and earnings is attributed to confounding factors such as state funding (some states fund public universities more than others) and wealth (wealthier areas attract students with more affluent backgrounds).

Overall, we expect acceptance rates and major breakdown to have the highest correlation with post-graduation income while location and graduation have less correlation.

## [World University Ranking](https://www.kaggle.com/datasets/mylesoneill/world-university-rankings)

We wanted to be conscious of the possible two-way correlation between income and ranking, where the income of graduates improves a school's ranking and vice versa. To do so, we added the Times dataset which includes a calcuation of universities' ranks broken down by quality of teaching staff, quality and influence of research, international outreach, and income of graduates. Then, we recalculated each school's ranking to exclude the income of graduates field, in order to generate new rankings for each university that are not two-way correlated with graduates' incomes.

````python
timesDF = pd.read_csv("data_times.csv")

# RECALCULATE RANKING EXCLUDING INCOME

# Recalculate score and filter only for years
timesDF = timesDF.loc[timesDF["year"] == 2016]
timesDF['teaching_score'] = timesDF["teaching"]*30.625
timesDF['citations_score'] = timesDF["citations"]*30.625
timesDF['research_score'] = timesDF["research"]*30.625
#timesDF["international"] = timesDF["international"].replace("-", "0")
timesDF["international"] = timesDF["international"].astype(float)
timesDF['international_score'] = timesDF["international"]*8.125
timesDF["total_score"] = timesDF["teaching_score"] + timesDF['citations_score'] + timesDF['research_score'] + timesDF['international_score']

#sort and rerank based on newly calculated score
timesDF = timesDF.sort_values('total_score', ascending=False)
timesDF = timesDF.reset_index()
timesDF['ranking'] = timesDF.index
timesDF['ranking'] = timesDF['ranking'] + 1
timesDF.drop(["index"], axis=1)
````

We conducted an isolated study of each university factor against `Median Earnings After 1 Year` by representing their relationships visually and through linear regression, and were able to gain a deeper understanding of how much (or how little) a particular factor affects graduate wages.

### Admission Rate

````python
adm_rate_df = pd.DataFrame(cs_df, columns=['INSTNM','ADM_RATE_SUPP', 'MD_EARN_WNE_1YR'])
adm_rate_df = adm_rate_df.dropna(subset=['ADM_RATE_SUPP'])
adm_rate_df = adm_rate_df[adm_rate_df['ADM_RATE_SUPP'] != 'PrivacySuppressed']

# format admission rate as percentage (rounded down)
def adm_rate_pct(string):
    string = str(string)
    string = float(string) * 100
    return int(string)

adm_rate_df['ADM_RATE_SUPP'] = adm_rate_df['ADM_RATE_SUPP'].apply(adm_rate_pct)
adm_rate_df.head()
````

````python
# linear regression
outcome_ar, predictors_ar = patsy.dmatrices('MD_EARN_WNE_1YR ~ ADM_RATE_SUPP', adm_rate_df)
mod_ar = sm.OLS(outcome_ar, predictors_ar)
res_ar = mod_ar.fit(cov_type='HC1')
print(res_ar.summary())
````

Based on our OLS regression, whenever admission rates go up by 1%, the median wage for the college is predicted to go down by $84.88. The coefficient has a t-statistic = -4.236, which is statistically significant at the 5% confidence level since |-4.236| > 1.96, meaning that it does play a role in determining wage.

````python
# scatterplot
graph=sns.scatterplot(x='ADM_RATE_SUPP', y='MD_EARN_WNE_1YR', data=adm_rate_df, s=100)

xs = np.arange(adm_rate_df['ADM_RATE_SUPP'].min(), adm_rate_df['ADM_RATE_SUPP'].max())
ys = -84.8820  * xs + 4.682e+04
plt.plot(xs, ys, '--k', linewidth=4, label='Admission Rate Fit Line')

plt.xlabel('Admission Rate in Percentage')
plt.ylabel('Median Earnings after 1 Year')
graph.xaxis.set_major_locator(ticker.MultipleLocator(10))
plt.title("Admission Rate Slight Negative Correlation with Earnings Immediately After College", fontsize=30)
````

Based on the holistic scatterplot containing the entire spectrum of admission rates, admission rate percentage in US universities do have a slight negative correlation with median earnings. ### SAT/ACT Score

````python
test_df = pd.read_csv('data_college_scorecard.csv')
test_df.columns
````

````python
# Manually convert ACT score to SAT
def act_converter(score):
    if score == 36: 
        return 1600
    elif score == 35:
        return 1560
    elif score == 34:
        return 1510
    elif score == 33:
        return 1480
    elif score == 32:
        return 1440
    elif score == 31:
        return 1410
    elif score == 30:
        return 1380
    elif score == 29:
        return 1350
    elif score == 28:
        return 1320
    elif score == 27:
        return 1290
    elif score == 26:
        return 1250 
    elif score == 25:
        return 1220
    elif score == 24:
        return 1190
    elif score == 23:
        return 1150
    elif score == 22:
        return 1120
    elif score == 21:
        return 1090
    elif score == 20:
        return 1050
    elif score == 19:
        return 1020
    elif score == 18:
        return 980
    elif score == 17:
        return 950
    elif score == 16:
        return 910
    elif score == 15:
        return 870
    elif score == 14:
        return 820
````

Here, we want to see if the ACT/SAT scores have an effect on median earnings post graduation. To do this, we created a new index called combined test scores, which uses ACT and SAT data, this was done in order to not bias to a single test. To make this, we used [this website](https://admissions.uiowa.edu/academics-undergraduate-programs-undergraduate/actsat-equivalency-table) and used the top of each range, then hardcoded a converter function.

````python
# linear regression for test scores
test_df['AVG_ACT'] = (test_df['ACTCM25'] + test_df['ACTCM75'] )// 2
test_df['AVG_SAT'] = (test_df['SATVRMID'] + test_df['SATMTMID'])
dropped_rows = len(test_df) - len(test_df.dropna(subset=['AVG_ACT', 'AVG_SAT']))
print("Number of dropped rows:", dropped_rows)
test_df = test_df.dropna(subset=['AVG_ACT', 'AVG_SAT'])
test_df['CONVERTED_ACT'] = test_df['AVG_ACT'].apply(act_converter)
test_df['COMBINED_TESTS'] = (test_df['CONVERTED_ACT'] + test_df['AVG_SAT']) // 2
test_df = test_df.dropna(subset=['MD_EARN_WNE_1YR'])

X = test_df['COMBINED_TESTS']
X = sm.add_constant(X)
y = test_df['MD_EARN_WNE_1YR']
model = sm.OLS(y, X).fit(cov_type='HC1')
print(model.summary())
````

Based on the linear regression, there is a positive correlation between the combined test scores and median earnings: institutions with higher incoming average SAT/ACT scores tend to produce graduates with higher median earnings. Our OLS regression analysis indicates that with every additional point in average SAT score, the predicted median wage increases by $34.92, and the t-statistic of 11.37 is statistically significant. 

````python
highest_sat = test_df['AVG_SAT'].max()
highest_act = test_df['AVG_ACT'].max()
highest_convert = test_df['CONVERTED_ACT'].max()
highest_combined = test_df['COMBINED_TESTS'].max()

print("Highest SAT score:", highest_sat)
print("Highest ACT score:", highest_act)
print("Highest converted ACT score:", highest_convert)
print("Highest combined score:", highest_combined)
````

````python
# scatterplot
graph = sns.scatterplot(x='COMBINED_TESTS', y='MD_EARN_WNE_1YR', data=test_df, s=100)

xs = np.arange(test_df['COMBINED_TESTS'].min(), test_df['COMBINED_TESTS'].max())

sns.regplot(x='COMBINED_TESTS', y='MD_EARN_WNE_1YR', data=test_df, scatter=False, color='red')

plt.xlabel('Combined Tests')
plt.ylabel('Median Earnings after 1 Year')
plt.title("Strong Positive Correlation between Average Standardized Test and a College's Median Earning After 1 Year", fontsize=30)
````

The overall trend could suggest that schools with more academically accomplished incoming students—reflected in higher standardized test scores—also see these students earning more after graduation. This trend could be explained by various factors: students who score higher on these exams may tend to care more about their career, they may score higher due to access to resources such as test prep and private schools and thus come from more affluent backgrounds, or they may have higher aptitude in certain ways which transfers into the real world.

The outlier in the graph shows a particularly high median earning relative to the combined test scores. This could represent a university with specialized programs that lead to high-paying industries. For example, a school with a strong alumni network, location advantages, or exceptional placement services in lucrative fields might skew the earnings data upwards. Alternatively, this could also be a data collection anomaly or reporting error. If the university in question has a small sample size of graduates, or if a handful of graduates land extraordinarily high-paying jobs, this would disproportionately influence the median earnings reported.

Despite having the highest R-squared value out of all the variables and being statistically significant, this factor still only accounts for less than 20% of the variance in wage after graduation. Some institutions with lower combined scores have graduates with relatively high earnings, and conversely, some with high scores don't necessarily have the highest-earning graduates. This variance indicates that other factors beyond test scores might influence graduate earnings, and the positive relationship seen isn’t perfectly predictive.

## Multiple Regression of University Factors

To get an idea of trends we might see later in our independent studies of each variable, we conducted a multiple regression of SAT/ACT scores, % STEM Majors, Admission Rate, and Ranking against the dependent variable of Median Earnings After 1 Year. We omitted location from the multiple regression since location is not a quantitative trait. Further in this study we will study location against median earnings and observe its independent effect on university graduate wages.

````python
combined_df = pd.merge(test_df, stempct_df, on='INSTNM')
combined_df = pd.merge(combined_df, rankingDF, on='INSTNM')

combined_df.columns
````

````python
# Standardizes our columns
scaler = StandardScaler()

# For some reason it wasn't working until columns are typecasted to float
combined_df['COMBINED_TESTS'] = combined_df['COMBINED_TESTS'].astype(float)
combined_df['STEM_PCT'] = combined_df['STEM_PCT'].astype(float)
combined_df['ADM_RATE_SUPP'] = combined_df['ADM_RATE_SUPP'].astype(float)
combined_df['RANKING_NO_INCOME'] = combined_df['RANKING_NO_INCOME'].astype(float)

y = combined_df['MD_EARN_WNE_1YR']

X = combined_df[['STEM_PCT', 'COMBINED_TESTS', 'ADM_RATE_SUPP', 'RANKING_NO_INCOME']]

X_scaled = scaler.fit_transform(X)
X_scaled = sm.add_constant(X_scaled)
model = sm.OLS(y, X_scaled).fit(cov_type='HC1')

print(model.summary())
````

The resulting multiple regression model overall has a relatively high R-squared value of 0.652, meaning that the model accounts for 65% of the variance in median earnings. In making this model, each variable is standardized. For example, the STEM_PCT coefficient 3782.94 means that if a school is one standard deviation above the mean in terms of % of students in STEM, that school's average salary would be \\$3,782.94 greater than if their STEM % was the mean, assuming all other variables are constant. Standardizing the variables in this way lets us better compare the coefficients for each variable to each other.

In our results, COMBINED_TESTS has the highest coefficient, followed by STEM %, then ranking, and finally that school's admission rate. The school admission rate appears negative due to an increase in admission rate representing the university being less competitive (easier to be accepted into) compared to others.

In terms of statistical significance, the STEM percentage and combined test variables seem to be very statistically significant, both with t-scores around 5. The t-score for school ranking and admission rate however, are much lower, meaning their results may not be statistically significant. The t-score for the admission rate is -1.3, a relatively low value that falls below the threshold for statistical significance at the 5% level.  The t-score for ranking is 2.149 which means that it is statistically significant at the 5% level but not the 1% level.

So, the differences in earnings we found from the admission rate are small enough that they may have occurred from random variance in the data. The differences in earnings corresponding to our other variables are much larger and less likely to have occurred solely from random variance, meaning admission rate is less statistically significant than our other variables.
