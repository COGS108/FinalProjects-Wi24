# Data

Here is the dataset that we are using for this project:
- [Car Features and MSRP](https://www.kaggle.com/datasets/CooperUnion/cardataset)

This dataset includes information about cars (ex. make, year, transmission type, etc) and its respective manufacturer's suggested retail price (MSRP). It is very convenient to access since it can be downloaded from Kaggle as a csv file. Some important variables include the make, year, city and highway mpg, vehicle style, and engine horsepower. The datatypes of the columns is a mix of numerical and categorical variables. We figure that we only need this one dataset since it has about 12,000 rows which is way more than enough, and it includes 16 variables which provides a variety of options for running a linear regression:

- Make: The company which produced the car (e.g. BMW, Toyota, Acura)
- Model: The specific name of the car (e.g. Civic,  Jetta)
- Year: The year that the car was first produced
- Engine Fuel Type: The type of fuel that the car requires, most commonly regular unleaded or premium unleaded
- Engine HP: The power the car's engine produces, measured in units of hp (horsepower)
- Engine Cylinders: How many cylinders are in the car's engine (cylinders are where power is generated and fuel is combusted)
- Transmission Type: The type of transmission of the car which is responsible for changing gears, most commonly automatic or manual
- Driven_Wheels: The car's system that transfers power from the engine to the wheels, can be one of three categories: front wheel drive, rear wheel drive, or four wheel drive
- Number of Doors: How many doors the car has, can be one of three categories: 2, 3, or 4.
- Market Category: Classification of the vehicle in terms of its configuration or price range (e.g. Luxury, Hatchback, Crossover)
- Vehicle Size: Size of the vehicle, can be one of three categories: Compact, Midsize, or Large
- Vehicle Style: Classification of the vehicle in terms of its target demographic or intended use (e.g. Sedan, SUV, Minivan)
- highway MPG: The car's average miles per gallon while driving on an open road at high speed without stopping
- city mpg: The car's average miles per gallon while driving in urban conditions
- Popularity: Used to rank popularity of the car
- MSRP: Stands for Manufacturer's Suggested Retail Price, which is the price recommended by the car's manufacturer to retailers

To preprocess our data, we would need to ensure that all string columns are standardized and written in the same format, and for our numerical columns we need to remove outliers that can be signifiers of missing values. In addition, the string values in the features should be one-hot encoded in order to allow the machine learning model to successfully predict MSRP with numeric values.

## Car Features and MSRP Dataset

```python
#Loading the data
import pandas as pd 
pre_data = pd.read_csv('data/car_features_and_msrp.csv')
pre_data.head()
```

```python
#Missingness per column
pre_data.isna().sum()
```

Before beginning the data cleaning process, we can see here that the data was almost entirely tidy, with the exception of some missing values in five of the columns. Of the 3850 missing values, 3742 of them are in the `Market Category` column, so we decided to drop that column from the dataset since the high amount of missingness would make it hard to work with, along with the fact that we do not expect that column to make a major impact on the predictions. The other columns have a negligible amount of missing values, so we decided to drop the rows where these values are missing. Making these two changes reduces the size of the dataset from 11,914 observations and 16 features to 11,812 observations and 15 features, which is still more than sufficient.

```python
#Dropping Market Category column and rows with missing values for the other columns
pre_data=pre_data.drop(columns='Market Category')
pre_data = pre_data[~pre_data['Engine Fuel Type'].isna()]
pre_data = pre_data[~pre_data['Engine HP'].isna()]
pre_data = pre_data[~pre_data['Engine Cylinders'].isna()]
pre_data = pre_data[~pre_data['Number of Doors'].isna()]
pre_data.isna().sum()
```

Next, we need to check for duplicate rows and drop them if they are present, since these will skew the result of our linear regression if they are not removed. We found 720 duplicated rows, so dropping them reduces the size of the dataset to 11,092.

```python
#Checking for duplicate rows
pre_data[pre_data.duplicated()]
```

```python
#Dropping the duplicate rows
pre_data = pre_data.drop_duplicates()
```

We also decided that it would be helpful to check the descriptive statistics of the dataset to see if there are any outliers in the numerical columns. Looking at the minimum, maximum, percentiles, mean, and standard deviation shows us that none of these values seem unreasonable. Because of this, we do not need to drop any outliers since none appear to be present.

```python
#Descriptive statistics
pre_data.describe()
```

Now, since most of the columns in this dataset are categorical, we are going to one hot encode them to make them more useful. Before doing this, we will check how many unique values each of the categorical variables have. We can see that most of the categorical columns have relatively few categories, aside from one, which is the `Model` column with 904 unique values. One hot encoding the `Model` column would cause the number of features in the dataset to increase by a huge margin, so we will drop that column since having too many features in the dataset can result in overfitting the data.

```python
#Checking number of unique categories for each categorical variable, dropping Model column
categorical_cols=['Make', 'Model', 'Year', 'Engine Fuel Type',
       'Engine Cylinders', 'Transmission Type', 'Driven_Wheels',
       'Number of Doors', 'Vehicle Size', 'Vehicle Style']
for c in categorical_cols:
    print(f'The {c} column has {pre_data[c].nunique()} unique values.')
categorical_cols.remove('Model')
pre_data=pre_data.drop(columns='Model')
```

```python
#One hot encoding the categorical columns
data=pd.get_dummies(pre_data, columns=categorical_cols)
data=data.reset_index()
data.head()
```

After one hot encoding, the number of features has increased from 14 to 129. 

We will likely come back to the data cleaning portion of this project and make changes once we start our EDA and modeling, since we may notice more issues with the dataset or ways to transform the columns to make them more useful. For example, we could realize that one of these one hot encoded categorical variables will not be useful for our model, leading us to drop it.

## Linear Regression

Some more words and stuff.  Remember notebooks work best if you interleave the code that generates a result with properly annotate figures and text that puts these results into context.

```python
from sklearn.linear_model import LinearRegression
from sklearn.model_selection import train_test_split
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score, mean_absolute_percentage_error

# Initialize design matrix and feature vector
X = data.drop(columns=['MSRP', 'Log_MSRP'])
y = data['MSRP']

# train test split
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=44)

# Fit the model
ols = LinearRegression().fit(X_train, y_train)

# Initialize y_hat for both train and test
train_y_hat = ols.predict(X_train)
test_y_hat = ols.predict(X_test)

# Display scores
print('Training scores:')
print('R^2: ', r2_score(y_train, train_y_hat))
print('MAE: ', mean_absolute_error(y_train, train_y_hat))
print('RMSE: ', mean_squared_error(y_train, train_y_hat, squared=False))
print('Test scores: ')
print('R^2: ', r2_score(y_test, test_y_hat))
print('MAE: ', mean_absolute_error(y_test, test_y_hat))
print('RMSE: ', mean_squared_error(y_test, test_y_hat, squared=False))
```

```python
# Set the style of seaborn for better aesthetics
sns.set_style("whitegrid")

plt.figure(figsize=(10, 6))

# Scatter plot for actual vs predicted values
sns.scatterplot(x=y_test, y=test_y_hat, alpha=0.2, color='orange', label='Predicted vs Actual')

# Ideal line for perfect predictions
plt.plot([y_test.min(), y_test.max()], [y_test.min(), y_test.max()], 'k--', lw=2, label='Ideal Prediction')

plt.title('Test Set: Actual vs. Predicted MSRP using Linear Regression')
plt.xlabel('Actual MSRP')
plt.ylabel('Predicted MSRP')
plt.legend()
```

## XGBoost

# Discussion and Conclusion

Building upon the insights gained from previous projects in car price prediction, our research aims to further explore the relationship between various car features and their influence on sale prices. Drawing from the methodologies and findings of projects like Car Price Prediction Multiple Linear Regression and Car Price Prediction using Machine Learning Techniques, we plan to employ multiple linear regression as our primary model while also considering the potential benefits of incorporating other machine learning techniques.

We began our data analysis by examining the distribution of various attributes in our dataset. We found that Chevorlet, Ford, and Toyota were the top three most common car models in our dataset, with more luxurious brands such as Lamborghini and Bugatti being less common. We chose to lograithmically tranform the MSRP column in order to normalize the distribution, and it displayed a right-skewed curve. This tells us that most of the values are clustered to the left, with the tail extending towards higher, more expensive values. We expected this for the distribution of car prices because most common cars tend to sell within similar price ranges, with a luxury brands selling their cars for much higher prices. Door count seemed to have no significant impact on MSRP, as seen by our violin plot that shows MSRP distribution for cars with different doors. Most cars were priced lower regardless of doorcount. Our exploratory data analysis also demonstrated that automated manual and direct drive transmissions exhibit greater variability and wider distributions, indicating a broader range of MSRPs within these categories. Interestingly, the 'unknown' transmission category shows a narrow distribution, suggesting less variability in MSRP for vehicles with unidentified transmission types. 

Overall, we were able to build relatively accurate models to predict car prices. Our aim with predictive analysis was to leverage these features to gain insights into their influence on car prices. Across all of our models that we build we had an average training R2 score of 0.96 and average test R2 score of 0.89. 

Our analysis has yielded compelling insights into the factors influencing car prices, supported by strong R-squared values and other performance metrics. The high R-squared values achieved by our predictive models indicate that a significant portion of the variance in car prices can be explained by the selected features, including transmission type, engine size, and fuel efficiency. This suggests that our models effectively capture the underlying relationships between these features and MSRP, enabling accurate predictions. Furthermore, the robust performance of our models on both training and test datasets showcases their reliability and generalization capabilities. Despite the complexity of the automotive market and the multitude of factors affecting car prices, our models demonstrate consistent predictive accuracy across different datasets. This suggests that the relationships identified by our models are robust and applicable to a wide range of scenarios.

We hypothesized that our linear regression model will yield the most accurate results, but it was actually CatBoost that had the most lowest testing RMSE. Our XGBoost model performed second best, and our linear regression model actually performed the worst.
