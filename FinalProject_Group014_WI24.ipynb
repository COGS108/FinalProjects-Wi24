## Background and Prior Work

Walkability refers to the degree of ease with which places and services in an area can be accessed on foot. It is comprised of factors related to pedestrian-friendly infrastructure, the proximity of destinations, and the presence of walkway systems. The concept of walkability often promotes the reduction of private vehicles and encourages the development of neighborhoods with greater access to amenities by walking as the sole mode of transportation. In addition, the walkability of an area could imply a more active lifestyle by incorporating physical activity into daily livelihoods. Thus, we are initially interested in the health benefits of walking and distinguishing locations in the United States that promote those benefits.

Walking can improve or maintain overall health by incorporating exercise into daily routines. This can increase cardiovascular strength, improve muscle growth, and reduce excess body fat,<a name="text4"></a>[<sup>4</sup>](#ref4) among other benefits - all of which are attributed to lower risk conditions such as high cholesterol, diabetes, obesity, heart disease, or stroke.<a name="text4"></a>[<sup>4</sup>](#ref4) On top of this, there is often a correlation between mental and physical health, which we investigate in this study as well.

## Data Cleaning

Now, we can import our data and check for any U.S. counties that might be unaccounted for in the dataset. ### Dataset #1: PLACES

Our first dataset displays health conditions and behaviors of indivviduals across the United States. These health conditions are quantified as a percentage of crude or age-adjusted prevalence of that health condition attributed to each County FIPS code. This is important to note so we can verify that we are accurately rename our columns to match the Data Dictionary names, and to transform the population data so that it represents the age-adjusted percentage of survey respondents.

```python
# Import Data
places = pd.read_csv('data/PLACES__County_Data__GIS_Friendly_Format___2020_release_20240220.csv')

# Basic Description
print(places.shape)
# There are 3142 entries, which matches the number of CountyFIPS codes in the United States (not including territories).
# There are a total of 3242 CountyFIPS codes in the USA, which includes 100 FIPS codes for U.S. territories.

places.head()
```

```python
# Check for Missingness
places.isna().any().any() # False, no missingness.
```

No missing counties are detected, so we can proceed with cleaning this dataset.

```python
# Data Cleaning

# Breaks down the combined FIPS into state and county FIPS codes, so that we can group the data by state if needed.
def breakdownFips(combined):
    return int(combined/1000), int(combined%1000)

places = places.assign(STATEFP = places.apply(lambda row: breakdownFips(row.CountyFIPS)[0],axis=1))
places = places.assign(COUNTYFP = places.apply(lambda row: breakdownFips(row.CountyFIPS)[1],axis=1))

# Log Transform population
places = places.assign(popuLog = np.log10(places.TotalPopulation))

# Clean up columns
# Filter the columns to only include the annual prevalence data for each state.
# Annual prevalence data is in units of: % of survey respondents, age-adjusted by standard U.S. population 
# distribution in 2000
placesCols = list(places.columns)
placesCols = [col for col in placesCols if "95CI" not in col and "Crude" not in col]
places = places[placesCols]

# Rename cols to remove "AdjPrev", which makes it match the Data Dictionary (included in the data_docs folder.)
placesCols = list(places.columns)
placesCols = [col if "_AdjPrev" not in col else col[:-8] for col in placesCols]
places.columns = placesCols

# Now, the columns match the Data Dictionary, making it easier to parse!

places.head()
```

This first dataset is prepped for merging, so we move on to the next dataset on the Walkability Index!

### Dataset #2: Walkability

Our second dataset scores the walkability of a city based on several environmental and infrastructural factors, quantified by the Walkability Index per County FIPS code. We will want to compare the ability to access resources via walking against the health conditions of the first dataset.

Let's perform the same steps as before to the Walkability dataset to import and check the data for any missing counties, then clean up the data for consistency.

```python
# Import Data
walk = pd.read_csv('data/EPA_SmartLocationDatabase_V3_Jan_2021_Final.csv')
# NOTE: You must extract the dataset manually: the raw file is too large to include in the Git repository, 
# so you need to extract data/Walkability.zip.

# Basic Description
print(walk.shape) 
# There are 220740 entries: Each entry is classified under STATEFP, COUNTYFP, TRACTCE, and BLKGRPCE in that order of 
# specificity. These are identifiers referring to different Census Tract areas of each state. Because our PLACES 
# data is only available for the CountyFIPS level of specificity, we need to group data by CountyFIPS.

walk.head()
```

```python
# Check for missingness
walk.isna().any().any() # True, there is missingness. Let's check where...
```

```python
# Check the percentage of missing values by column
walk.isna().mean()[walk.isna().any() == True] 
# The missingness occurs in columns that do not affect our analysis. 
# For more information, refer to the table below for details on missingness.
```

Unlike the Places dataset, we notice missing values occur in the following columns of the Walkability data:

| Column | Description |
|--------|-------------|
| GEOID10 | 2010 FIPS code (We use 2018 instead, which is more updated) |
| CSA, CSA_Name, CBSA, CBSA_Name | Statistical Area codes that we do not use. We identify areas via CountyFIPS instead. |
| CountHU, HH | Count of total vs. occupied Housing Units. We use population data instead, which is available for all areas |
| D1a | Density of housing units. We use Walkability Index instead |
| D1c8_Off | Office employment density |
| D4d, D4e | Frequency of transit service, aggregated per square mile vs. per capita |

Luckily, these inconsistencies do not affect our analysis! We continue to clean the data by averaging the walkability index for each county FIPS code to later be used in our exploratory analysis.

```python
# Data Cleaning

# Combine the STATEFP and COUNTYFP to form a proper, unique CountyFIPS code that identifies each county
def combineFips(state,county):
    s = str(state)
    c = str(county)

    if county < 10:
        return int(s + '00' + c)
    elif county < 100:
        return int(s + '0' + c)
    else:
        return int(s + '' + c)
    
walk = walk.assign(CountyFIPS = walk.apply(lambda row: combineFips(row.STATEFP, row.COUNTYFP), axis=1))
# This results in 3233 counties, which is more than the 3143 counties in the United States.
# This is because there's extra data for U.S. territories. The PLACES data only covers data from the 50 United States,
# so they will be filtered out when merging (left joining) with the PLACES data.

# Get the mean Walkability Index by County
# This is because: The Walkability data is further classified under CountyFIPS using TractCE and BlkGrpCE 
# (more specific Census Tract areas), but the PLACES data does not go into that level of detail.
walk_mean = pd.DataFrame(walk.groupby('CountyFIPS').NatWalkInd.mean())

# Log transform Walkability
walk_mean = walk_mean.assign(walkLog = np.log10(walk_mean.NatWalkInd))

# No further cleaning is required, because only the Walkability Index will be left joined into the PLACES data.

walk_mean.head()
```

## Combining Datasets

Now that the data has been cleaned, we take advantage of the common usage of FIPS county codes in both datasets to merge the Walkability and Places survey data. This will allow statistics to be compared and displayed more efficiently.

```python
# Compare the number of counties before merging
print(len(places.CountyFIPS.unique()), len(walk.CountyFIPS.unique()), len(walk_mean))
# 3142 counties in PLACES data, and 3233 in Walkability data.
# They will be left joined on the U.S. States data, so this makes sense.

# Left join the walkability index into the Places dataset.
merged = pd.merge(places, walk_mean, on='CountyFIPS', how='left')

merged.head()
```

## Linear Regression

Now, we can conduct a linear regression to investigate relationships between individual health conditions and the Walkability Index of each county.

The Ordinary Least Squares (OLS) regression is a technique that forms a linear model between our predictor variable (a selected health condition from the Places dataset) with the walkability index (from the Walkability dataset) such that the least variance from the linear model is obtained across all compared values.

Here's an example of the OLS Regression Results for the `PHLTH` variable (Physical Health)

```python
# Linear Regression
outcome1, predictors1 = patsy.dmatrices('PHLTH ~ NatWalkInd', merged)
model1 = sm.OLS(outcome1, predictors1)
results1 = model1.fit()
print(results1.summary())
results1.bse
```

This process is then repeated across all analyzed variables.

For our research, we are interested in the correlations and predictability of the relationship, as well as any outliers or extrema that can be observed.

Thus, we create a regression table to find the intercepts and coefficients for the relationship between the walkability index and all the observed health outcomes. 

```python
intercepts = []
coefficients = []
bse_intercept = []
p_coefficients = []

for i in outcomeCols:
    outcome, predictors = patsy.dmatrices(f'{i} ~ NatWalkInd', merged)
    model = sm.OLS(outcome, predictors)
    result = model.fit()
    intercepts.append(result.params[0])
    coefficients.append(result.params[1])
    bse_intercept.append(result.bse[0])
    p_coefficients.append(result.pvalues[1])
    
regression_df = pd.DataFrame({'Health Conditions':outcomeColsNames, 'Coefficients':coefficients, 'Intercepts':intercepts, 
                              'Standard Error':bse_intercept, 'p_value':p_coefficients}).set_index('Health Conditions')
regression_df
```

Across the models, the coefficients are all negative, indicating a negative relationship between the walkability index and physical health, diabetes, obesity, and mental health states of surveyors across the United States. More explicitly, as the walkability index of a city increases, these health issues among residents in that city tend to decrease. 

From all the linear regression models conducted above, we notice that all p-values in the tables are around 0. Since p-values demonstrate the likelihood that a relationship is random, this suggests strong evidence against the null hypothesis, indicating a highly predictable linear correlation between the walkability index and various health outcomes. 

Intercepts from the regression results are also listed, to show the highest percent prevalence of that health condition for that singular condition's regression line. The standard error states the averaged variance that data values possess in relation to the regression line.


Results from the linear regression are our first indication that the negative relationships between health issues and the walkability of a location exist, which we also discovered in the correlation matrix.

## Outliers & Extrema

Now that we have established common trends between health conditions and the walkability index of counties in the United States, we must identify the outliers within these relationships to verify the validity of our collected data.

Let's observe values that fall in the top and bottom quarter quartile for each dataset, and find the locations of those responses:

```python
# Helper function to identify the outliers:
def findOutliers(col):
    outcome = merged[col]
    outcome_Q1 = np.percentile(outcome, 25)
    outcome_Q3 = np.percentile(outcome, 75)
    outcome_IQR = outcome_Q3 - outcome_Q1
    outcome_lower = outcome_Q1 - 1.5 * outcome_IQR
    outcome_upper = outcome_Q3 + 1.5 * outcome_IQR
    outcome_outliers = outcome[(outcome < outcome_lower) | (outcome > outcome_upper)]

    outcome_indices = outcome_outliers.index

    # The subset of the dataframe with all outliers identified
    return merged.loc[outcome_indices] 
```

```python
# Outliers in physical health:
outliers_PHLTH = findOutliers('PHLTH')

# Find the geographical area that outliers are concentrated in:
ret1, counts1 = np.unique(outliers_PHLTH.get('StateDesc'), return_counts=True)
ret1, counts1/outliers_PHLTH.shape[0] # Alaska(0.12), Kentury(0.35), Mississippi(0.12), South Dakota(0.12)
```

```python
# Outliers in diabetes:
outliers_diabetes = findOutliers('DIABETES')

ret2, counts2 = np.unique(outliers_diabetes.get('StateDesc'), return_counts=True)
ret2, counts2/outliers_diabetes.shape[0] # Alabama(0.14),Georgia(0.09), Mississippi(0.25), and Texas(0.25)
```

```python
# Outliers in CHD:
outliers_CHD = findOutliers('CHD')

ret3, counts3 = np.unique(outliers_CHD.get('StateDesc'), return_counts=True)
ret3, counts3/outliers_CHD.shape[0] # kentucky(0.49), South Dakota(0.2)
```

```python
# Outliers in obesity
outliers_obesity = findOutliers('OBESITY')


ret4, counts4 = np.unique(outliers_obesity.get('StateDesc'), return_counts=True)
ret4, counts4/outliers_obesity.shape[0] # California(0.13), Colorado(0.35)
```

```python
# Outliers in mental health:
outliers_MHLTH = findOutliers('MHLTH')

ret5, counts5 = np.unique(outliers_MHLTH.get('StateDesc'), return_counts=True)
ret5, counts5/outliers_MHLTH.shape[0] # Kentucky(0.46), West Virginia(0.31)
```
