We want our *Mon-Yr* column to look similar to our Date column and hence, we wrote two `standardize_year()` that convert the *Mon-Yr* column into a format that can be easily converted into the `pd.datetime` format

```python
def standardize_year(string):
    if(string == "Jan-00" ):
        string = string.replace(string, "2000-Jan")
    if(string == "Feb-00" ):
        string = string.replace(string, "2000-Feb")
    if(string == "Mar-00" ):
        string = string.replace(string, "2000-Mar")
    if(string == "Apr-00" ):
        string = string.replace(string, "2000-Apr")
    if(string == "May-00" ):
        string = string.replace(string, "2000-May")
    if(string == "Jun-00" ):
        string = string.replace(string, "2000-Jun")
    if(string == "Jul-00" ):
        string =  string.replace(string, "2000-Jul")
    if(string == "Aug-00" ):
        string = string.replace(string, "2000-Aug")
    if(string == "Sep-00" ):
        string = string.replace(string, "2000-Sep")
    if(string == "Oct-00" ):
        string = string.replace(string, "2000-Oct")
    if(string == "Nov-00" ):
        string = string.replace(string, "2000-Nov")
    if(string == "Dec-00" ):
        string = string.replace(string, "2000-Dec")
    return string
```

```python
def standardize_yr(string):
    retval = string
    num2 = string[0:2]
    num = string[0]
    
    if(len(string)==5):
        if(num.isnumeric()):
            num = int(num)
            if(num <= 9 and num >0):
                num = num + 2000
                string = str(num)
                con = retval[1:5]
                string = string + con
    elif(len(string)==6):
        if(num2.isnumeric()):
            num2 = int(num2)
            if(num2 <= 24 and num2 >= 10):
                num2 = num2 + 2000
                string = str(num2)
                con = retval[2:6]
                string = string + con
    else : 
        string = retval
    return string
```

```python
hprice["Mon-Yr"] = hprice["Mon-Yr"].apply(standardize_year)
hprice["Mon-Yr"] = hprice["Mon-Yr"].apply(standardize_yr)
hprice['Mon-Yr']= pd.to_datetime(hprice["Mon-Yr"])
```
