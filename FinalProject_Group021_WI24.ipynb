## Section 3 of EDA - Linear Regression of Mechanical Properties


### we consider our result to be significant if p-value is less than 0.05

```python
# Create a smaller dataset for multiple linear regression
V_clean = clean.drop(['Material','Heat_Treatment','Vickers_Hardness_Number','Description_of_the_Material','material_name'],axis=1)
V_clean.rename(columns={'Poisson\'s_Ratio': 'Poissons_Ratio'},inplace=True)
V = V_clean.columns
```

```python
# Create a list of column names to loop through
L = []
for i in V:
    print(i)
    L.append(i)
print(len(L))
```

```python
# Regression with one Mechanical Property as the dependent and all others as the independent
M=[]
for i in range(len(L)):
    k=i
    D=[]
    I=[]
    for i in range(0,len(L)):
        if i==k:
            D.append(L[i])
        else:
            I.append(L[i])
    S=''
    S=S+D[0]+' ~ '
    for i in range(len(I)-1):
        S=S+" "+I[i]+" +"
    S=S+" "+I[len(I)-1]
    outcome, predictors = patsy.dmatrices(S, V_clean)
    mod = sm.OLS(outcome, predictors)
    res = mod.fit()
    #M.append(S)
    M.append(res)
```

We provided all the linear regressions for one property as the dependent variable and all other properties  as the independent variables. We stored them in M. But we found that most of the results are not significant. Thus we focus on the regression between the correlated varibales we found in the correlation matrix above.

```python
clean.columns
```

###   Ultimate_Tensile_Strength's relationship with Yield_Strength and Brinell_Hardness_Number

```python
outcome, predictors = patsy.dmatrices('Ultimate_Tensile_Strength ~ Yield_Strength + Brinell_Hardness_Number', clean)
mod = sm.OLS(outcome, predictors)
res = mod.fit()
print(res.summary())
```

From the summary, we verify our idea that Ultimate_Tensile_Strength has a linear relationship with Yield_Strength and Brinell_Hardness_Number each with a positive coefficient. We also found that the intercept is not significant.

$$ Ultimate_Tensile_Strength = 0.5792  * Yield_Strength + 0.5254 * Brinell_Hardness_Number $$

```python
from mpl_toolkits.mplot3d import Axes3D

fig = plt.figure()
ax = Axes3D(fig)
fig.add_axes(ax)
ax.scatter3D(clean['Yield_Strength'],clean['Brinell_Hardness_Number'],clean['Ultimate_Tensile_Strength'])
#range has considered the maxs and mins for both Yield_Strength and Brinell_Hardness_Number
xs = np.arange(-1.5,4.5) 
ys = np.arange(-1.5,4.5)
xs, ys = np.meshgrid(xs, ys)
zs = 0.5792  * xs + 0.5254 * ys
ax.set_zlabel('Ultimate_Tensile_Strength')
ax.set_xlabel('Yield_Strength')
ax.set_ylabel('Brinell_Hardness_Number')
ax.zaxis.labelpad=-2
ax.plot_surface(xs,ys,zs,alpha=0.6)
ax.set_title('Ultimate_Tensile_Strength\'s relationship with Yield_Strength and Brinell_Hardness_Number')

plt.show()
```

###   Brinell_Hardness_Number's relationship with Yield_Strength

```python
outcome, predictors = patsy.dmatrices('Yield_Strength ~ Brinell_Hardness_Number', clean)
mod = sm.OLS(outcome, predictors)
res = mod.fit()
print(res.summary())
```

From the summary, we verify our idea that Yield_Strength has a linear relationship with Brinell_Hardness_Number with a positive coefficient.

# Ethics & Privacy

Our dataset, sourced from the Autodesk Material Library, contains detailed information on the mechanical properties of materials, including their Brinell hardness numbers. However, we've encountered a challenge that's common in real-world data: missing values. Missing data can significantly impact our analysis, potentially skewing the results and leading to incomplete conclusions. For example, if our dataset lacks complete hardness numbers for certain materials or details about their heat treatment processes, our understanding of the relationship between heat treatment and material hardness could be compromised. This gap might lead to biased insights, especially if the missing values are not randomly distributed but are instead related to specific materials or treatment types.
To address these issues in a professional and methodical manner, we might employ strategies like data imputation to fill in missing values where possible, or sensitivity analysis to understand how our conclusions might change with different assumptions about the missing data. By doing so, we aim to ensure that our analysis is robust and reliable, providing clear insights into how heat treatment processes affect material hardness and uncovering any other trends that emerge from the data.

We did not want to discard any properties, so we decided that the best way to do our analysis was by using correlation matrices and heatmaps. These correlation matrices allowed us to see the strength and direction of all correlations at a glance. The heatmaps made it easy to interpret them. We were then able to look more deeply into any correlations which stood out by using single variate and multivariate regressions on properties of interest. 

Our group started by looking at property correlations between all materials in order to get a sense of what to expect. As expected, we found that there was a very strong and positive correlation between the tensile and yield strength. Here we define "strong" as a correlation with an coefficient over ~|0.8|. There were also very strong and positive correlations between the Brinell hardness number and the tensile strength and yield strength. Further, there was a really strong positive correlation between Vicker's hardness number and the pressure at yield. Curiously, there were no strong negative correlations at all, with the strongest negative correlation having a coefficient of -0.5. 

We then narrow our analysis down to look at material property correlations within specific materials. Our group noted that there was a very wide range of materials in the dataset, with some not having very many observations. So, we decided to only look at materials whih had over 100 observations in the dataset. These materials ended up being steel, aluminum, CSN, and DIN. Within steel (not specifying carbon content), the notable differences were a strong positive correlation between the density and its shear modulus and no available correlation between its pressure at yield and its Vicker's hardness number due to lack of data. Within aluminum, there were no significant differences from the correlations between all materials. However, there was a lot of missing data as well as data which had constant values which made the analysis difficult to use. There was little difference to note within CSN except that the negative correlation between the elongation at the material's break or strain and its yield strength became 1.5x stronger, although still not strong. DIN had some interesting patterns, where density had strong positive correlations with the elastic modulus, shear modulus, and its Poisson's ratio. Similar to CSN, the negative correlation between its elongation and yield strength became 1.5x stronger as well.

The next step in our analysis was analyzing material property correlations within the low, medium, and high carbon subsections of steel. We failed to find any real significant differences in the correlations. Due to this, it seems that the different carbon content of steels has little to no effect on the relationships of its material properties.

Finally, our group conducted three regression analyses on properties of interest. These properties inluded the tensile strength, yield strength, Brinell hardness number, elastic modulus, shear modulus, the pressure at yield, and Vicker's hardness number. We conducted regressions for the yield strength and Brinell hardness number on tensile strength, Brinell hardness number on yield strength, shear modulus on elastic modulus, and Vicker's hardness number on the pressure at yield. Every regression that we ran held significant results at the .001 level for the independent variables. It should be noted that these relationships aren't necessarily causal, and deeper analysis would be necessary to determine a causal relationship.

These relationships were largely reflected within specific materials like steel, aluminum, CSN, and DIN. These specific materials had some significant differences from the materials as a whole. Interestingly, there were no significant negative correlations to note. We found that the strong relationships we found through our correlation matrices were significant by running multiple regression analyses.
