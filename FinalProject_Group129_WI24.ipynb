<h1>Abstract</h1>

The topic of this project is mainly to investigate the correlation between rankings of US colleges and the median salaries of their bachelor graduates. This topic is significant for people who are in the process of making college decisions and their parents, as the results of our research topic can provide them guidance on how much weight to put on college rankings to obtain high income after graduation. We decide to use Times Higher Education ranks as the data for college rankings and PayScale datasets as the data for the median graduates’ salaries of each college. In the research process, we divide the median salary into starting and mid-career salary, where starting salary indicates 0-5 years of work experience and mid-career salary indicates 10+ years of work experience. We also choose two specific time periods to investigate, which are 2012-2018 and 2023-2024 respectively, because these two time periods are the most recent time periods that have not been affected by Covid, which causes great fluctuations in income and thus potential inaccuracies in our results. Moreover, the salaries of each year are adjusted at the standard inflation rate to ensure consistency. 

The distribution of the Times datasets showed that there were too few universities with ranks greater than 800 to be confident they reflect the larger population, so we only included universities that didn't exceed 800. The distribution of both the starting and mid-career salary show an increase over time after adjusting for inflation. Most importantly, we found a positive correlation between university rank and salary for both starting career and mid-career median salaries using scatterplots. We also specifically investigate the salaries of UCs and Ivy Leagues over the years to compare their respective salaries, as these two systems of universities are famous and usually attract more students due to their widespread popularity.

Statistical analysis is conducted via t-tests and used to give insight on how likely it is that the salaries of the top and bottom 30% ranked universities come from the same population (also split by year). Using p-values of 5% to determine significance, we found significant differences between the two groups for all years we analyzed in both starting and mid-career salaries, indicating that graduates from top 30% universities earn significantly more than those from bottom 30%. While there is a noticeable difference in salary between top 30% and bottom 30%, the difference in salary of universities within top 30% range is not great, meaning that, for example, top 10% universities do not differ greatly from top 30% universities in terms of median salaries.

<h3> Times Dataset </h3>

Times datasets are cleaned and concatonated. We also renamed the "name country region" to "name" and remove unnecessary elements like "United States" or "US" and "Explore" etc from the name column. Universities without rank will be removed as well. The first 200 ranks from the Times Rankings are given individually, after which Times judges certain ranking intervals of universities to be negligibly different in rankings. This results in ranks 201-225 having an identical rank, and we chose to preserve the highest rank as it is a tie. This causes a banding pattern when plotting rank against other variables. <br>
The tied groups are 201-225, 226-250, 251-275, 276-300, 301-350, 351-400, 401-500, 501-600, 601-800, 801-1000, 1001+.

```python
dfs = []

for df in (times13, times14, times15, times16, times17, times18, times24):
    # lower, strip, rename columns
    df.columns = df.columns.str.lower().str.strip()
    df = df.rename(columns={'name\ncountry/region': 'name'})

    # remove region and name artifacts from csv
    df.name = df.name.str.replace('United States','')
    df.name = df.name.str.replace(', US','')
    df.name = df.name.str.replace("\nExplore",'')
    df.name = df.name.str.replace(' (Main campus)','')
    df.name = df.name.str.replace('-Pittsburgh campus','')
    df.name = df.name.str.replace('University of South Florida (Tampa)','University of South Florida')
    
    # keep useful columns
    df = df.get(['name','rank','year'])
    
    dfs.append(df)

# combine dataframes
times = pd.concat(dfs, join='outer')

# clear repeated indices
times = times.reset_index().drop(labels='index', axis=1)

# clean rank data
times['rank'] = times['rank'].convert_dtypes(convert_string = True).str.strip()
if times['rank'].str.contains('-').sum() > 0:
        times['rank'] = times['rank'].str.replace(',','')
        times['rank'] = times['rank'].str.replace('+','')
        times['rank'] = times['rank'].str.replace('–','-')
        for i in [
            '-225', '-250', '-275', '-300', '-350',
            '-400', '-500', '-600', '-800',
            '-1000', '-1200','-1500']:
            times['rank'] = times['rank'].str.replace(i,'')
        times['rank'] = times['rank'].str.replace('Reporter','0')
        
        times['rank'] = times['rank'].apply(int)

times = times[times['rank'] > 0] # remove Reporter ranked universities (don't have a rank)
times['year'] = times['year'].apply(int)

times.head()
```

<h3> Payscale Dataset </h3>

Payscale datasets are cleaned and concatenated. To maintain consistency and avoid duplicates in the Payscale dataset, which lists multiple names for the same university, we standardized the names by consolidating the different variations into a single, specific name for each institution.

<h3> Overall Dataset </h3>

We merged the "times" dataframe and "pay" dataframe together to form the overall "df" dataframe. Then, we adjusted the starting and mid-career salaries based on each year's inflation rate comparing to 2024 and stored these values in two new columns "starting salary adj" and "mid-career salary adj."

```python
# inflation rates defined as $1 January of 'year' -> $1 January of 2024
# https://data.bls.gov/cgi-bin/cpicalc.pl?

inflation = {2012:1.36, 2013:1.34, 2014:1.32, 2015:1.32, 2016:1.3, 2017:1.27, 2023:1.03}
def inflate_rate(year_in):
    return inflation[year_in]

# merge on name, year to prevent any possible duplicates
df = pd.merge(times, pay, how='inner', on = ['name', 'year']).set_index('name')

# correct for inflation based on year
df['starting salary adj'] = df['starting salary'] * df['year'].apply(inflate_rate)
df['mid-career salary adj'] = df['mid-career salary'] * df['year'].apply(inflate_rate)

df = df.sort_values(by=['year', 'rank'], ascending = [True, True])
df.head()
```

The above graphs show the distributions of starting salaries and mid-career salaries adjusted for inflation over the years. There is a general positive increase trend as time goes on. Most of the Pre-COVID (2012-2018) observations keep 75% of their salaries between 60,000-75000 USD, with about a 10,000 USD increase to the Post-COVID year 2023. A similar pattern fits the mid-career salaries: Pre-COVID observations are mostly within 110,000-140,000 USD, with the Post-COVID year 2023 showing a 10,000 USD increase.

```python
iqr = {}
def year2outlier(year):
    return iqr[year]

def find_outliers(column, df):
    for i in list(range(2012,2018))+list(range(2023,2024)):
        Q1 = df[df['year'] == i][column].quantile(0.25)
        Q3 = df[df['year'] == i][column].quantile(0.75)
        iqr[i] = Q3+1.5*(Q3 - Q1)
    return df[df[column] > df['year'].apply(year2outlier)]

ssOut = find_outliers('starting salary adj', df)
msOut = find_outliers('mid-career salary adj', df)

pd.concat([ssOut,msOut]).groupby(by='name').count().rename(columns={'rank':'count'}).get(['count'])
```

The outliers of the boxplots are shown above. CalTech, MIT, Princeton, and Stanford are known for being very selective and prestigious universities, and seeing as they are outliers to the salary boxplots (in the positive direction), it's not surprising that their alumni earn comparatively high salaries. While the Colorado School of Mines is ranked in the 100-200 range over the years, many of their alumni enter lucrative businesses related to fossil fuels, explaining the comparatively high salary. We decide not to remove any outliers at this point in time as there are no errors.

For lower whisker, median, and upper whisker boxplot labels:

```python
import matplotlib.patheffects as path_effects

def add_median_labels(ax: plt.Axes, fmt: str = ".1f") -> None:
    lines = ax.get_lines()
    boxes = [c for c in ax.get_children() if "Patch" in str(c)]
    for i in range(5):
        if ((i == 2) | (i == 3) | (i == 4)):
            if not boxes:  # seaborn v0.13 => fill=False => no patches => +1 line
                boxes = [c for c in ax.get_lines() if len(c.get_xdata()) == 5]
                #start += 1
            lines_per_box = len(lines) // len(boxes)
            for median in lines[i::lines_per_box]:
                x, y = (data.mean() for data in median.get_data())
                # choose value depending on horizontal or vertical plot orientation
                value = x if len(set(median.get_xdata())) == 1 else y
                text = ax.text(x, y, f'{value:{fmt}}', ha='center', va='center',
                            fontweight='bold', color='white')
                # create median-colored border around white text for contrast
                text.set_path_effects([
                    path_effects.Stroke(linewidth=3, foreground=median.get_color()),
                    path_effects.Normal(),
                ])
```

<h2>Statistical Analysis</h2>

Based on the boxplots above and the gaps between the top 30% and bottom 30% schools graduates' salaries, we perform t-tests below to compare the means of top 30% universities and bottom 30% universities starting and mid-career salaries to determine if they are statistically significantly different from each other across years 2012-2017 and 2023.

```python
years = [2012, 2013, 2014, 2015, 2016, 2017, 2023]
from scipy.stats import ttest_ind
start_salary_results = []
mid_career_salary_results = []
salary_time = ['starting salary adj','mid-career salary adj']
for y in years:
    for salary in salary_time:
        amount = int(len(df[df.year == y]) * proportion)
        max_idx = df[df.year == y].shape[0]
        if salary == 'starting salary adj':
            top_30 = df[df.year == y].sort_values(by='rank', ascending=True).iloc[:amount]['starting salary adj']  
            bot_30 = df[df.year == y].sort_values(by='rank', ascending=False).iloc[:amount]['starting salary adj']
            t_stat, p_val = ttest_ind(top_30, bot_30, nan_policy='omit')
            start_salary_results.append({'Year': y, 'T-Statistic': t_stat, 'P-Value': p_val})
        else:
            top_30 = df[df.year == y].sort_values(by='rank', ascending=True).iloc[:amount]['mid-career salary adj']  
            bot_30 = df[df.year == y].sort_values(by='rank', ascending=False).iloc[:amount]['mid-career salary adj']
            t_stat, p_val = ttest_ind(top_30, bot_30, nan_policy='omit')
            mid_career_salary_results.append({'Year': y, 'T-Statistic': t_stat, 'P-Value': p_val})
start_salary_results_df = pd.DataFrame(start_salary_results, index=None)
mid_career_salary_results_df = pd.DataFrame(mid_career_salary_results,index=None)

print("Starting Salary T-Test Results")
print(start_salary_results_df.to_string(index=False))
print("\n" + "-" * 33 + "\n")
print("Mid-Career Salary T-Test Results")
print(mid_career_salary_results_df.to_string(index=False))
```

We selected the top 30% and bottom 30% of universities to ensure adequate sample sizes for conducting the t-test. Our null hypothesis is that there is no significant differences in the starting and mid-career salaries of graduates from the top 30% of US universities compared to those from the bottom 30%, across each examined year (2012-2017, and 2023). The results of the t-tests reveal that the p-values for both starting and mid-career salaries during these years are well below the 0.05 threshold, indicating statistical significance. Notably, the p-values for mid-career salaries are exceptionally low, all falling beneath 1e-6. Based on these findings, we reject the null hypothesis, affirming significant statistical disparities between the salaries of graduates from the top 30% and bottom 30% of US universities. These p-values align with our earlier boxplot comparisons of starting and mid-career salaries from the top 30% versus bottom 30% universities that there are significant differences between them.

# Ethics & Privacy

Payscale data: Since the payscale dataset is completely collected through online self-reported survey, multiple concerns can arise from this data colllection method. Firstly, since the survey is online, it is accessible and can be completed by anyone who visits the Payscale website. Anyone who successfully complete the payscale employee survey can record their self-reported data into this dataset. Therefore, there is potential possibility that people may fake the results of their data in order to get a certain university to have high income rate. Secondly, self-reported data can always be biased and create inconsistencies, even unintentionally. For example, if the people taking survey just recently gets promoted, they may report either the salary before promotion or after promotion, creating inconsistencies between each individual data. Thirdly, the dataset itself may be a biased reflection of the true average salaries, because people with higher incomes usually are more active in searching and participating in a employment survey, thus the dataset's reported salary may be higher than the true salary. Since the nature of the collection method of our dataset will lead to inconsistencies and bias, we need to note these potential concerns to the readers while drawing our conclusions, even if Payscale claims to have their own proprietary data-correction algorithm and analysts that verify their results. We should always incorporate these concerns into our conclusions and avoid drawing the conclusion firmly.  
  
Times data: The dataset we choose to base our university rankings on is the Times Higher Education ranking. While it is generally considered to be a reliable and popular rank, there are still potential problems associated with it. The Times ranking, according to its ranking methodology, focuses heavily on research and citation, which means the rank is mostly based on the academic achivements and reputations of the universities. While academic aspect is an important component for a university, it neglects some other aspects that are also important such as the career aspect and fails to take into account the professional prospect of the graduates. Therefore, the rank methodology may create some bias and inaccuracies. This problem can be solved by basing our project on multiple reliable university ranking instead of only one. This way we can minimize the bias arising from a single ranking methodology.
