## Commerical Fertilizer Annual Report Data
### Web scraping for the data
Since the fertilizer grade distribution dataset is not alrealy avilable for us in csv format, we decided to scrape the data from the data source. We will use `BeautifulSoup` for our web scraping on the website: https://otscweb.tamu.edu/Reports/Annual.aspx

```python
def parser(year: str):

    # Get the HTML Comment based on the year
    html_content = requests.get(url = "https://otscweb.tamu.edu/Reports/Annual/FertByCountyAllQtr.aspx?RegistrationType=2&FiscalYear={year}".format(year=year)).text
    soup = BeautifulSoup(html_content, 'html.parser')

    # Find all the h4 tags and then for each tag, subsequent div tag containing the table.
    h4_tag = list(soup.findAll('h4'))
    data = []
    for tag in h4_tag:
        table = tag.find_next_sibling('div').find('table')

        # Find the row with the 'Total' cell.
        for row in table.find_all('tr'):
            cells = row.find_all('td')
            if cells and 'Total' == cells[0].text:
                # Extract the total values for nitrogen, phosphate, and potash.
                nitrogen = cells[1].text
                phosphate = cells[2].text
                potash = cells[3].text
                
                data.append({
                    'year': year,
                    'county': str(tag).replace("<h4>",'').replace("</h4>", '').strip(),
                    'Nitrogen': nitrogen,
                    'Phosphate': phosphate,
                    'Potash': potash, 
                })

    # Create a pandas DataFrame.
    df = pd.DataFrame(data)
    return df
```

```python
def parse_and_return_df(year):
    return parser(str(year))

if not os.path.exists('texas-fertilizer-data.csv'):
    # Initialize an empty list to store future results
    futures = []

    with ThreadPoolExecutor(max_workers=10) as executor:
        for i in range(2002, 2024):
            futures.append(executor.submit(parse_and_return_df, i))

    # Collect results as they are completed
    data_frames = [future.result() for future in futures]

    # Concatenate all the DataFrames in the list
    df = pd.concat(data_frames, ignore_index=True)
    df.to_csv('texas-fertilizer-data.csv', index=False)
```

Next, we will perform mean imputation, conditional on county and year. 

```python
'''
helper function that takes in a series, 
and returns a imputed series with the mean values of the observed data 
'''
def mean_impute(s):
    return s.fillna(s.mean())

# mean imputation for pH, oxygen, temperature, and conductivity
water['pH'] = water.groupby(['County', 'Year'])['pH'].transform(mean_impute)
water['Oxygen (mg/L)'] = water.groupby(['County', 'Year'])['Oxygen (mg/L)'].transform(mean_impute)
water['Temperature (C)'] = water.groupby(['County', 'Year'])['Temperature (C)'].transform(mean_impute)
water['Conductivity (µs/cm)'] = water.groupby(['County', 'Year'])['Conductivity (µs/cm)'].transform(mean_impute)
```

### Pearson Correlation Test

In this section we will be calculating the pearson correlation coefficient. For those who are not aware, this coefficient is a statistical measure that quantifies the linear relationship between two continuous variables. 

The coefficient ranges from -1 to +1.
- A value of +1 indicates a perfect positive linear relationship, meaning that as one variable increases, the other variable increases proportionally.
- A value of -1 indicates a perfect negative linear relationship, meaning that as one variable increases, the other variable decreases proportionally.
- A value of 0 indicates no linear relationship between the variables.

Further, The absolute value of the coefficient represents the strength of the linear relationship between the variables.
- Values closer to +1 or -1 indicate a strong linear relationship.
- Values closer to 0 indicate a weak or no linear relationship.

We are going to calculate this coefficient for all possible variables in our two datasets. For doing this, we are doing to be using `.corr()` function with the method being `pearson`. Further we will plot these coefficients using a heatmap (plotted using sns)

Now let us try to visualize a heatmap of correlations between the different variables from both the datasets.

```python
#Create a merged dataset by county with all relevant features
merged_data = pd.merge(fertilizer_avg, oxygen_avg, on='County', how='inner')
merged_data = pd.merge(merged_data, avg_pH.reset_index(), on='County', how='inner')
merged_data = pd.merge(merged_data, avg_temperature.reset_index(), on='County', how='inner')
merged_data = pd.merge(merged_data, avg_conductivity.reset_index(), on='County', how='inner')
merged_data = pd.merge(merged_data, avg_nitrogen.reset_index(), on='County', how='inner')
merged_data = pd.merge(merged_data, avg_potash.reset_index(), on='County', how='inner')
merged_data = pd.merge(merged_data, avg_phosphate.reset_index(), on='County', how='inner')
merged_data.columns = ['county', 'total_fertilizer', 'oxygen', 'pH', 'temperature', 'conductivity', 'Nitrogen', 'Potash', 'Phosphate']
merged_data
```

```python
# Calculate the Pearson correlation matrix
corr_matrix = merged_data.drop(columns=['county']).corr(method='pearson')

# Create a heatmap of the correlation matrix
sns.heatmap(corr_matrix, annot=True, vmin=-1, vmax=1, center=0, square=True, cmap='RdBu')

plt.title('Correlation of All Features')
plt.show()
```

As we see from the heat map above, there is no value that is close to -1 or 1 except for those variables that belong to the same dataset. However, we see some values that belong to the two different datasets, that are weak correlations.

As described above, a coefficient calue closer to 0 implies weak or no correlation at all. From this information, we can say that the a pearson correlation coefficient of greater than -0.25 (closer to 0) will represent a weaker negative correlation that is more likely to tell us there is no correlation at all. So let us try to find those variables that have a correlation coefficient of less than -0.25.

```python
# Create a boolean mask for correlation coefficients less than -0.25
mask = corr_matrix < -0.25

# Get the row and column labels of the cells that satisfy the condition
variable_pairs = [(corr_matrix.index[i], corr_matrix.columns[j]) for i, j in zip(*np.where(mask))]

# Create a set to store unique variable pairs
unique_pairs = set()

# Print the unique variable pairs and their correlation coefficients
for pair in variable_pairs:
    if pair[0] != pair[1]:  # Exclude diagonal elements (self-correlation)
        # Check if the reversed pair already exists in the set
        if (pair[1], pair[0]) not in unique_pairs:
            corr_coef = corr_matrix.loc[pair[0], pair[1]]
            print(f"Variables: {pair[0]} and {pair[1]}, Correlation Coefficient: {corr_coef:.2f}")
            unique_pairs.add(pair)
```

| Phosphate     |  Conductivity    | -0.28                   | Negative            | Weak                     | Higher Phosphate is weakly associated with lower Conductivity levels.                                 |

- The strength of correlation is based on the following general guidelines:
  - 0 to ±0.3: Weak correlation
  - ±0.3 to ±0.7: Moderate correlation
  - ±0.7 to ±1.0: Strong correlation

## Analysis: Linear Regression

The Pearson correlation coefficient (r) that we calculated above measures the strength and direction of the linear relationship between two variables X and Y.

The goal of linear regression, on the other hand, is to estimate the values of a and b that minimize the sum of squared differences between the observed data points and the fitted line. This line can then be used to predict the value of Y for any given value of X.

The two techniques are complementary. The Pearson coefficient indicates if a linear model is appropriate, while linear regression provides the actual model equation if linearity holds. A high absolute Pearson r value supports using linear regression, while a value close to 0 suggests a linear model may not be suitable. This implies that the variables that we ignored above (whose coefficients were not high enough, or in our case, low enough) will not be fit for linear regression. Hence ignoring them was a good decision.

The p-value in linear regression serves to assess the statistical significance of the relationship between the independent variable(s) X and the dependent variable Y.

Specifically, the p-value helps to determine if the estimated coefficients (slopes) in the linear regression model are statistically different from zero. A low p-value indicates that the coefficient is unlikely to be zero, suggesting that changes in the independent variable(s) are associated with changes in the dependent variable.

Now we are going to run Linear regression on the variables that we found weak to moderate negative correlation for above. 

For the purpose of this regression, we will keep our alpha value of **0.05**

```python
# Function to run linear regression
def linear_regression(X, y):
    outcome, predictors = patsy.dmatrices(f'{y} ~ {X}', merged_data)
    model = sm.OLS(outcome, predictors)
    return model.fit()
```

However, when we plotted the water variables over the same period of time, we saw that they generally just remained the same. This gave us the first indication that this might go against our hypothesis.

As we progressed, we calculated Pearson coefficients for each of the independent variables (fertilizer variables) with each of the dependent variables (water quality variables). The coefficient values for **pH against phosphate**, **pH against potash usage**, and **oxygen against phosphate usage** suggested a very weak negative correlation. As mentioned above, running linear regression on them would have yielded no results, as the Pearson coefficient and linear regression go hand in hand.

For **oxygen against potash** and **conductivity against nitrogen**, the correlation values were enough to perform linear regression. However, performing linear regression on these variables with a coefficient value of 0.05 yielded poor-fitting models and p-values that were not low enough to prove that there is a correlation between them. We also performed linear regression for other variables like **oxygen vs. total fertilizer use**, **conductivity vs. potash use**, and **conductivity vs. phosphate use**, as their coefficient values were sufficient enough, but they also yielded similar results.

Based on the data we had, the tests we conducted on it (after cleaning and standardizing), and the results we obtained, we do not have enough statistical evidence to reject our null hypothesis, i.e., **there is no correlation between fertilizer usage and key water quality variables in the state of Texas across different counties between 2002 and 2023**.
